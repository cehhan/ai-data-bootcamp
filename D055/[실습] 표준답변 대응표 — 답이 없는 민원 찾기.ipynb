````python
# 실습 준비 ② 환경 확인 — 한글 폰트 · 그림 색 · DEVICE · 문장 인코더 설정 · 버전
import glob
import os
import sys
import warnings

import matplotlib.font_manager as fm
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")

# 한글 폰트 — 설치한 나눔 ttf 를 폰트 목록에 직접 등록합니다.
#  설치는 디스크에 파일만 만듭니다. 이미 불러온 폰트 목록에는 새 파일이 들어 있지 않으므로 addfont 로
#  등록해야 이름이 보입니다(등록하지 않으면 DejaVu Sans 로 바뀌어 한글이 전부 □ 로 나옵니다).
for _p in glob.glob("/usr/share/fonts/truetype/nanum/Nanum*.ttf"):
    try: fm.fontManager.addfont(_p)
    except Exception: pass
try: fm._get_font.cache_clear()
except Exception: pass
installed = {f.name for f in fm.fontManager.ttflist}
FONT = next((n for n in ["Pretendard", "Apple SD Gothic Neo", "NanumGothic"] if n in installed), "DejaVu Sans")
plt.rcParams["font.family"] = FONT
plt.rcParams["axes.unicode_minus"] = FONT in {"Pretendard", "DejaVu Sans"}   # 나머지 폰트에는 U+2212 가 없습니다
%config InlineBackend.print_figure_kwargs = {"bbox_inches": None}   # 그림마다 폭이 깎이지 않게
plt.rcParams["figure.dpi"] = plt.rcParams["savefig.dpi"] = 150       # 폭 FIG_W x 150dpi = 1500px 고정
plt.rcParams["figure.autolayout"] = True
plt.rcParams.update({"font.size": 11, "axes.titlesize": 15, "axes.labelsize": 12,
                     "legend.fontsize": 11, "figure.titlesize": 17, "font.weight": "regular",
                     "text.color": "black", "axes.labelcolor": "black", "axes.titlecolor": "black",
                     "axes.edgecolor": "#757575", "xtick.color": "#757575", "ytick.color": "#757575",
                     "xtick.labelcolor": "black", "ytick.labelcolor": "black",
                     "axes.spines.top": False, "axes.spines.right": False, "axes.grid": False})

# 이 실습의 그림 전체가 함께 쓰는 색 — 기준 = 남색, 강조 = 하늘색, 문맥 = 회색
COL = {"남색": "#051C2C", "파랑": "#2251FF", "하늘색": "#00A9F4",
       "회색": "#757575", "연회색": "#B3B3B3"}
FIG_W = 10.0     # 캔버스 폭은 이 실습의 그림 모두 이 값으로 고정하고 높이만 그림마다 조절합니다

# DEVICE — 이 실습은 문장 인코더를 실행하므로 GPU 런타임을 권합니다
import torch
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
if DEVICE == "cuda":
    _gpu = torch.cuda.get_device_properties(0)
    print(f"GPU : {_gpu.name} · 메모리 {_gpu.total_memory / 1024 ** 3:.1f}GB")
else:
    print("GPU 가 없습니다 — 작은 문장 인코더와 항공 민원 4,000건으로 같은 코드를 실행합니다")

# 문장 인코더 설정 — DEVICE 에 따라 정해집니다. 버전(revision)을 고정해 교안의 수를 계산한 가중치를 불러옵니다
#  GPU : nlpai-lab/KURE-v1 (MIT) — 접두어 없음, 항공 민원 12,232건 전체, 리뷰 3,000건
#  CPU : intfloat/multilingual-e5-small (MIT) — 접두어 「query: 」, 항공 민원 4,000건(시드 0), 리뷰 1,000건
if DEVICE == "cuda":
    MODEL_NAME, MODEL_REV, PREFIX = "nlpai-lab/KURE-v1", "8b418a58414668e75532ed045c22d9ca018ae2b2", ""
    N_AIR, N_REV = None, 3000
else:
    MODEL_NAME, MODEL_REV, PREFIX = "intfloat/multilingual-e5-small", "614241f622f53c4eeff9890bdc4f31cfecc418b3", "query: "
    N_AIR, N_REV = 4000, 1000
print(f"DEVICE : {DEVICE} · 문장 인코더 : {MODEL_NAME} · "
      + ("항공 민원 12,232건 전체" if N_AIR is None else f"항공 민원 {N_AIR:,}건(시드 0)")
      + f" · 리뷰 {N_REV:,}건 | 사용 폰트 : {FONT}")

import kiwipiepy
import sentence_transformers
import transformers
print(f"kiwipiepy {kiwipiepy.__version__} · sentence-transformers {sentence_transformers.__version__}"
      f" · transformers {transformers.__version__} · torch {torch.__version__.split('+')[0]}")
if kiwipiepy.__version__ != "0.23.1":
    print("참고 — 교안의 수는 kiwipiepy 0.23.1 로 계산했습니다. 다른 버전에서는 명사 목록이 조금 다를 수 있습니다.")
````

````python
#@title [채점] 이 셀을 먼저 실행하세요 (한 번이면 됩니다)
#   미션의 「PROMPT 셀」을 실행하면 그 셀의 출력과 그림을 받아 실행 결과에 채점 결과를 찍습니다.
#   통과하면 그 자리에서 완성본 풀이가 열리고, 아니면 항목별 힌트와 튜터에 보낼 질문이 나옵니다.
#   읽지 않고 실행만 해도 됩니다 — 기대값과 풀이는 이 셀에만 있고 「PROMPT 셀」에는 없습니다.
_G0 = dict(globals())                     # 이 셀이 만들기 전의 이름 — 맨 아래에서 이 셀이 만든 이름을 가려냅니다
_CANON = "0.4.3"                          # 모든 실습이 함께 쓰는 채점 코드의 판 번호입니다

import base64
import glob
import io
import json
import re
import shutil
import subprocess
import sys

import matplotlib.font_manager as fm      # 아래 폰트 자급이 쓴다
import matplotlib.pyplot as plt

#  채점 코드는 아래 별칭만 씁니다 — 학생 셀의 `from numpy import *`(min·max·any·round·sum·abs·bool)와
#  `from pandas import *`(io) 가 같은 이름을 덮어도, 학생이 re·json 같은 이름에 값을 넣어도 채점은 그대로 돕니다.
import base64 as _b64
import builtins as _bi
import io as _io
import json as _json
import re as _re
import sys as _sys
import matplotlib.pyplot as _plt


def _font_bootstrap():
    """한글 폰트 자급 — 맨 위 설치·폰트 셀을 건너뛰어도 그래프의 한글이 깨지지 않게.

    LMS 는 이 셀만 자동 실행하므로 폰트 준비도 여기서 한 번 더 한다. 설치·폰트 셀과 같은
    규칙이라 둘 다 돌아도 결과가 같고, apt-get 이 없는 자리(로컬)에서는 곧바로 지나간다.
    설치한 ttf 는 `addfont` 로 폰트 목록에 직접 넣는다 — 넣지 않으면 목록이 옛 상태라
    Colab 에서도 한글이 □ 로 깨진다.
    """
    try:
        cands = ["Pretendard", "Apple SD Gothic Neo", "NanumGothic"]
        installed = {f.name for f in fm.fontManager.ttflist}
        if not _bi.any(n in installed for n in cands) and shutil.which("apt-get"):
            subprocess.run(["apt-get", "-y", "-qq", "install", "fonts-nanum"],
                           check=False, timeout=120)
        # 설치는 파일만 만든다 — 이미 떠 있는 폰트 목록에 addfont 로 넣어야 이름이 보인다.
        # (설치를 건너뛴 자리도 나눔 ttf 가 이미 있으면 이 줄들만으로 잡힌다.)
        for _p in glob.glob("/usr/share/fonts/truetype/nanum/Nanum*.ttf"):
            try: fm.fontManager.addfont(_p)
            except Exception: pass
        try: fm._get_font.cache_clear()
        except Exception: pass
        installed = {f.name for f in fm.fontManager.ttflist}
        _font = next((n for n in cands if n in installed), "DejaVu Sans")
        _plt.rcParams["font.family"] = _font
        _plt.rcParams["axes.unicode_minus"] = _font in {"Pretendard", "DejaVu Sans"}
    except Exception:
        print("한글 폰트를 못 찾아 그래프 글자가 깨질 수 있습니다 —"
              " 맨 위 설치·폰트 셀을 실행하세요")


_font_bootstrap()


def _dec_json(s):                          # base64(JSON) → 값. 리스트는 원래 튜플이라 되돌립니다
    def _t(v):
        return ({k: _t(x) for k, x in v.items()} if isinstance(v, dict)
                else tuple(_t(x) for x in v) if isinstance(v, list) else v)
    return _t(_json.loads(_b64.b64decode(s).decode("utf-8")))


#  아래 두 표지 줄 사이만 실습마다 다릅니다 — _ANS · _REVEAL · _HINT (· _CAUSE) 블롭, 튜터 질문 틀(_ASK · _ASK_HEAD ·
#  _ASK_FORM), 빈 PROMPT 에 잠그는 미션 `_blocks(key, reason)`, 권 설정 `_BOOK`(minus · skip · spec · pre · checks ·
#  error_hint), 그 실습에만 있는 함수. 표지 줄 밖은 모든 실습이 같은 채점 코드입니다.
# ━━ 이 실습의 채점 기준과 설정 — 이 줄부터 「채점 기준 끝」 줄까지만 실습마다 다릅니다 ━━
# 채점 기준 — base64(JSON), 통과 판정에만 쓴다. DEVICE 에 따라 by_device 의 두 벌 가운데 하나를 덮어씁니다
#  cuda : KURE-v1 · 항공 민원 12,232건 전체    cpu : e5-small · 항공 민원 4,000건(시드 0)
_ANS = _dec_json(
               "eyJtMSI6IHsiY3VzdG9tIjogIm0xIiwgInBsb3QiOiB7Iuy2leudvOuyqCI6IHRydWV9LCAiYnlfZGV2aWNlIjogeyJjdWRhIjog"
               "eyJ2YWx1ZXMiOiB7IuydtOybgyAyMOqxtCDspJEg44CM7ZmY67aI44CNIOyXhuuKlCDrr7zsm5AiOiBbMSwgMSwgIuqxtCJdLCAi"
               "66y47J6lIOyehOuyoOuUqSDquLjsnbQiOiBbMC45OTksIDEuMDAxXX0sICJ0cmFwcyI6IHsi7J207JuDIDIw6rG0IOykkSDjgIzt"
               "mZjrtojjgI0g7JeG64qUIOuvvOybkCI6IFtbMjAsIDIwLCAi7J20IOqwkuydgCDsnbTsm4MgMjDqsbTsnYQg7L2U7IKs7J247J20"
               "IOyekeydgCDsiJzshJzroZwg6rOo656Q7J2EIOuVjCDrgpjsmKTripQg6rCS7J2066+A66GcIGBucC5hcmdzb3J0KC1jb3MpWzpU"
               "T1BfS11gIOyymOufvCDtgbAg7Iic7ISc66GcIOqzqOuekOuKlOyngCDrtIXri4jri6QiXSwgWzE5LCAxOSwgIuydtCDqsJLsnYAg"
               "44CM7ZmY67aI44CN7J20IOyeiOuKlCDrr7zsm5DsnYQg7IWM7J2EIOuVjCDrgpjsmKTripQg6rCS7J2066+A66GcIGDrqoXsgqxg"
               "IOuqqeuhneyXkCBgS0VZV09SRGAg6rCAIOyXhuuKlCDrr7zsm5DsnYQg7IWM64qU7KeAIOu0heuLiOuLpCJdLCBbMywgMywgIuyd"
               "tCDqsJLsnYAg7KCc66qp7J2EIOqzteuwseycvOuhnCDsnpDrpbgg7Ja07KCI7JeQ7IScIOOAjO2ZmOu2iOOAjeydhCDssL7slZjs"
               "nYQg65WMIOuCmOyYpOuKlCDqsJLsnbTrr4DroZwgYOuqheyCrGAg7Je07JeQ7IScIOywvuyVmOuKlOyngCDrtIXri4jri6QiXV19"
               "fSwgImNwdSI6IHsidmFsdWVzIjogeyLsnbTsm4MgMjDqsbQg7KSRIOOAjO2ZmOu2iOOAjSDsl4bripQg66+87JuQIjogWzIsIDIs"
               "ICLqsbQiXSwgIuusuOyepSDsnoTrsqDrlKkg6ri47J20IjogWzAuOTk5LCAxLjAwMV19LCAidHJhcHMiOiB7IuydtOybgyAyMOqx"
               "tCDspJEg44CM7ZmY67aI44CNIOyXhuuKlCDrr7zsm5AiOiBbWzIwLCAyMCwgIuydtCDqsJLsnYAg7J207JuDIDIw6rG07J2EIOy9"
               "lOyCrOyduOydtCDsnpHsnYAg7Iic7ISc66GcIOqzqOuekOydhCDrlYwg64KY7Jik64qUIOqwkuydtOuvgOuhnCBgbnAuYXJnc29y"
               "dCgtY29zKVs6VE9QX0tdYCDsspjrn7wg7YGwIOyInOyEnOuhnCDqs6jrnpDripTsp4Ag67SF64uI64ukIl0sIFsxOCwgMTgsICLs"
               "nbQg6rCS7J2AIOOAjO2ZmOu2iOOAjeydtCDsnojripQg66+87JuQ7J2EIOyFjOydhCDrlYwg64KY7Jik64qUIOqwkuydtOuvgOuh"
               "nCBg66qF7IKsYCDrqqnroZ3sl5AgYEtFWVdPUkRgIOqwgCDsl4bripQg66+87JuQ7J2EIOyFjOuKlOyngCDrtIXri4jri6QiXSwg"
               "WzgsIDgsICLsnbQg6rCS7J2AIOygnOuqqeydhCDqs7XrsLHsnLzroZwg7J6Q66W4IOyWtOygiOyXkOyEnCDjgIztmZjrtojjgI3s"
               "nYQg7LC+7JWY7J2EIOuVjCDrgpjsmKTripQg6rCS7J2066+A66GcIGDrqoXsgqxgIOyXtOyXkOyEnCDssL7slZjripTsp4Ag67SF"
               "64uI64ukIl1dfX19fSwgIm0yIjogeyJyZWxzIjogW3sia2luZCI6ICJyYXRpb19taW4iLCAiYSI6ICJlNS1zbWFsbCDsnoTqs4Tq"
               "sJIgMC45MCDsjI0iLCAiYiI6ICJLVVJFLXYxIOyehOqzhOqwkiAwLjkwIOyMjSIsICJtaW4iOiAxLjMsICJuYW1lIjogIuqwmeyd"
               "gCAwLjkwIOydmCDsjI0g7IiYIOu5hCIsICJoaW50IjogInJlbF9yYXRpbyIsICJydWxlIjogImU1LXNtYWxsIMO3IEtVUkUtdjEg"
               "4omlIDEuMyJ9LCB7ImtpbmQiOiAiZ3QiLCAiYSI6ICLrs4TsoJAg6rCI66awIOu5hOycqCBlNS1zbWFsbCIsICJiIjogIuuzhOyg"
               "kCDqsIjrprAg67mE7JyoIEtVUkUtdjEiLCAibmFtZSI6ICLrs4TsoJAg6rCI66awIOu5hOycqCDruYTqtZAiLCAiaGludCI6ICJy"
               "ZWxfc3BsaXQiLCAicnVsZSI6ICJlNS1zbWFsbCA+IEtVUkUtdjEifSwgeyJraW5kIjogIm5lYXIiLCAiYSI6ICLri6Tsi5wg7KCV"
               "7ZWcIOyehOqzhOqwkuyXkOyEnCBlNS1zbWFsbCDsjI0iLCAiYiI6ICJLVVJFLXYxIOyehOqzhOqwkiAwLjkwIOyMjSIsICJ0b2wi"
               "OiAwLjEsICJuYW1lIjogIuuLpOyLnCDsoJXtlZwg7J6E6rOE6rCS7J2YIOyMjSDsiJgiLCAiaGludCI6ICJyZWxfbmVhciIsICJy"
               "dWxlIjogIktVUkUtdjEgMC45MCDsjI0g7IiY7J2YIMKxMTAlIOyViCJ9LCB7ImtpbmQiOiAicGN0X29mX24iLCAiYSI6ICLqsbDs"
               "nZgg6rCZ7J2AIOq4gOydtCDsnojripQg66as67ewIiwgIm5fbmFtZSI6ICJOX1JFViIsICJuYW1lIjogIuqxsOydmCDqsJnsnYAg"
               "6riAIOu5hOycqOydmCDrtoTrqqgiLCAiaGludCI6ICJyZWxfcGN0IiwgInJ1bGUiOiAi6rG07IiYIMO3IOumrOu3sCDsiJggw5cg"
               "MTAwID0g67mE7JyoIn1dLCAiZm9yYmlkIjogWyLtjJDri6ggOiIsICLtjJDri6g6IiwgIuuzhOygkCDtj4nqt6DsnYQg66+/7J2E"
               "IOyImCDsl4bri6QiXSwgInBsb3QiOiB7Iuy2leudvOuyqCI6IHRydWV9LCAiYnlfZGV2aWNlIjogeyJjdWRhIjogeyJ2YWx1ZXMi"
               "OiB7IktVUkUtdjEg7J6E6rOE6rCSIDAuOTAg7IyNIjogWzExMCwgMTEwLCAi7IyNIl0sICJlNS1zbWFsbCDsnoTqs4TqsJIgMC45"
               "MCDsjI0iOiBbMTc1NDEwLCAxNzU3NzAsICLsjI0iXSwgIuuzhOygkCDqsIjrprAg67mE7JyoIEtVUkUtdjEiOiBbNy43LCA4Ljcs"
               "ICIlIl0sICLrs4TsoJAg6rCI66awIOu5hOycqCBlNS1zbWFsbCI6IFsyMi43LCAyMy43LCAiJSJdLCAiZTUtc21hbGwg64uk7Iuc"
               "IOygle2VnCDsnoTqs4TqsJIiOiBbMC45NiwgMC45N10sICLri6Tsi5wg7KCV7ZWcIOyehOqzhOqwkuyXkOyEnCBlNS1zbWFsbCDs"
               "jI0iOiBbOTksIDEyMSwgIuyMjSJdLCAi6rGw7J2YIOqwmeydgCDquIDsnbQg7J6I64qUIOumrOu3sCI6IFsyLjc1LCAzLjI1LCAi"
               "JSJdfSwgInRyYXBzIjogeyJLVVJFLXYxIOyehOqzhOqwkiAwLjkwIOyMjSI6IFtbMjIwLCAyMjAsICLsnbQg6rCS7J2AIOq4sOyk"
               "gOydmCDrkZAg67Cw66GcLCDrjIDqsIHshKDrp4wg67qAIOy9lOyCrOyduCDtlonroKwg7KCE7LK07JeQ7IScIOqwmeydgCDsjI3s"
               "nYQgKGksIGopIOyZgCAoaiwgaSkg66GcIOuRkCDrsogg7IWM7J2EIOuVjCDrgpjsmKTripQg6rCS7J2066+A66GcIOychCDsgrzq"
               "sIEoaSA8IGop66eMIOyFjOuKlOyngCDrtIXri4jri6QiXV0sICJlNS1zbWFsbCDsnoTqs4TqsJIgMC45MCDsjI0iOiBbWzM1MDgy"
               "MCwgMzUxNTQwLCAi7J20IOqwkuydgCDquLDspIDsnZgg65GQIOuwsOuhnCwg64yA6rCB7ISg66eMIOu6gCDsvZTsgqzsnbgg7ZaJ"
               "66CsIOyghOyytOyXkOyEnCDqsJnsnYAg7IyN7J2EIChpLCBqKSDsmYAgKGosIGkpIOuhnCDrkZAg67KIIOyFjOydhCDrlYwg64KY"
               "7Jik64qUIOqwkuydtOuvgOuhnCDsnIQg7IK86rCBKGkgPCBqKeunjCDshYzripTsp4Ag67SF64uI64ukIl0sIFszNTM4MjAsIDM1"
               "NDU0MCwgIuydtCDqsJLsnYAg7L2U7IKs7J24IO2WieugrCDsoITssrTrpbwg7IS47Ja0IOyekOq4sCDsnpDsi6Dqs7zsnZgg7IyN"
               "6rO8IOqwmeydgCDsjI3snZgg7KSR67O17J20IOuqqOuRkCDrk6TslrTqsJTsnYQg65WMIOuCmOyYpOuKlCDqsJLsnbTrr4DroZwg"
               "7JyEIOyCvOqwgShpIDwgainrp4wg7IWM64qU7KeAIOu0heuLiOuLpCJdXSwgIuuLpOyLnCDsoJXtlZwg7J6E6rOE6rCS7JeQ7ISc"
               "IGU1LXNtYWxsIOyMjSI6IFtbMjIwLCAyMjAsICLsnbQg6rCS7J2AIOq4sOykgOydmCDrkZAg67Cw66GcLCDrjIDqsIHshKDrp4wg"
               "67qAIOy9lOyCrOyduCDtlonroKwg7KCE7LK07JeQ7IScIOqwmeydgCDsjI3snYQgKGksIGopIOyZgCAoaiwgaSkg66GcIOuRkCDr"
               "sogg7IWM7J2EIOuVjCDrgpjsmKTripQg6rCS7J2066+A66GcIOychCDsgrzqsIEoaSA8IGop66eMIOyFjOuKlOyngCDrtIXri4jr"
               "i6QiXV0sICLrs4TsoJAg6rCI66awIOu5hOycqCBLVVJFLXYxIjogW1swLjA1LCAwLjE1LCAi7J20IOqwkuydgCAlIOqwgCDslYTr"
               "i4jrnbwgMCDqs7wgMSDsgqzsnbTsnZgg67mE66GcIOywjeyXiOydhCDrlYwg64KY7Jik64qUIOqwkuydtOuvgOuhnCAxMDAg7J2E"
               "IOqzse2VtCAlIOuhnCDstpzroKXtlanri4jri6QiXSwgWzIyLjcsIDIzLjcsICLsnbQg6rCS7J2AIGU1LXNtYWxsIOyMjeycvOuh"
               "nCDshYzsnYQg65WMIOuCmOyYpOuKlCDqsJLsnbTrr4DroZwg7J6E6rOE6rCSIDAuOTAg7JeQ7IScIOustuyduCBLVVJFLXYxIOyM"
               "jeycvOuhnCDshYzripTsp4Ag67SF64uI64ukIl1dLCAi67OE7KCQIOqwiOumsCDruYTsnKggZTUtc21hbGwiOiBbWzAuMiwgMC4z"
               "LCAi7J20IOqwkuydgCAlIOqwgCDslYTri4jrnbwgMCDqs7wgMSDsgqzsnbTsnZgg67mE66GcIOywjeyXiOydhCDrlYwg64KY7Jik"
               "64qUIOqwkuydtOuvgOuhnCAxMDAg7J2EIOqzse2VtCAlIOuhnCDstpzroKXtlanri4jri6QiXSwgWzkuNSwgMTAuNSwgIuydtCDq"
               "sJLsnYAgMC45MCDsnbQg7JWE64uI6528IOuLpOyLnCDsoJXtlZwg7J6E6rOE6rCS7JeQ7IScIOyehOqzhOqwkiDsnbTsg4Hsnbgg"
               "7IyN7Jy866GcIOyFjOydhCDrlYwg64KY7Jik64qUIOqwkuydtOuvgOuhnCDsnoTqs4TqsJIgMC45MCDsl5DshJwg66y27J24IGU1"
               "LXNtYWxsIOyMjeycvOuhnCDshYzripTsp4Ag67SF64uI64ukIl0sIFs3LjcsIDguNywgIuydtCDqsJLsnYAgS1VSRS12MSDsjI3s"
               "nLzroZwg7IWM7J2EIOuVjCDrgpjsmKTripQg6rCS7J2066+A66GcIOyehOqzhOqwkiAwLjkwIOyXkOyEnCDrrLbsnbggZTUtc21h"
               "bGwg7IyN7Jy866GcIOyFjOuKlOyngCDrtIXri4jri6QiXV0sICLqsbDsnZgg6rCZ7J2AIOq4gOydtCDsnojripQg66as67ewIjog"
               "W1szLjk1LCA0LjA1LCAi7J20IOqwkuydgCBLVVJFLXYxIOydmCAwLjkwIOyMjeyXkCDrgpjsmKgg66as67ew66W8IOyFjOydhCDr"
               "lYwg64KY7Jik64qUIOqwkuydtOuvgOuhnCBlNS1zbWFsbCDsnZgg64uk7IucIOygle2VnCDsnoTqs4TqsJIg7J207IOB7J24IOyM"
               "jeyXkCDrgpjsmKgg66as67ew66W8IOyFjOuKlOyngCDrtIXri4jri6QiXV19fSwgImNwdSI6IHsidmFsdWVzIjogeyJLVVJFLXYx"
               "IOyehOqzhOqwkiAwLjkwIOyMjSI6IFsxMywgMTMsICLsjI0iXSwgImU1LXNtYWxsIOyehOqzhOqwkiAwLjkwIOyMjSI6IFsxOTYz"
               "MCwgMTk2ODAsICLsjI0iXSwgIuuzhOygkCDqsIjrprAg67mE7JyoIEtVUkUtdjEiOiBbMCwgMC41LCAiJSJdLCAi67OE7KCQIOqw"
               "iOumsCDruYTsnKggZTUtc21hbGwiOiBbMjMuOSwgMjQuOSwgIiUiXSwgImU1LXNtYWxsIOuLpOyLnCDsoJXtlZwg7J6E6rOE6rCS"
               "IjogWzAuOTYsIDAuOTddLCAi64uk7IucIOygle2VnCDsnoTqs4TqsJLsl5DshJwgZTUtc21hbGwg7IyNIjogWzEyLCAxNCwgIuyM"
               "jSJdLCAi6rGw7J2YIOqwmeydgCDquIDsnbQg7J6I64qUIOumrOu3sCI6IFsxLjM1LCAxLjQ1LCAiJSJdfSwgInRyYXBzIjogeyJL"
               "VVJFLXYxIOyehOqzhOqwkiAwLjkwIOyMjSI6IFtbMjYsIDI2LCAi7J20IOqwkuydgCDquLDspIDsnZgg65GQIOuwsOuhnCwg64yA"
               "6rCB7ISg66eMIOu6gCDsvZTsgqzsnbgg7ZaJ66CsIOyghOyytOyXkOyEnCDqsJnsnYAg7IyN7J2EIChpLCBqKSDsmYAgKGosIGkp"
               "IOuhnCDrkZAg67KIIOyFjOydhCDrlYwg64KY7Jik64qUIOqwkuydtOuvgOuhnCDsnIQg7IK86rCBKGkgPCBqKeunjCDshYzripTs"
               "p4Ag67SF64uI64ukIl1dLCAiZTUtc21hbGwg7J6E6rOE6rCSIDAuOTAg7IyNIjogW1szOTI2MCwgMzkzNjAsICLsnbQg6rCS7J2A"
               "IOq4sOykgOydmCDrkZAg67Cw66GcLCDrjIDqsIHshKDrp4wg67qAIOy9lOyCrOyduCDtlonroKwg7KCE7LK07JeQ7IScIOqwmeyd"
               "gCDsjI3snYQgKGksIGopIOyZgCAoaiwgaSkg66GcIOuRkCDrsogg7IWM7J2EIOuVjCDrgpjsmKTripQg6rCS7J2066+A66GcIOyc"
               "hCDsgrzqsIEoaSA8IGop66eMIOyFjOuKlOyngCDrtIXri4jri6QiXSwgWzQwMjYwLCA0MDM2MCwgIuydtCDqsJLsnYAg7L2U7IKs"
               "7J24IO2WieugrCDsoITssrTrpbwg7IS47Ja0IOyekOq4sCDsnpDsi6Dqs7zsnZgg7IyN6rO8IOqwmeydgCDsjI3snZgg7KSR67O1"
               "7J20IOuqqOuRkCDrk6TslrTqsJTsnYQg65WMIOuCmOyYpOuKlCDqsJLsnbTrr4DroZwg7JyEIOyCvOqwgShpIDwgainrp4wg7IWM"
               "64qU7KeAIOu0heuLiOuLpCJdXSwgIuuLpOyLnCDsoJXtlZwg7J6E6rOE6rCS7JeQ7IScIGU1LXNtYWxsIOyMjSI6IFtbMjYsIDI2"
               "LCAi7J20IOqwkuydgCDquLDspIDsnZgg65GQIOuwsOuhnCwg64yA6rCB7ISg66eMIOu6gCDsvZTsgqzsnbgg7ZaJ66CsIOyghOyy"
               "tOyXkOyEnCDqsJnsnYAg7IyN7J2EIChpLCBqKSDsmYAgKGosIGkpIOuhnCDrkZAg67KIIOyFjOydhCDrlYwg64KY7Jik64qUIOqw"
               "kuydtOuvgOuhnCDsnIQg7IK86rCBKGkgPCBqKeunjCDshYzripTsp4Ag67SF64uI64ukIl1dLCAi67OE7KCQIOqwiOumsCDruYTs"
               "nKggS1VSRS12MSI6IFtbMjMuOSwgMjQuOSwgIuydtCDqsJLsnYAgZTUtc21hbGwg7IyN7Jy866GcIOyFjOydhCDrlYwg64KY7Jik"
               "64qUIOqwkuydtOuvgOuhnCDsnoTqs4TqsJIgMC45MCDsl5DshJwg66y27J24IEtVUkUtdjEg7IyN7Jy866GcIOyFjOuKlOyngCDr"
               "tIXri4jri6QiXV0sICLrs4TsoJAg6rCI66awIOu5hOycqCBlNS1zbWFsbCI6IFtbMC4yLCAwLjMsICLsnbQg6rCS7J2AICUg6rCA"
               "IOyVhOuLiOudvCAwIOqzvCAxIOyCrOydtOydmCDruYTroZwg7LCN7JeI7J2EIOuVjCDrgpjsmKTripQg6rCS7J2066+A66GcIDEw"
               "MCDsnYQg6rOx7ZW0ICUg66GcIOy2nOugpe2VqeuLiOuLpCJdLCBbMCwgMC4wNSwgIuydtCDqsJLsnYAgMC45MCDsnbQg7JWE64uI"
               "6528IOuLpOyLnCDsoJXtlZwg7J6E6rOE6rCS7JeQ7IScIOyehOqzhOqwkiDsnbTsg4Hsnbgg7IyN7J2064KYIEtVUkUtdjEg7IyN"
               "7Jy866GcIOyFjOydhCDrlYwg64KY7Jik64qUIOqwkuydtOuvgOuhnCDsnoTqs4TqsJIgMC45MCDsl5DshJwg66y27J24IGU1LXNt"
               "YWxsIOyMjeycvOuhnCDshYzripTsp4Ag67SF64uI64ukIl1dLCAi6rGw7J2YIOqwmeydgCDquIDsnbQg7J6I64qUIOumrOu3sCI6"
               "IFtbMi4yNSwgMi4zNSwgIuydtCDqsJLsnYAgS1VSRS12MSDsnZggMC45MCDsjI3sl5Ag64KY7JioIOumrOu3sOulvCDshYzsnYQg"
               "65WMIOuCmOyYpOuKlCDqsJLsnbTrr4DroZwgZTUtc21hbGwg7J2YIOuLpOyLnCDsoJXtlZwg7J6E6rOE6rCSIOydtOyDgeyduCDs"
               "jI3sl5Ag64KY7JioIOumrOu3sOulvCDshYzripTsp4Ag67SF64uI64ukIl1dfX19fX0=")


_REVEAL = {                                # 완성본 풀이 — base64, 통과할 때만 디코드해 인쇄합니다
    "m1": (
               "8J+UkyDrr7jshZggMSDsmYTshLHrs7guIEFJIOy9lOuTnCDtkZzsi5zshKAg7JWE656Y7JeQIOuTpOyWtOyZgOyVvCDtlaAg7L2U"
               "65Oc64qUIOydtCDsoJXrj4TsnoXri4jri6QKCuKWoCDsnbQg7L2U65Oc66W8IOu2gOuluCDtlITroaztlITtirgg7JiICgogICAg"
               "7KSA67mEIOyFgOydmCBBSVIsIFFVRVJZLCBLRVlXT1JELCBUT1BfSywgREVWSUNFLCBNT0RFTF9OQU1FLCBNT0RFTF9SRVYsIFBS"
               "RUZJWCwgcm5kIOuKlCDsnbTrr7gg7J6I7Jy864uIIOuLpOyLnCDrp4zrk6Tsp4Ag66eILgogICAgc2VudGVuY2VfdHJhbnNmb3Jt"
               "ZXJzIOydmCBTZW50ZW5jZVRyYW5zZm9ybWVyIOulvCBNT0RFTF9OQU1FIMK3IHJldmlzaW9uPU1PREVMX1JFViDCtyBkZXZpY2U9"
               "REVWSUNFIOuhnCDrtojrn6zsmYAKICAgIG1heF9zZXFfbGVuZ3RoIOulvCAyNTYg7Jy866GcIOuRkOqzoCwg66qo64247J20IOye"
               "iOuKlCDqs7MobW9kZWwuZGV2aWNlKeydhCDtlZwg7ZaJIOywjeyWtCDspJguCiAgICBBSVIg7J2YIOyCrOqxtOygnOuqqeqzvCBR"
               "VUVSWSDslZ7sl5Ag66qo65GQIFBSRUZJWCDrpbwg65GQ6rOgIG5vcm1hbGl6ZV9lbWJlZGRpbmdzPVRydWUgwrcg67Cw7LmYIDY0"
               "IOuhnCDsnbjsvZTrlKntlZjqs6AsCiAgICDssqsg7KCc66qp7J2YIOusuOyepSDsnoTrsqDrlKkg6ri47J2066W8ICLrrLjsnqUg"
               "7J6E67Kg65SpIOq4uOydtCA9ICjshozsiJgg64S37Ke4IOyekOumrCkiIOuhnCDtlZwg7ZaJIHByaW50IO2VtCDspJguCiAgICBB"
               "SVJfRU1CIEAgcSDroZwg7L2U7IKs7J247J2EIOq1rO2VtCDtgbAg7Iic7ISc66GcIFRPUF9LIOqxtOydhCDqs6DrpbTqs6AsIDHs"
               "nITsmYAgMjDsnIQg7L2U7IKs7J24LCAyMOqxtOydmCDsiJzsnIQgwrcg7L2U7IKs7J24IMK3IOygnOuqqeydhCDssI3slrQg7KSY"
               "LgogICAgMjDqsbTrp4jri6Qg66qF7IKsIOyXtOyXkCBLRVlXT1JEIOqwgCDsnojripTsp4Ag67O06rOgLCDsl4bripQg6rG07IiY"
               "66W8ICLsnbTsm4MgMjDqsbQg7KSRIOOAjO2ZmOu2iOOAjSDsl4bripQg66+87JuQID0gKOygleyImCkg6rG0IiDsnLzroZwg7ZWc"
               "IO2WiSBwcmludCDtlbQg7KSYLgogICAgMjDqsbQg6rCA7Jq0642wIOqwmeydgCDsoJzrqqnsnbQg6rCA7J6lIOunjuydtCDrkJjt"
               "koDsnbTrkJwg7IiY64+EIOywjeyWtCDspJguCiAgICDqt7jrnpjtlITripQg7ZWcIOyepeycvOuhnCDqt7jroKQg7KSYLiDqsIDr"
               "oZwg66eJ64yAKGJhcmgpIDIw6rCcLCAx7JyE6rCAIOunqCDsnIQsIOOAjO2ZmOu2iOOAjSDsnojsnYzqs7wg7JeG7J2M7J2EIOuR"
               "kCDsg4nsnLzroZwg64KY64iE6rOgIOuylOuhgOulvCDri6zqs6AsCiAgICDqsIDroZzstpUg7J2066aEICLsvZTsgqzsnbgiLCDs"
               "hLjroZzstpUg7J2066aEICLsnbTsm4Mg7Iic7JyEIiDroZwg7ZW0IOykmC4KCiAgICBmcm9tIHNlbnRlbmNlX3RyYW5zZm9ybWVy"
               "cyBpbXBvcnQgU2VudGVuY2VUcmFuc2Zvcm1lcgoKICAgIG1vZGVsID0gU2VudGVuY2VUcmFuc2Zvcm1lcihNT0RFTF9OQU1FLCBk"
               "ZXZpY2U9REVWSUNFLCByZXZpc2lvbj1NT0RFTF9SRVYpICAgIyDrsoTsoITsnYQg6rOg7KCV7ZW0IERFVklDRSDroZwg7Jiu6ri0"
               "64ukCiAgICBtb2RlbC5tYXhfc2VxX2xlbmd0aCA9IDI1NgogICAgcHJpbnQoZiLrqqjrjbgg7JyE7LmYIDoge21vZGVsLmRldmlj"
               "ZX0iKQoKICAgIEFJUl9FTUIgPSBtb2RlbC5lbmNvZGUoW1BSRUZJWCArIHQgZm9yIHQgaW4gQUlSWyLsgqzqsbTsoJzrqqkiXV0s"
               "IGJhdGNoX3NpemU9NjQsCiAgICAgICAgICAgICAgICAgICAgICAgICAgIG5vcm1hbGl6ZV9lbWJlZGRpbmdzPVRydWUsIGNvbnZl"
               "cnRfdG9fbnVtcHk9VHJ1ZSwgc2hvd19wcm9ncmVzc19iYXI9RmFsc2UpCiAgICBxID0gbW9kZWwuZW5jb2RlKFtQUkVGSVggKyBR"
               "VUVSWV0sIG5vcm1hbGl6ZV9lbWJlZGRpbmdzPVRydWUsIGNvbnZlcnRfdG9fbnVtcHk9VHJ1ZSlbMF0gICAjIOyniOydmOyXkOuP"
               "hCDqsJnsnYAg7KCR65GQ7Ja0CiAgICBwcmludChmIuuvvOybkCDrsqHthLAgOiB7QUlSX0VNQi5zaGFwZVswXTosfeqxtCDDlyB7"
               "QUlSX0VNQi5zaGFwZVsxXX3ssKjsm5AiKQogICAgcHJpbnQoZiLrrLjsnqUg7J6E67Kg65SpIOq4uOydtCA9IHtybmQobnAubGlu"
               "YWxnLm5vcm0oQUlSX0VNQlswXSksIDQpOi40Zn0iKQoKICAgIGNvcyA9IEFJUl9FTUIgQCBxICAgICAgICAgICAgICAgICAgICAg"
               "ICAgICAgICAgICAgICAgIyDquLjsnbQgMSDrsqHthLDrgbzrpqzsnZgg64K07KCBID0g7L2U7IKs7J24CiAgICB0b3AgPSBucC5h"
               "cmdzb3J0KC1jb3MpWzpUT1BfS10gICAgICAgICAgICAgICAgICAgICAgICMg7L2U7IKs7J247J20IO2BsCDsiJzshJzroZwgMjDq"
               "sbQKICAgIGNvczIwID0gY29zW3RvcF0KICAgIGhhcyA9IEFJUlsi66qF7IKsIl0uaWxvY1t0b3BdLm1hcChsYW1iZGEgbnM6IEtF"
               "WVdPUkQgaW4gbnMpLnRvX251bXB5KCkgICAjIOygnOuqqSDrrLjsnpDsl7TsnbQg7JWE64uI6528IO2Yle2DnOyGjCDrqoXsgqwg"
               "66qp66Gd7JeQ7IScIOuzuOuLpAogICAgdGl0bGVzID0gQUlSWyLsgqzqsbTsoJzrqqkiXS5pbG9jW3RvcF0udG9saXN0KCkKCiAg"
               "ICBwcmludChmIuydtOybgyAx7JyEIOy9lOyCrOyduCB7cm5kKGNvczIwWzBdLCAzKTouM2Z9IMK3IDIw7JyEIOy9lOyCrOyduCB7"
               "cm5kKGNvczIwWy0xXSwgMyk6LjNmfSIpCiAgICBmb3IgcmFuaywgKGMsIGgsIHQpIGluIGVudW1lcmF0ZSh6aXAoY29zMjAsIGhh"
               "cywgdGl0bGVzKSwgc3RhcnQ9MSk6CiAgICAgICAgcHJpbnQoZiJ7cmFuazo+Mn3snIQge3JuZChjLCAzKTouM2Z9IHsnICAnIGlm"
               "IGggZWxzZSAnw5cgJ317dH0iKQogICAgcHJpbnQoZiLsnbTsm4MgMjDqsbQg7KSRIOOAjO2ZmOu2iOOAjSDsl4bripQg66+87JuQ"
               "ID0ge2ludCgofmhhcykuc3VtKCkpfSDqsbQiKQogICAgZHVwID0gcGQuU2VyaWVzKHRpdGxlcykudmFsdWVfY291bnRzKCkKICAg"
               "IHByaW50KGYi6rCZ7J2AIOygnOuqqeydtCDqsIDsnqUg66eO7J20IOuQmO2SgOydtOuQnCDsiJggOiB7ZHVwLmlsb2NbMF1967KI"
               "IOKAlCB7ZHVwLmluZGV4WzBdfSIpCgogICAgZmlnLCBheCA9IHBsdC5zdWJwbG90cyhmaWdzaXplPShGSUdfVyAqIDAuOCwgNikp"
               "CiAgICB5ID0gbnAuYXJhbmdlKFRPUF9LKQogICAgYXguYmFyaCh5W2hhc10sIGNvczIwW2hhc10sIGNvbG9yPUNPTFsi7Jew7ZqM"
               "7IOJIl0sIGxhYmVsPSLjgIztmZjrtojjgI0g7J6I7J2MIikKICAgIGF4LmJhcmgoeVt+aGFzXSwgY29zMjBbfmhhc10sIGNvbG9y"
               "PUNPTFsi7ZWY64qY7IOJIl0sIGxhYmVsPSLjgIztmZjrtojjgI0g7JeG7J2MIikKICAgIGF4LnNldF95dGlja3MoeSkKICAgIGF4"
               "LnNldF95dGlja2xhYmVscyh5ICsgMSkKICAgIGF4LmludmVydF95YXhpcygpICAgICAgICAgICAgICAgICAgICAgICAgICAgICAg"
               "ICAgICAgIyAx7JyE6rCAIOunqCDsnIQKICAgIGF4LnNldF94bGltKGNvczIwLm1pbigpIC0gMC4wMiwgY29zMjAubWF4KCkgKyAw"
               "LjAxKQogICAgYXguc2V0X3hsYWJlbCgi7L2U7IKs7J24IikKICAgIGF4LnNldF95bGFiZWwoIuydtOybgyDsiJzsnIQiKQogICAg"
               "YXgubGVnZW5kKGxvYz0ibG93ZXIgcmlnaHQiKQogICAgYXguc2V0X3RpdGxlKGYi7KeI7J2Y7JmAIOqwgOyepSDqsIDquYzsmrQg"
               "7ZWt6rO1IOuvvOybkCAyMOqxtCDigJQge01PREVMX05BTUUuc3BsaXQoJy8nKVstMV19IikKICAgIHBsdC5zaG93KCkKCu2VtOyE"
               "pAoKLSDsmIjsg4Eg7IWLIOykkSDri7XsnYAg65GQIOuyiOynuCwg44CMMe+9njXqsbTjgI0g7J6F64uI64ukLiBHUFUg65+w7YOA"
               "7J6E7J2YIEtVUkUtdjEgwrcg7ZWt6rO1IOuvvOybkCAxMiwyMzLqsbTsl5DshJzripQgMeqxtOydtOqzoCwgQ1BVIOydmCBlNS1z"
               "bWFsbCDCtyA0LDAwMOqxtOyXkOyEnOuKlCAy6rG07J6F64uI64ukLiDsp4jsnZjsmYAg6rCZ7J2AIOyCrOygleydhCDsk7Qg66+8"
               "7JuQIOuMgOu2gOu2hOydgCDjgIztmZjrtojjgI3snYQg7JOw7KeA66eMIOyghOu2gOuKlCDslYTri5nri4jri6QuCi0gS1VSRS12"
               "MSDsnZggMeqxtOydgCA47JyEIOOAjOy9lOuhnOuCmOuhnCDqsrDtla3rkJwg7ZWt6rO16raMIO2ZmOq4iSDqsbDrtoDjgI3snoXr"
               "i4jri6QuIOy9lOyCrOyduOydgCAwLjg5NiDsnbTqs6AsIOOAjO2ZmOu2iOOAjSDrjIDsi6Ag44CM7ZmY6riJ44CN7J2EIOyNvOyd"
               "hCDrv5Ag7KeI7J2Y7JmAIOqwmeydgCDsgqzsoJXsnoXri4jri6QuCi0gZTUtc21hbGwgNCwwMDDqsbTsnZggMuqxtOydgCAz7JyE"
               "IOOAjOy9lOuhnOuCmOuhnCDsnbjtlZwg7ZWt6rO16raMIOy3qOyGjCDsmpTssq0g6rGw67aAIOqxtOOAjeqzvCAxM+ychCDjgIzs"
               "vZTroZzrgpjroZwg7J247ZWcIO2Vreqzteq2jCDst6jshowg67aI6rCA44CN7J206rOgLCDsvZTsgqzsnbjsnYAgMC45Njcgwrcg"
               "MC45NjEg7J6F64uI64ukLiDrkZgg64ukIOuPiCDsnbTslbzquLAg7JeG7J20IOy3qOyGjOulvCDqsbDrtoDri7ntlojri6Tqs6Ag"
               "7I287Iq164uI64ukLgotIOuRkCDrqqjrjbjsnZgg7L2U7IKs7J247J2AIO2BrOq4sOqwgCDri6TrpoXri4jri6QuIOydtOybgyAx"
               "7JyEIO+9niAyMOychOqwgCBLVVJFLXYxIOydgCAwLjk0OSDvvZ4gMC44OTAsIGU1LXNtYWxsIDQsMDAw6rG07J2AIDAuOTczIO+9"
               "niAwLjk1OSDsnoXri4jri6QuIOq3uOuemOyEnCDqsIDquYzsmrQg7KCV64+E66GcIOqzoOultOuKlCDsnoTqs4TqsJLsnYAg66qo"
               "642466eI64ukIOuLpOyLnCDsoJXtlanri4jri6QuCi0g7J207JuDIDIw6rG07JeQ64qUIOqwmeydgCDsoJzrqqnsnbQg65CY7ZKA"
               "7J2065Cp64uI64ukLiAxMiwyMzLqsbQg7KCE7LK07JeQ7ISc64qUIOuRkCDrqqjrjbgg66qo65GQIOOAjOy9lOuhnOuCmOuhnCDs"
               "t6jshoztlZwg7ZWt6rO16raMIO2ZmOu2iCDsmpTqtazjgI3qsIAgMTEg772eIDE57JyE7JeQIDnrsogsIGU1LXNtYWxsIDQsMDAw"
               "6rG07JeQ7ISc64qUIDUg772eIDfsnITsl5AgM+uyiCDrgpjsmLXri4jri6QuIOq4gOyekOqwgCDqsJnsnYAg7KCc66qp7J2AIOuy"
               "oe2EsOuPhCDqsJnslYQg7J207JuDIOuqqeuhneydhCDssKjsp4Dtlanri4jri6QuIOydtOufsCDqsbDsnZgg6rCZ7J2AIOq4gOyd"
               "mCDruYTsnKjsnYQg6rWs7ZWY64qUIOqyg+ydtCDrr7jshZggMiDsnoXri4jri6QuCgrsvZTrk5zqsIAg64u17J2EIOunjOuTnOuK"
               "lCDrsKnsi50uIGBtb2RlbC5lbmNvZGVgIOqwgCDsoJzrqqnrp4jri6Qg67Kh7YSwIO2VmOuCmOulvCDrp4zrk6Tqs6AgYG5vcm1h"
               "bGl6ZV9lbWJlZGRpbmdzPVRydWVgIOuhnCDquLjsnbTrpbwgMSDroZwg66ee7Lal64uI64ukLiDstpzroKXsnZgg44CM66y47J6l"
               "IOyehOuyoOuUqSDquLjsnbQgPSAxLjAwMDDjgI3snbQg6re4IO2ZleyduOydtOqzoCwg7J20IOyLpOyKteydmCDrkZAg66qo6424"
               "7J2AIOygleq3nO2ZlCDri6jqs4Trpbwg7J2066+4IO2PrO2VqO2VmOqzoCDsnojslrQg7J247J6Q6rCAIOyXhuyWtOuPhCAxIOye"
               "heuLiOuLpC4g7KeI7J2Y64+EIOqwmeydgCDrqqjrjbggwrcg6rCZ7J2AIGBQUkVGSVhgIOuhnCDsnbjsvZTrlKntlojsnLzrr4Dr"
               "oZwgYEFJUl9FTUIgQCBxYCDqsIAg66+87JuQ66eI64ukIOyniOydmOyZgOydmCDsvZTsgqzsnbjsnoXri4jri6QuIGBucC5hcmdz"
               "b3J0KC1jb3MpWzpUT1BfS11gIOqwgCDsvZTsgqzsnbjsnbQg7YGwIDIw6rG07J2EIOqzoOultOqzoCwg44CM7ZmY67aI44CNIOyX"
               "rOu2gOuKlCDspIDruYQg7IWA7J20IO2Yle2DnOyGjOuhnCDrgpjriIggYOuqheyCrGAg66qp66Gd7JeQ7IScIOu0heuLiOuLpC4g"
               "6re4656Y7IScIOOAjO2ZmOu2iOyngOyXsOydhOOAjeyymOufvCDrnYTslrQg7JOw7KeAIOyViuydgCDsoJzrqqnrj4Qg44CM7ZmY"
               "67aI44CN7J20IOyeiOuKlCDsqr3snLzroZwg65Ok7Ja06rCR64uI64ukLgo="),
    "m2": (
               "8J+UkyDrr7jshZggMiDsmYTshLHrs7guIEFJIOy9lOuTnCDtkZzsi5zshKAg7JWE656Y7JeQIOuTpOyWtOyZgOyVvCDtlaAg7L2U"
               "65Oc64qUIOydtCDsoJXrj4TsnoXri4jri6QKCuKWoCDsnbQg7L2U65Oc66W8IOu2gOuluCDtlITroaztlITtirgg7JiICgogICAg"
               "QUkg7L2U65OcIO2RnOyLnOyEoCDslYTrnpgg7KCE7J6EIOy9lOuTnOuKlCBLVVJFLXYxIOycvOuhnCDsoJXtlZwg7J6E6rOE6rCS"
               "IDAuOTAg7J2EIGU1LXNtYWxsIOyXkCDqt7jrjIDroZwg7I2o7IScLCDqsbDsnZgg6rCZ7J2AIOq4gOydhCDsp4DrgpjsuZjqsowg"
               "66eO7J20IOustuyWtC4KICAgIOykgOu5hCDshYDsnZggUkVWLCBMT1csIEVOQ09ERVJTLCBlbWJlZCwgT0xEX1RIUkVTSE9MRCwg"
               "cm5kIOuKlCDsnbTrr7gg7J6I7Jy864uIIOuLpOyLnCDrp4zrk6Tsp4Ag66eQ6rOgIOq3uOuMgOuhnCDsjagg7KSYLgogICAg65GQ"
               "IOuqqOuNuCDrqqjrkZAg6rCZ7J2AIFJFVlsi67O466y4X+ygleumrCJdIOulvCBlbWJlZCDroZwg7J247L2U65Sp7ZWY6rOgLCDs"
               "nIQg7IK86rCBKGkgPCBqKSDsjI3snZgg7L2U7IKs7J247J2EIOuqqOuNuOuniOuLpCDqtaztlbQg7KSYLgogICAg7J6E6rOE6rCS"
               "IDAuOTAg7JeQ7IScIOuqqOuNuOuniOuLpCDsjI0g7IiY7JmAIOuzhOygkCDqsIjrprAg67mE7JyoKExPVyDqsIAg7ISc66GcIOuL"
               "pOuluCDsjI3snZggJSnsnYQg6rWs7ZW07IScCiAgICAiS1VSRS12MSDsnoTqs4TqsJIgMC45MCDsjI0gPSAo7KCV7IiYKSDsjI0i"
               "LCAiZTUtc21hbGwg7J6E6rOE6rCSIDAuOTAg7IyNID0gKOygleyImCkg7IyNIiwKICAgICLrs4TsoJAg6rCI66awIOu5hOycqCBL"
               "VVJFLXYxID0gKOyGjOyImCDssqvsp7gg7J6Q66asKSAlIiwgIuuzhOygkCDqsIjrprAg67mE7JyoIGU1LXNtYWxsID0gKOyGjOyI"
               "mCDssqvsp7gg7J6Q66asKSAlIiDrpbwg7ZWcIO2WieyUqSBwcmludCDtlbQg7KSYLgogICAgZTUtc21hbGwg7L2U7IKs7J247J2E"
               "IO2BsCDsiJzshJzroZwg7KCV66Cs7ZW0IEtVUkUtdjEgMC45MCDsjI0g7IiYIOuyiOynuCDqsJLsnYQg64uk7IucIOygle2VnCDs"
               "noTqs4TqsJLsnLzroZwg7ZWY6rOgLAogICAgImU1LXNtYWxsIOuLpOyLnCDsoJXtlZwg7J6E6rOE6rCSID0gKOyGjOyImCDrhLfs"
               "p7gg7J6Q66asKSIsICLri6Tsi5wg7KCV7ZWcIOyehOqzhOqwkuyXkOyEnCBlNS1zbWFsbCDsjI0gPSAo7KCV7IiYKSDsjI0iIOyd"
               "hCBwcmludCDtlbQg7KSYLgogICAg6re4IOyehOqzhOqwkiDsnbTsg4Hsnbgg7IyN7JeQIO2VnCDrsogg7J207IOBIOuCmOyYqCDr"
               "pqzrt7Ag7IiY7JmAIOumrOu3sCDsiJjroZwg64KY64iIIOu5hOycqOydhAogICAgIuqxsOydmCDqsJnsnYAg6riA7J20IOyeiOuK"
               "lCDrpqzrt7AgPSAo7KCV7IiYKSDqsbQgKCjshozsiJgg7LKr7Ke4IOyekOumrCkgJSkiIOuhnCBwcmludCDtlZjqs6AsICLtjJDr"
               "i6ggOiIg7ZaJ7J2AIOyngOybjCDspJguCiAgICDrkZAg66qo64247J2YIOyMjSDsiJgg67Cw7JyoLCDri6Tsi5wg7KCV7ZWcIOye"
               "hOqzhOqwkuydmCDrs4TsoJAg6rCI66awIOu5hOycqCwgS1VSRS12MSAwLjkwIOydmCDqsbDsnZgg6rCZ7J2AIOq4gCDruYTsnKgs"
               "CiAgICDri6Tsi5wg7KCV7ZWcIOyehOqzhOqwkuyXkCDqsIDsnqUg6rCA6rmM7Jq0IGU1LXNtYWxsIOyMjSAz6rCc7J2YIOy9lOyC"
               "rOyduCDCtyDrs4TsoJAgwrcg67O466y464+EIOywjeyWtCDspJguCiAgICDqt7jrnpjtlITripQg7ZWcIOyepeycvOuhnCDqt7jr"
               "oKQg7KSYLiDsnoTqs4TqsJIgMC44NSDvvZ4gMC45NyAoMC4wMSDqsITqsqkp66eI64ukIOuRkCDrqqjrjbjsnZgg7J6E6rOE6rCS"
               "IOydtOyDgeyduCDsjI0g7IiY66W8IOyEoCAy6rCc66GcLCDshLjroZzstpXsnYAg66Gc6re4IOyKpOy8gOydvCwKICAgIDAuOTAg"
               "6rO8IOuLpOyLnCDsoJXtlZwg7J6E6rOE6rCS7JeQIOyEuOuhnOyEoCwg67KU66GALCDqsIDroZzstpUg7J2066aEICLsnoTqs4Tq"
               "sJIiLCDshLjroZzstpUg7J2066aEICLsnoTqs4TqsJIg7J207IOB7J24IOyMjSDsiJgiIOuhnCDtlbQg7KSYLgoKICAgIEUgPSB7"
               "bmFtZTogZW1iZWQobmFtZSwgUkVWWyLrs7jrrLhf7KCV66asIl0pIGZvciBuYW1lIGluIEVOQ09ERVJTfSAgICMg6rCZ7J2AIFJF"
               "ViDtlonsnYQg65GQIOuqqOuNuOuhnCwgREVWSUNFIOyXkOyEnAogICAgaXUgPSBucC50cml1X2luZGljZXMobGVuKFJFViksIDEp"
               "ICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICAjIOychCDsgrzqsIEg4oCUIOuMgOqwgeyEoOqzvCDqsJnsnYAg7IyN7J2Y"
               "IOykkeuzteydhCDruoDri6QKICAgIHNwbGl0ID0gTE9XW2l1WzBdXSAhPSBMT1dbaXVbMV1dICAgICAgICAgICAgICAgICAgICAg"
               "ICAgICAgICAgICAgIyDrs4TsoJAgMSDCtyAy7KCQ6rO8IDQgwrcgNeygkOydtCDsp53snbgg7IyNCiAgICBjb3MgPSB7bmFtZTog"
               "KFggQCBYLlQpW2l1XSBmb3IgbmFtZSwgWCBpbiBFLml0ZW1zKCl9ICAgICAgICAgICAgICMg66qo642466eI64ukIOuqqOuToCDs"
               "jI3snZgg7L2U7IKs7J24CgoKICAgIGRlZiBidW5kbGUoYywgdGhyKToKICAgICAgICAjIOyehOqzhOqwkiB0aHIg7J207IOB7J24"
               "IOyMjSDsiJggwrcg6re4IOqwgOyatOuNsCDrs4TsoJAg6rCI66awIOu5hOycqCglKSDCtyDtlZwg7IyNIOydtOyDgeyXkCDrgpjs"
               "mKgg66as67ewIOyImAogICAgICAgIG0gPSBjID49IHRocgogICAgICAgIGlucGFpciA9IG5wLnplcm9zKGxlbihSRVYpLCBib29s"
               "KQogICAgICAgIGlucGFpcltpdVswXVttXV0gPSBUcnVlCiAgICAgICAgaW5wYWlyW2l1WzFdW21dXSA9IFRydWUKICAgICAgICBy"
               "ZXR1cm4gaW50KG0uc3VtKCkpLCAocm5kKDEwMCAqIHNwbGl0W21dLm1lYW4oKSwgMSkgaWYgbS5hbnkoKSBlbHNlIDAuMCksIGlu"
               "dChpbnBhaXIuc3VtKCkpCgoKICAgIG5faywgcF9rLCBrX2sgPSBidW5kbGUoY29zWyJLVVJFLXYxIl0sIE9MRF9USFJFU0hPTEQp"
               "CiAgICBuX2UsIHBfZSwga19lID0gYnVuZGxlKGNvc1siZTUtc21hbGwiXSwgT0xEX1RIUkVTSE9MRCkKICAgIHByaW50KGYi66as"
               "67ewIOuyoe2EsCA6IHtsZW4oUkVWKTosfeqxtCDDlyDrkZAg66qo6424IMK3IERFVklDRSB7REVWSUNFfSIpCiAgICBwcmludChm"
               "IktVUkUtdjEg7J6E6rOE6rCSIDAuOTAg7IyNID0ge25fazosfSDsjI0iKQogICAgcHJpbnQoZiJlNS1zbWFsbCDsnoTqs4TqsJIg"
               "MC45MCDsjI0gPSB7bl9lOix9IOyMjSIpCiAgICBwcmludChmIiAg6rCZ7J2AIDAuOTAg7JeQ7IScIGU1LXNtYWxsIOyMjSDsiJjr"
               "ipQgS1VSRS12MSDsnZgge3JuZChuX2UgLyBuX2ssIDEpOix967CwIikKICAgIHByaW50KGYi67OE7KCQIOqwiOumsCDruYTsnKgg"
               "S1VSRS12MSA9IHtwX2s6LjFmfSAlIikKICAgIHByaW50KGYi67OE7KCQIOqwiOumsCDruYTsnKggZTUtc21hbGwgPSB7cF9lOi4x"
               "Zn0gJSIpCgogICAgdF9uZXcgPSBmbG9hdChucC5zb3J0KGNvc1siZTUtc21hbGwiXSlbOjotMV1bbl9rIC0gMV0pICAgICAgICAg"
               "ICAjIGU1LXNtYWxsIOy9lOyCrOyduOydhCDtgbAg7Iic7ISc66GcIOuGk+qzoCBuX2sg67KI7Ke4IOqwkgogICAgbl90LCBwX3Qs"
               "IGtfdCA9IGJ1bmRsZShjb3NbImU1LXNtYWxsIl0sIHRfbmV3KQogICAgcHJpbnQoZiJlNS1zbWFsbCDri6Tsi5wg7KCV7ZWcIOye"
               "hOqzhOqwkiA9IHtybmQodF9uZXcsIDQpOi40Zn0iKQogICAgcHJpbnQoZiLri6Tsi5wg7KCV7ZWcIOyehOqzhOqwkuyXkOyEnCBl"
               "NS1zbWFsbCDsjI0gPSB7bl90Oix9IOyMjSIpCiAgICBwcmludChmIiAg6re4IOyMjSDqsIDsmrTrjbAg67OE7KCQIOqwiOumsCDs"
               "jI0gOiB7cF90Oi4xZn0gJSIpCiAgICBwcmludChmIuqxsOydmCDqsJnsnYAg6riA7J20IOyeiOuKlCDrpqzrt7AgPSB7a190Oix9"
               "IOqxtCAoe3JuZCgxMDAgKiBrX3QgLyBsZW4oUkVWKSwgMSk6LjFmfSAlKSIpCiAgICBwcmludChmIiAg67mE6rWQ7ZWgIOqwkiDi"
               "gJQgS1VSRS12MSAwLjkwIOycvOuhnCDqtaztlZjrqbQgOiB7a19rOix9IOqxtCAoe3JuZCgxMDAgKiBrX2sgLyBsZW4oUkVWKSwg"
               "MSk6LjFmfSAlKSIpCiAgICBwcmludCgiICDri6Tsi5wg7KCV7ZWcIOyehOqzhOqwkuyXkCDqsIDsnqUg6rCA6rmM7Jq0IGU1LXNt"
               "YWxsIOyMjSAz6rCcIOKAlCDqsJnsnYAg6riA66GcIOustuyWtOuPhCDrkJjripTsp4Ag7J297Iq164uI64ukIikKCiAgICBuZWFy"
               "ID0gbnAuYXJnc29ydChucC5hYnMoY29zWyJlNS1zbWFsbCJdIC0gdF9uZXcpKVs6M10gICAgICAgICAgICMg64uk7IucIOygle2V"
               "nCDsnoTqs4TqsJIg67CU66GcIOqzgeydmCDsjI0g7IWLIOKAlCDsp4HsoJEg7J2964qU64ukCiAgICBmb3IgaiBpbiBuZWFyOgog"
               "ICAgICAgIGEsIGIgPSBpdVswXVtqXSwgaXVbMV1bal0KICAgICAgICBwcmludChmIiAge3JuZChjb3NbJ2U1LXNtYWxsJ11bal0s"
               "IDQpOi40Zn0gwrcge1JFVlsn67OE7KCQJ11bYV197KCQIOOAjHtSRVZbJ+uzuOusuF/soJXrpqwnXVthXVs6NDBdfeOAjSIKICAg"
               "ICAgICAgICAgICBmIiDCtyB7UkVWWyfrs4TsoJAnXVtiXX3soJAg44CMe1JFVlsn67O466y4X+ygleumrCddW2JdWzo0MF1944CN"
               "IikKCiAgICBncmlkID0gbnAucm91bmQobnAuYXJhbmdlKDAuODUsIDAuOTcwMSwgMC4wMSksIDIpICAgICAgICAgICAgICAgICMg"
               "7J6E6rOE6rCSIDAuODUg772eIDAuOTcsIDEz6rCcCiAgICBmaWcsIGF4ID0gcGx0LnN1YnBsb3RzKGZpZ3NpemU9KEZJR19XICog"
               "MC43LCA0LjApKQogICAgZm9yIG5hbWUsIGNvbG9yIGluICgoIktVUkUtdjEiLCBDT0xbIuuCqOyDiSJdKSwgKCJlNS1zbWFsbCIs"
               "IENPTFsi7ZWY64qY7IOJIl0pKToKICAgICAgICBheC5wbG90KGdyaWQsIFtpbnQoKGNvc1tuYW1lXSA+PSB0KS5zdW0oKSkgZm9y"
               "IHQgaW4gZ3JpZF0sIG1hcmtlcj0ibyIsIGNvbG9yPWNvbG9yLCBsYWJlbD1uYW1lKQogICAgYXguYXh2bGluZShPTERfVEhSRVNI"
               "T0xELCBjb2xvcj1DT0xbIu2ajOyDiSJdLCBsaW5lc3R5bGU9Ii0tIiwgbGFiZWw9IuyYmyDsnoTqs4TqsJIgMC45MCIpCiAgICBh"
               "eC5heHZsaW5lKHRfbmV3LCBjb2xvcj1DT0xbIu2VmOuKmOyDiSJdLCBsaW5lc3R5bGU9IjoiLCBsYWJlbD1mIuuLpOyLnCDsoJXt"
               "lZwgZTUtc21hbGwg7J6E6rOE6rCSIHtybmQodF9uZXcsIDQpOi40Zn0iKQogICAgYXguc2V0X3lzY2FsZSgibG9nIikgICAgICAg"
               "ICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICAjIOyMjSDsiJjqsIAg7IiY7Iut7JeQ7IScIOyImOyLreunjOq5"
               "jOyngOudvCDroZzqt7gg7Iqk7LyA7J28CiAgICBheC5zZXRfeGxhYmVsKCLsnoTqs4TqsJIiKQogICAgYXguc2V0X3lsYWJlbCgi"
               "7J6E6rOE6rCSIOydtOyDgeyduCDsjI0g7IiYICjroZzqt7gg7Iqk7LyA7J28KSIpCiAgICBheC5sZWdlbmQoZm9udHNpemU9OSkK"
               "ICAgIGF4LnNldF90aXRsZShmIuumrOu3sCB7bGVuKFJFVik6LH3qsbQg4oCUIOyehOqzhOqwkuuzhCDsnoTqs4TqsJIg7J207IOB"
               "7J24IOyMjSDsiJgiKQogICAgcGx0LnNob3coKQoK7ZW07ISkCgotIOyYiOyDgSDshYsg7KSRIOuLteydgCDssqsg67KI7Ke4LCDj"
               "gIxLVVJFLXYxIOyXkOyEnCDsjI3snYQg7J297Ja0IOygle2VnCAwLjkwIOydhCBlNS1zbWFsbCDsl5Ag6re464yA66GcIOyNvOuL"
               "pOOAjSDsnoXri4jri6QuIEtVUkUtdjEg7JeQ7IScIDAuOTAg7J2AIDk567Cx67aE7JyE7IiYIDAuNzA3IOuztOuLpCDsnITrnbwg"
               "65Oc66y4IOyMjeunjCDrhJjsp4Drp4wsIGU1LXNtYWxsIOyXkOyEnOuKlCA5Oeuwseu2hOychOyImCAwLjkxMyDrs7Tri6Qg7JWE"
               "656Y7J6F64uI64ukLiDqt7jrnpjshJwg6rCZ7J2AIDAuOTAg7JeQ7IScIOumrOu3sCAzLDAwMOqxtOydmCDsnoTqs4TqsJIg7J20"
               "7IOB7J24IOyMjeydtCBLVVJFLXYxIDExMOyMjSwgZTUtc21hbGwgMTc1LDU5MOyMjeycvOuhnCAxLDU5NuuwsOyeheuLiOuLpC4g"
               "Q1BVIOydmCAxLDAwMOqxtOyXkOyEnOuKlCAxM+yMjeqzvCAxOSw2NTTsjI3snoXri4jri6QuIOOAjHF1ZXJ5OiDjgI0g7KCR65GQ"
               "7Ja064qUIGU1LXNtYWxsIOydmCDrqqjrjbgg7ISk66qF7J20IOuNlO2VmOudvOqzoCDslYjrgrTtlZjripQg6rKD7J206rOgLCDr"
               "pqzrt7Drpbwg64qY66Ck64+EIOuqqOuNuOyXkCDrp57sp4Ag7JWK64qUIOyehOqzhOqwkuydgCDqt7jrjIDroZzsnoXri4jri6Qu"
               "Ci0gMC45MCDsl5DshJwg7J6E6rOE6rCSIOydtOyDgeyduCDsjI0g6rCA7Jq0642wIOuzhOygkCDqsIjrprAg67mE7Jyo7J2AIEtV"
               "UkUtdjEgOC4yJSwgZTUtc21hbGwgMjMuMiUg7J6F64uI64ukKENQVSAxLDAwMOqxtCAwLjAlIMK3IDI0LjQlKS4gZTUtc21hbGwg"
               "7J2YIDAuOTAg7J2AIOuzhOygkOydtCDrsJjrjIDsnbgg6riA6rmM7KeAIOqwmeydgCDquIDroZwg66y27Iq164uI64ukLgotIGU1"
               "LXNtYWxsIOydmCDri6Tsi5wg7KCV7ZWcIOyehOqzhOqwkuydgCDrpqzrt7AgMywwMDDqsbTsl5DshJwgMC45NjQ3LCAxLDAwMOqx"
               "tOyXkOyEnCAwLjk2NDYg7J6F64uI64ukLiDqt7gg7J6E6rOE6rCS7JeQ7IScIOustuyduCBlNS1zbWFsbCDsjI3snYAgS1VSRS12"
               "MSDqs7wg6rCZ7J2AIDExMOyMjeqzvCAxM+yMjeydtOqzoCwg67OE7KCQIOqwiOumsCDruYTsnKjsnYAgMTAuMCUg7JmAIDAuMCUg"
               "7J6F64uI64ukLiDriIjrjIDspJHsnLzroZwgMC45NSDroZwg7Jis66as66m0IDMsMDAw6rG07JeQ7IScIDg3NOyMjSwgMSwwMDDq"
               "sbTsl5DshJwgODjsjI3snbTrnbwg7Jes7KCE7Z6IIEtVUkUtdjEg7IyNIOyImOydmCA267Cw6rCAIOuEmOyKteuLiOuLpC4KLSDq"
               "sbDsnZgg6rCZ7J2AIOq4gOydtCDsnojripQg66as67ew64qUIDMsMDAw6rG07JeQ7IScIOyghOyehCDsvZTrk5zsnZggMiw4NDHq"
               "sbQgwrcgOTQuNyUg6rCAIDky6rG0IMK3IDMuMSUg66GcIOykhOyWtOuTreuLiOuLpC4gMSwwMDDqsbTsl5DshJzripQgODky6rG0"
               "IMK3IDg5LjIlIOqwgCAxNOqxtCDCtyAxLjQlIOuhnCDspITslrTrk63ri4jri6QuIEtVUkUtdjEgMC45MCDsnLzroZwg6rWs7ZWY"
               "66m0IDEyMOqxtCDCtyA0LjAlIOyZgCAyM+qxtCDCtyAyLjMlIOudvCDqsJnsnYAg7YGs6riw7J6F64uI64ukLiDjgIzrpqzrt7Ag"
               "64yA67aA67aE7J20IOuzteyCrO2VnCDquIDjgI3snbTrnbzripQg6rKw66Gg7J2AIOyduOy9lOuNlOyXkCDrp57sp4Ag7JWK64qU"
               "IOyehOqzhOqwkuydtCDrp4zrk6Ag6rKw66Gg7J206rOgLCDqs6DsuZwg65Kk7JeQ64qUIOumrOu3sCDrqocgJeydmCDsnbTslbzq"
               "uLDsnoXri4jri6QuCi0g64uk7IucIOygle2VnCDsnoTqs4TqsJLsnYAg7IyN7J2YIOyImOunjCDrp57stpgg7Lac67Cc7KCQ7J6F"
               "64uI64ukLiDstpzroKXsnZgg44CM6rCA7J6lIOqwgOq5jOyatCDsjI0gM+qwnOOAjeulvCDsnb3qs6Ag6rCZ7J2AIOq4gOuhnCDr"
               "rLbslrTrj4Qg65CY64qU7KeAIO2ZleyduO2VnCDrkqQg6riw7KSA7Jy866GcIOyUgeuLiOuLpC4KCuy9lOuTnOqwgCDri7XsnYQg"
               "66eM65Oc64qUIOuwqeyLnS4gYGVtYmVkYCDqsIAg6rCZ7J2AIGBSRVZgIOulvCDrkZAg66qo642466GcIOyduOy9lOuUqe2VmOqz"
               "oCwg6ri47J20IDEg67Kh7YSw7J2YIO2WieugrOqzsSBgWCBAIFguVGAg6rCAIOuqqOuToCDsjI3snZgg7L2U7IKs7J247J6F64uI"
               "64ukLiBgbnAudHJpdV9pbmRpY2VzYCDroZwg7JyEIOyCvOqwgeunjCDqurzrgrQg7J6Q6riwIOyekOyLoOqzvOydmCDsjI3qs7wg"
               "6rCZ7J2AIOyMjeydmCDspJHrs7XsnYQg67qN64uI64ukLiDri6Tsi5wg7KCV7ZWcIOyehOqzhOqwkuydgCBlNS1zbWFsbCDsvZTs"
               "gqzsnbjsnYQg64K066a87LCo7Iic7Jy866GcIOygleugrO2VnCDrsLDsl7TsnZggYG5fayAtIDFgIOuyiCDsnITsuZjsnZgg6rCS"
               "7J206528LCDqt7gg6rCSIOydtOyDgeyduCDsjI3snbQg7KCV7ZmV7Z6IIGBuX2tgIOqwnCDrgqjsirXri4jri6QuIOqxsOydmCDq"
               "sJnsnYAg6riA7J20IOyeiOuKlCDrpqzrt7DripQg7J6E6rOE6rCSIOydtOyDgeyduCDsjI3snZgg65GQIOumrOu3sOulvCBib29s"
               "IOuwsOyXtOyXkOyEnCDrqqjrkZAgVHJ1ZSDroZwg67CU6r68IOuSpCBUcnVlIOyduCDrpqzrt7Drpbwg7IS46rOgLCDrpqzrt7Ag"
               "7IiY66GcIOuCmOuIleuLiOuLpC4K"),
}


# 항목별 힌트 — base64(JSON)
_HINT = _dec_json(
               "eyJtMSI6IHsi7J207JuDIDIw6rG0IOykkSDjgIztmZjrtojjgI0g7JeG64qUIOuvvOybkCI6IFsi7J297J2AIOqwkuycvOuhnCDs"
               "i6TsiJjrpbwg7Kea7Iq164uI64ukLiAwIOydtOuptCDjgIztmZjrtojjgI3snbQg7JeG64qUIOuvvOybkOydtCDtlZjrgpjrj4Qg"
               "7JeG64uk64qUIOucu+ydtOudvCDsnbTsm4PsnYQg6rOg66W064qUIOywqOuhgOuCmCDshLjripQg64yA7IOB7J2EIOu0heuLiOuL"
               "pC4gMjAg67O064ukIO2BrOuptCDsnbTsm4MgMjDqsbTsnbQg7JWE64uI6528IO2VreqztSDrr7zsm5Ag7KCE7LK07JeQ7IScIOyF"
               "jOqzoCwg7J2M7IiY66m0IOu2gO2YuOqwgCDrtpnsl4jsirXri4jri6QuIOuwlOuluCDqsJLsnYAg7KeI7J2Y7JmAIOy9lOyCrOyd"
               "uOydtCDtgbAgMjDqsbQg6rCA7Jq0642wIGDrqoXsgqxgIOuqqeuhneyXkCBgS0VZV09SRGAg6rCAIOyXhuuKlCDrr7zsm5Ag7IiY"
               "7J6F64uI64ukLiDjgIzsnbTsm4MgMjDqsbQg7KSRIOOAjO2ZmOu2iOOAjSDsl4bripQg66+87JuQID0g7Iir7J6QIOqxtOOAjSDt"
               "lZwg7ZaJ7Jy866GcIOy2nOugpe2VqeuLiOuLpCIsICJgQUlSX0VNQiBAIHFgIOuhnCDsvZTsgqzsnbjsnYQg6rWs7ZW0IGBucC5h"
               "cmdzb3J0KC1jb3MpWzpUT1BfS11gIOyymOufvCDtgbAg7Iic7ISc66GcIDIw6rG07J2EIOqzoOultOqzoCwg6re4IDIw6rG07J2Y"
               "IGDrqoXsgqxgIOuqqeuhneyXkCBgS0VZV09SRGAg6rCAIOyXhuuKlCDqsbTsiJjrpbwg7IWJ64uI64ukLiDsp4jsnZjsmYAg7KCc"
               "66qpIOuqqOuRkCDslZ7sl5AgYFBSRUZJWGAg66W8IOuRoeuLiOuLpCIsICLrgrQg7L2U65Oc6rCAIOydtOybgyAyMOqxtOydhCDs"
               "lrTripAg67Cp7Zal7Jy866GcIOygleugrO2VtCDqs6jrnpDqs6Ag44CM7ZmY67aI44CNIOyXrOu2gOulvCDslrTripAg7Je07JeQ"
               "7IScIOu0pOuKlOyngCJdLCAi66y47J6lIOyehOuyoOuUqSDquLjsnbQiOiBbIuydveydgCDqsJLsnLzroZwg7Iuk7IiY66W8IOyn"
               "muyKteuLiOuLpC4gNjMuMihDUFUg7J2YIDQsMDAw6rG0KeuCmCAxMTAuNihHUFUg7J2YIDEyLDIzMuqxtCnsspjrn7wgMSDrs7Tr"
               "i6Qg7YGs66m0IO2WiSDtlZjrgpjqsIAg7JWE64uI6528IOuwsOyXtCDsoITssrTsnZgg6ri47J20IGBucC5saW5hbGcubm9ybShB"
               "SVJfRU1CKWAg66W8IOq1rO2WiOyKteuLiOuLpC4g6ri47J20IDEg7J24IO2WiSBOIOqwnOulvCDrqqjsnYAg67Cw7Je07J2YIOq4"
               "uOydtOuKlCBOIOydmCDsoJzqs7Hqt7zsnoXri4jri6QoNCwwMDAg7J2YIOygnOqzseq3vOydgCA2My4yKS4g7ZaJIO2VmOuCmCBg"
               "QUlSX0VNQlswXWAg7J2YIOq4uOydtOulvCDjgIzrrLjsnqUg7J6E67Kg65SpIOq4uOydtCA9IOyIq+yekOOAjSDtlZwg7ZaJ7Jy8"
               "66GcLCDshozsiJgg64S37Ke4IOyekOumrOq5jOyngCDstpzroKXtlanri4jri6QiLCAiYGVuY29kZSguLi4sIG5vcm1hbGl6ZV9l"
               "bWJlZGRpbmdzPVRydWUpYCDroZwg66eM65OgIOusuOyepSDsnoTrsqDrlKkg7ZWY64KY7J2YIOq4uOydtCBgbnAubGluYWxnLm5v"
               "cm0oQUlSX0VNQlswXSlgIOulvCDstpzroKXtlanri4jri6QuIOqwkuydtCAxIOyXkOyEnCDrspfslrTrgpjrqbQg7KSA67mEIOyF"
               "gOydmCBgTU9ERUxfTkFNRWAgwrcgYE1PREVMX1JFVmAg66GcIOu2iOufrOyZlOuKlOyngCDrs7TshLjsmpQiLCAi64K0IOy9lOuT"
               "nOqwgCDrrLjsnqUg7J6E67Kg65SpIOq4uOydtOulvCDrsLDsl7TsnZgg7Ja064qQIOu2gOu2hOycvOuhnCDqs4TsgrDtlojripTs"
               "p4AiXSwgInJlY291bnQiOiBbIuyxhOygkOydgCDrgrQg7L2U65Oc6rCAIOunjOuToCBgQUlSX0VNQmAg7JmAIGBxYCDroZwgYEFJ"
               "Ul9FTUIgQCBxYCDsnZgg7YGwIOyInOyEnCAyMOqxtOydhCDri6Tsi5wg6rOo6528LCBg66qF7IKsYCDrqqnroZ3sl5Ag44CM7ZmY"
               "67aI44CN7J20IOyXhuuKlCDrr7zsm5DsnYQg7IWJ64uI64ukLiDsnIQg7ZaJ7J2YIOuRkCDsiJjqsIAg64uk66W066m0IOygleug"
               "rCDrsKntlqUoYG5wLmFyZ3NvcnQoLWNvcylgKSwg7IS464qUIOuMgOyDgSjjgIztmZjrtojjgI3snbQg7JeG64qUIOuvvOybkCks"
               "IOywvuuKlCDsl7QoYOuqheyCrGApIOqwgOyatOuNsCDtlZjrgpjqsIAg7Ja06riL64Ks7Iq164uI64ukLiDsnIQg7ZaJIOuBneyd"
               "mCDshKTrqoXsnbQg7Ja064qQIOqyg+yduOyngCDsp5rsirXri4jri6QiLCAiYHRvcCA9IG5wLmFyZ3NvcnQoLWNvcylbOlRPUF9L"
               "XWAg66GcIO2BsCDsiJzshJwgMjDqsbTsnYQg6rOg66W06rOgLCBgQUlSW1wi66qF7IKsXCJdLmlsb2NbdG9wXWAg7JeQ7IScIGBL"
               "RVlXT1JEYCDqsIAg7JeG64qUIO2WieydhCDshYnri4jri6QuIOygnOuqqSDrrLjsnpDsl7TsnYQg6rO167Cx7Jy866GcIOyekOul"
               "tOuptCDjgIztmZjrtojsp4Dsl7DsnYTjgI0g6rCZ7J2AIOyWtOygiOydtCDjgIztmZjrtojjgI3qs7wg6rCZ7KeAIOyViuyVhCDs"
               "l4bripQg7Kq97Jy866GcIOuTpOyWtOqwkeuLiOuLpCIsICLrgrQg7L2U65Oc6rCAIOydtOybgyAyMOqxtOydhCDslrTripAg67Cp"
               "7Zal7Jy866GcIOqzqOuekOqzoCwg44CM7ZmY67aI44CNIOyXrOu2gOulvCDslrTripAg7Je07JeQ7IScIOustOyXh+ydhCDshYzr"
               "ipTsp4AiXSwgInBsb3QiOiBbIuq3uOuemO2UhOulvCDsnbQg7IWA7JeQ7IScIOq3uOuguOuKlOyngCwg6rCA66Gc7LaV6rO8IOyE"
               "uOuhnOy2leyXkCDsnbTrpoQo44CM7L2U7IKs7J2444CNwrfjgIzsnbTsm4Mg7Iic7JyE44CNKeydhCDri6zslZjripTsp4Ag67O0"
               "7IS47JqULiDqt7jrnpjtlITripQg6re466a87J20IOyeiOuKlOyngOyZgCDrkZAg7LaV7J2YIOydtOumhOunjCDssYTsoJDtlanr"
               "i4jri6QiLCAiYGF4LnNldF94bGFiZWwoXCLsvZTsgqzsnbhcIilgIOqzvCBgYXguc2V0X3lsYWJlbChcIuydtOybgyDsiJzsnIRc"
               "IilgIOydhCDqt7jrpqzripQg7L2U65Oc7JeQIOuEo+ycvOudvOqzoCBQUk9NUFQg7JeQIOyggeycvOyEuOyalC4g66eJ64yA7J2Y"
               "IOyImCDCtyDsg4kgwrcg67KU66GA64qUIOyxhOygkO2VmOyngCDslYrsirXri4jri6QiLCAi64K0IOq3uOuemO2UhOydmCDqsIDr"
               "oZzstpXCt+yEuOuhnOy2lSDsnbTrpoTsnbQg7Ja065SU7IScIOu5oOyhjOuKlOyngCJdLCAiZXJyb3IiOiBbIk5hbWVFcnJvciDr"
               "qbQg67CU66GcIOychCDjgIzrr7jshZggMSDspIDruYQg7IWA44CN7J2EIOuovOyggCDsi6TtlontlojripTsp4DrtoDthLAg7ZmV"
               "7J247ZWY7IS47JqULiDsi6TtlontlojripTrjbDrj4Qg7Jik66WY6rCAIOuCmOuptCDsmKTrpZgg66y47J6lIOyyqyDtlonsnYQg"
               "6re464yA66GcIEFJIO2KnO2EsOyXkCDrs7Tsl6wg7KO86rOgLCDjgIzrr7jshZggMSDspIDruYQg7IWA44CN7J20IOunjOuToCDs"
               "nbTrpoQoQUlSIMK3IFFVRVJZIMK3IEtFWVdPUkQgwrcgVE9QX0sgwrcgREVWSUNFIMK3IE1PREVMX05BTUUgwrcgTU9ERUxfUkVW"
               "IMK3IFBSRUZJWCDCtyBybmQgwrcgQ09MIMK3IEZJR19XKeydgCDri6Tsi5wg66eM65Ok7KeAIOunkOqzoCDquIDsnpAg6re464yA"
               "66GcIOyTsOudvOqzoCDsoIHsnLzshLjsmpQuIOuqqOuNuOydgCBgU2VudGVuY2VUcmFuc2Zvcm1lcihNT0RFTF9OQU1FLCBkZXZp"
               "Y2U9REVWSUNFLCByZXZpc2lvbj1NT0RFTF9SRVYpYCDroZwg67aI65+s7Ji164uI64ukIiwgIuydtCDsmKTrpZjqsIAg7Ja064qQ"
               "IO2WieyXkOyEnCDsmZwg64Ks64qU7KeAIl0sICJwcmVmaXgiOiBbImU1LXNtYWxsIOydgCDsp4jsnZjsmYAg66+87JuQIOygnOuq"
               "qSDrqqjrkZAg7JWe7JeQIGBQUkVGSVhgKOOAjHF1ZXJ5OiDjgI0p66W8IOuRkOqzoCDsnbjsvZTrlKntlanri4jri6QuIOychCDt"
               "lonsnbQg7Kea7J2AIOyqveydhCBgbW9kZWwuZW5jb2RlKFtQUkVGSVggKyBRVUVSWV0sIOKApilgIMK3IGBbUFJFRklYICsgdCBm"
               "b3IgdCBpbiBBSVJbXCLsgqzqsbTsoJzrqqlcIl1dYCDsspjrn7wg64uk7IucIOyduOy9lOuUqe2VmOudvOqzoCBQUk9NUFQg7JeQ"
               "IOyggeycvOyEuOyalCIsICLssYTsoJDsnYAg64K0IOy9lOuTnOqwgCDrtojrn6zsmKgg66y47J6lIOyduOy9lOuNlOuhnCDsp4js"
               "nZjsmYAg7LKrIOygnOuqqeydhCDsoJHrkZDslrTrpbwg642U7ZW0IMK3IOu5vOqzoCDtlZwg67KI7JSpIOyduOy9lOuUqe2VtCDr"
               "grQg67Kh7YSw7JmAIOu5hOq1kO2VqeuLiOuLpC4g7ZWc7Kq97JeQ66eMIOygkeuRkOyWtOqwgCDsnojsnLzrqbQg7J2067KIIOyn"
               "iOydmOyXkOyEnOuKlCDqsbTsiJjqsIAg6rCZ6rKMIOuCmOyZgOuPhCDri6Trpbgg7KeI7J2Y7JeQ7ISc64qUIOydtOybgyDsiJzs"
               "nITqsIAg64us65287KeR64uI64ukIiwgIuuCtCDsvZTrk5zqsIAg7KeI7J2Y7JmAIOygnOuqqSDqsIDsmrTrjbAg7Ja064qQIOyq"
               "veyXkCBQUkVGSVgg66W8IOuRkOyXiOuKlOyngCJdfSwgIm0yIjogeyJLVVJFLXYxIOyehOqzhOqwkiAwLjkwIOyMjSI6IFsi7J29"
               "7J2AIOqwkuycvOuhnCDsi6TsiJjrpbwg7Kea7Iq164uI64ukLiDrpqzrt7AgMSwwMDDqsbQoQ1BVKeyXkOyEnCAxLDAyNiwgMyww"
               "MDDqsbQoR1BVKeyXkOyEnCAzLDIyMCDsspjrn7wg66as67ewIOyImOuztOuLpCDtgazrqbQg7L2U7IKs7J24IO2WieugrCDsoITs"
               "srTrpbwg7IWM7Iq164uI64ukLiDsnpDquLAg7J6Q7Iug6rO87J2YIOyMjSjsvZTsgqzsnbggMSnsnbQg66as67ewIOyImOunjO2B"
               "vCDrjZTtlbTsp4Dqs6Ag6rCZ7J2AIOyMjeydtCDrkZAg67KIIOuTpOyWtOqwkeuLiOuLpCgxMyDDlyAyICsgMSwwMDAgPSAxLDAy"
               "NikuIOydjOyImOuptCDrtoDtmLjqsIAg67aZ7JeI7Iq164uI64ukLiDsnIQg7IK86rCBKGBucC50cml1X2luZGljZXMobGVuKFJF"
               "ViksIDEpYCnsl5DshJwgMC45MCDsnbTsg4Hsnbgg7IyN7J2EIOyEuOyWtCDjgIxLVVJFLXYxIOyehOqzhOqwkiAwLjkwIOyMjSA9"
               "IOygleyImCDsjI3jgI0g7ZWcIO2WieycvOuhnCDstpzroKXtlanri4jri6QiLCAiYGVtYmVkKFwiS1VSRS12MVwiLCBSRVZbXCLr"
               "s7jrrLhf7KCV66asXCJdKWAg66GcIOqwmeydgCDrpqzrt7Drpbwg7J247L2U65Sp7ZWY6rOgLCDsnIQg7IK86rCBIOy9lOyCrOyd"
               "uCDqsIDsmrTrjbAgYE9MRF9USFJFU0hPTERgIOydtOyDgeyduCDsjI3snYQg7IWJ64uI64ukLiDsnbTrpoTsnZggMC45MCDrj4Qg"
               "6re464yA66GcIOyggeyKteuLiOuLpCIsICLrgrQg7L2U65Oc6rCAIEtVUkUtdjEg7IyNIOyImOulvCDsvZTsgqzsnbgg7ZaJ66Cs"
               "7J2YIOyWtOuKkCDrtoDrtoTsl5DshJwg7IWM64qU7KeAIl0sICJlNS1zbWFsbCDsnoTqs4TqsJIgMC45MCDsjI0iOiBbIuydveyd"
               "gCDqsJLsnLzroZwg7Iuk7IiY66W8IOynmuyKteuLiOuLpC4gS1VSRS12MSDsjI0g7IiY7JmAIOqwmeydgCDsnpDrpr/siJgoQ1BV"
               "IDEzIMK3IEdQVSAxMTAp66m0IOuLpOyLnCDsoJXtlZwg7J6E6rOE6rCS7JeQ7IScIOyFjOqxsOuCmCBLVVJFLXYxIOuhnCDsnbjs"
               "vZTrlKntlojsirXri4jri6QuIOqwmeydgCAwLjkwIOyXkOyEnOuKlCBlNS1zbWFsbCDsjI3snbQgQ1BVIDE5LDY1NCDCtyBHUFUg"
               "MTc1LDU5MCjrmJDripQgMTc1LDU5MSkg7Jy866GcIOyghOyehCDsvZTrk5zsnZgg44CM7J6E6rOE6rCSIOydtOyDgeyduCDsjI3j"
               "gI3qs7wg6rCZ7Iq164uI64ukLiDjgIxlNS1zbWFsbCDsnoTqs4TqsJIgMC45MCDsjI0gPSDsoJXsiJgg7IyN44CNIO2VnCDtlons"
               "nLzroZwg7Lac66Cl7ZWp64uI64ukIiwgIuqwmeydgCBgUkVWYCDrpbwgYGVtYmVkKFwiZTUtc21hbGxcIiwg4oCmKWAg66GcIOyd"
               "uOy9lOuUqe2VtCBLVVJFLXYxIOqzvCDqsJnsnYAg67Cp67KVKOychCDsgrzqsIEgwrcgMC45MCDsnbTsg4Ep7Jy866GcIOyFieuL"
               "iOuLpC4g7KCE7J6EIOy9lOuTnOydmCDsnbjsvZTrlKnqs7wg7IS464qUIOuwqeuyleydgCDti4Drpqzsp4Ag7JWK7JWY6rOgLCDt"
               "i4DrprAg6rKD7J2AIOydtCDsiJjroZwg64K066awIOqysOuhoOyeheuLiOuLpCIsICLrgrQg7L2U65Oc6rCAIGU1LXNtYWxsIOyM"
               "jSDsiJjrpbwg7Ja064qQIOyehOqzhOqwkiDCtyDslrTripAg66as67ew66GcIOyFjOuKlOyngCJdLCAi67OE7KCQIOqwiOumsCDr"
               "uYTsnKggS1VSRS12MSI6IFsi7J297J2AIOqwkuycvOuhnCDsi6TsiJjrpbwg7Kea7Iq164uI64ukLiDsnYzsiJjrqbQg67aA7Zi4"
               "6rCAIOu2meyXiOqxsOuCmCDrkZAg67mE7Jyo7J2YIOywqOulvCDsoIHsl4jqs6AsIDEwMCDrs7Tri6Qg7YGs66m0IDEwMCDsnYQg"
               "65GQIOuyiCDqs7HtlojsirXri4jri6QuIDAuOTAg7JeQ7IScIOustuyduCBLVVJFLXYxIOyMjSDqsIDsmrTrjbAgYExPV1tpXSAh"
               "PSBMT1dbal1gIOyduCDsjI3snZgg67mE7JyoKCUp7J2EIOOAjOuzhOygkCDqsIjrprAg67mE7JyoIEtVUkUtdjEgPSDsiKvsnpAg"
               "JeOAjSDtlZwg7ZaJ7Jy866GcIOy2nOugpe2VqeuLiOuLpCIsICLrtoTrqqjripQg7J6E6rOE6rCSIOydtOyDgeyduCDsjI3snZgg"
               "7IiY7J6F64uI64ukLiDsnoTqs4TqsJIg7J207IOB7J24IOyMjSDqsIDsmrTrjbAg67OE7KCQ7J20IOqwiOumsCDsjI3snbQg7ZWY"
               "64KY64+EIOyXhuycvOuptCAwLjAgJSDrpbwg7Lac66Cl7ZWp64uI64ukIiwgIuuCtCDsvZTrk5zqsIAg67OE7KCQIOqwiOumsCDr"
               "uYTsnKjsnZgg67aE7J6Q7JmAIOu2hOuqqOulvCDrrLTsl4fsnLzroZwg7IWM64qU7KeAIl0sICLrs4TsoJAg6rCI66awIOu5hOyc"
               "qCBlNS1zbWFsbCI6IFsi7J297J2AIOqwkuycvOuhnCDsi6TsiJjrpbwg7Kea7Iq164uI64ukLiDsnYzsiJjrqbQg67aA7Zi46rCA"
               "IOu2meyXiOqxsOuCmCDrkZAg67mE7Jyo7J2YIOywqOulvCDsoIHsl4jqs6AsIDEwMCDrs7Tri6Qg7YGs66m0IDEwMCDsnYQg65GQ"
               "IOuyiCDqs7HtlojsirXri4jri6QuIDAuOTAg7JeQ7IScIOustuyduCBlNS1zbWFsbCDsjI0g6rCA7Jq0642wIGBMT1dbaV0gIT0g"
               "TE9XW2pdYCDsnbgg7IyN7J2YIOu5hOycqCglKeydhCDjgIzrs4TsoJAg6rCI66awIOu5hOycqCBlNS1zbWFsbCA9IOyIq+yekCAl"
               "44CNIO2VnCDtlonsnLzroZwg7Lac66Cl7ZWp64uI64ukIiwgIuu2hOuqqOuKlCDsnoTqs4TqsJIg7J207IOB7J24IOyMjeydmCDs"
               "iJjsnbTqs6AsIOuLpOyLnCDsoJXtlZwg7J6E6rOE6rCS7J20IOyVhOuLiOudvCAwLjkwIOyXkOyEnCDshLwg6rCS7J6F64uI64uk"
               "IiwgIuuCtCDsvZTrk5zqsIAgZTUtc21hbGwg7J2YIOuzhOygkCDqsIjrprAg67mE7Jyo7J2EIOyWtOuKkCDsnoTqs4TqsJLsl5Ds"
               "hJwg7IWM64qU7KeAIl0sICJlNS1zbWFsbCDri6Tsi5wg7KCV7ZWcIOyehOqzhOqwkiI6IFsi7J297J2AIOqwkuycvOuhnCDsi6Ts"
               "iJjrpbwg7Kea7Iq164uI64ukLiAwLjkg66m0IOyYmyDsnoTqs4TqsJIgMC45MCDsnYQg6re464yA66GcIOyggeyXiOyKteuLiOuL"
               "pC4gMC45NSDsspjrn7wg64iI64yA7KSR7Jy866GcIOqzoOuluCDqsJLsnbTrgpggZTUtc21hbGwg7L2U7IKs7J247J2YIDk567Cx"
               "67aE7JyE7IiYKOumrOu3sCAxLDAwMOqxtCAwLjkxNCDCtyAzLDAwMOqxtCAwLjkxMynripQg65GQIOuqqOuNuOydtCDrrLbripQg"
               "7IyN7J2YIOyImOulvCDrp57stpTsp4Ag66q77ZWp64uI64ukLiDsnYzsiJjrqbQg67aA7Zi46rCAIOu2meyXiOyKteuLiOuLpC4g"
               "ZTUtc21hbGwg7J2YIOychCDsgrzqsIEg7L2U7IKs7J247J2EIO2BsCDsiJzshJzroZwg7KCV66Cs7ZW0IEtVUkUtdjEgMC45MCDs"
               "jI0g7IiYIOuyiOynuCDqsJLsnYQg44CMZTUtc21hbGwg64uk7IucIOygle2VnCDsnoTqs4TqsJIgPSDshozsiJgg64S37Ke4IOye"
               "kOumrOOAjSDtlZwg7ZaJ7Jy866GcIOy2nOugpe2VqeuLiOuLpCIsICJgbnAuc29ydChlNSDsvZTsgqzsnbgpWzo6LTFdW0tVUkUg"
               "7IyNIOyImCAtIDFdYCDsspjrn7wgS1VSRS12MSDsnZgg7IyNIOyImOulvCDsiJzsnITroZwg7JSB64uI64ukLiDqt7gg6rCSIOyd"
               "tOyDgeyduCBlNS1zbWFsbCDsjI3snbQg7KCV7ZmV7Z6IIEtVUkUtdjEg7IyNIOyImOunjO2BvCDrgqjsirXri4jri6QiLCAi64K0"
               "IOy9lOuTnOqwgCDri6Tsi5wg7KCV7ZWcIOyehOqzhOqwkuydhCDrrLTsl4fsl5Ag66ee7LawIOqzqOuekOuKlOyngCJdLCAi64uk"
               "7IucIOygle2VnCDsnoTqs4TqsJLsl5DshJwgZTUtc21hbGwg7IyNIjogWyLsnb3snYAg6rCS7Jy866GcIOyLpOyImOulvCDsp5rs"
               "irXri4jri6QuIDAg7J2064KYIOydjOyImOuptCDsnoTqs4TqsJLsnYQgMSDrs7Tri6Qg7YGwIOqwkuycvOuhnCDrkZDsl4jqsbDr"
               "gpgg67aA7Zi46rCAIOu2meyXiOyKteuLiOuLpC4g64uk7IucIOygle2VnCDsnoTqs4TqsJIg7J207IOB7J24IGU1LXNtYWxsIOyM"
               "jeydmCDsiJjrpbwg44CM64uk7IucIOygle2VnCDsnoTqs4TqsJLsl5DshJwgZTUtc21hbGwg7IyNID0g7KCV7IiYIOyMjeOAjSDt"
               "lZwg7ZaJ7Jy866GcIOy2nOugpe2VqeuLiOuLpCIsICLsnbQg7IiY64qUIEtVUkUtdjEgMC45MCDsnZgg7IyNIOyImOyZgCDqsJns"
               "lYTslbwg7ZWp64uI64ukLiDrsJjsmKzrprztlZwg7J6E6rOE6rCS7Jy866GcIOuLpOyLnCDshLjrqbQg66qHIOyMjSDri6zrnbzs"
               "p4gg7IiYIOyeiOyWtCwg7KCV66Cs7JeQ7IScIOq6vOuCuCDqsJIg6re464yA66GcIOyFieuLiOuLpCIsICLrgrQg7L2U65Oc6rCA"
               "IOydtCDsjI0g7IiY66W8IOyWtOuKkCDsnoTqs4TqsJIg6rCS7Jy866GcIOyFjOuKlOyngCJdLCAi6rGw7J2YIOqwmeydgCDquIDs"
               "nbQg7J6I64qUIOumrOu3sCI6IFsi7J297J2AIOqwkuycvOuhnCDsi6TsiJjrpbwg7Kea7Iq164uI64ukLiA4OS4yKENQVSnrgpgg"
               "OTQuNyhHUFUp7J2066m0IOyVhOyngSAwLjkwIOyXkOyEnCDshLwg6rCS7J206rOgLCA3LjgoQ1BVKeydtOuCmCAxMS45KEdQVSnr"
               "qbQgMC45NSDsl5DshJwg7IS8IOqwkuydtOupsCwgNjQuMShDUFUp7J2066m0IGU1LXNtYWxsIOy9lOyCrOyduOydmCA5Oeuwseu2"
               "hOychOyImOyXkOyEnCDshLwg6rCS7J6F64uI64ukLiDri6Tsi5wg7KCV7ZWcIGU1LXNtYWxsIOyehOqzhOqwkiDsnbTsg4Hsnbgg"
               "7IyN7JeQIO2VnCDrsogg7J207IOBIOuCmOyYqCDrpqzrt7DsnZgg6rG07IiY7JmALCDqt7gg6rG07IiY66W8IOumrOu3sCDsiJgg"
               "YGxlbihSRVYpYCDroZwg64KY64iIIOu5hOycqOydhCDjgIzqsbDsnZgg6rCZ7J2AIOq4gOydtCDsnojripQg66as67ewID0gayDq"
               "sbQgKHIgJSnjgI0g7ZWcIO2WieycvOuhnCDstpzroKXtlanri4jri6QuIOyxhOygkOydgCDqtITtmLgg7JWI7J2YICUg66W8IOyd"
               "veyKteuLiOuLpCIsICLrtoTrqqjripQg7IyNIOyImOqwgCDslYTri4jrnbwg66as67ewIOyImOyeheuLiOuLpC4g7J6E6rOE6rCS"
               "IOydtOyDgeyduCDsjI3snZgg65GQIOumrOu3sOulvCBib29sIOuwsOyXtOyXkOyEnCBUcnVlIOuhnCDrsJTqvrjqs6AgVHJ1ZSDs"
               "nbgg66as67ew66W8IOyFieuLiOuLpCIsICLrgrQg7L2U65Oc6rCAIOqxsOydmCDqsJnsnYAg6riA7J20IOyeiOuKlCDrpqzrt7Dr"
               "pbwg7Ja064qQIOyehOqzhOqwkuyXkOyEnCwg66y07JeH7Jy866GcIOuCmOuIoCDshYzripTsp4AiXSwgInJlbF9yYXRpbyI6IFsi"
               "65GQIO2WieydmCDsjI0g7IiY66W8IOqwmeydgCBgUkVWYCDCtyDqsJnsnYAg7J6E6rOE6rCSIDAuOTAg7JeQ7IScIOyFjOuKlOyn"
               "gCDrs7TshLjsmpQuIOqwmeydgCDsobDqsbTsnbTrqbQgZTUtc21hbGwg7IyNIOyImOqwgCBLVVJFLXYxIOydmCDsspwg67Cw6rCA"
               "IOuEmOyKteuLiOuLpChDUFUgMTksNjU0IMO3IDEzID0g7JW9IDEsNTEy67CwKSIsICLtlZzsqr0g66qo642466eMIOuLpOuluCDs"
               "noTqs4TqsJLsnbTrgpgg64uk66W4IOumrOu3sOuhnCDshYzsnLzrqbQg7J20IOq0gOqzhOqwgCDquajsp5Hri4jri6QuIOuRkCDr"
               "qqjrjbgg66qo65GQIGBPTERfVEhSRVNIT0xEYCDroZwg7IWJ64uI64ukIiwgIuuCtCDsvZTrk5zqsIAg65GQIOuqqOuNuOydmCDs"
               "jI0g7IiY66W8IOqwmeydgCDrpqzrt7Agwrcg6rCZ7J2AIOyehOqzhOqwkuycvOuhnCDshYzripTsp4AiXSwgInJlbF9zcGxpdCI6"
               "IFsi6rCZ7J2AIDAuOTAg7JeQ7IScIOuRkCDrqqjrjbjsnZgg67OE7KCQIOqwiOumsCDruYTsnKjsnYQg7IWM64qU7KeAIOuztOyE"
               "uOyalC4gZTUtc21hbGwg7J2YIDAuOTAg7J2AIOuzhOygkOydtCDrsJjrjIDsnbgg6riA6rmM7KeAIOustuyWtCBLVVJFLXYxIOuz"
               "tOuLpCDruYTsnKjsnbQg7YG964uI64ukKENQVSAyNC40ICUg64yAIDAuMCAlKS4gZTUtc21hbGwg7Kq97J2EIOuLpOyLnCDsoJXt"
               "lZwg7J6E6rOE6rCS7JeQ7IScIOyEuOuptCDsnbQg6rSA6rOE6rCAIOq5qOynkeuLiOuLpCIsICLrkZAg67mE7JyoIOuqqOuRkCBg"
               "TE9XW2l1WzBdXSAhPSBMT1dbaXVbMV1dYCDrpbwgMC45MCDsnbTsg4HsnLzroZwg7J6E6rOE6rCSIOydtOyDgeyduCDsjI3sl5Dr"
               "p4wg6rOo6528IO2Pieq3oO2VnCDqsJLsnoXri4jri6QiLCAi64K0IOy9lOuTnOqwgCDrkZAg66qo64247J2YIOuzhOygkCDqsIjr"
               "prAg67mE7Jyo7J2EIOqwmeydgCDsnoTqs4TqsJLsl5DshJwg7IWM64qU7KeAIl0sICJyZWxfbmVhciI6IFsi64uk7IucIOygle2V"
               "nCDsnoTqs4TqsJLsnZggZTUtc21hbGwg7IyNIOyImOqwgCBLVVJFLXYxIDAuOTAg7IyNIOyImOyZgCDqsbDsnZgg6rCZ7JWE7JW8"
               "IO2VqeuLiOuLpC4gMC45NSDroZwg64iI64yA7KSR7ZWY66m0IENQVSDsl5DshJwgODjsjI3snbTrnbwgS1VSRS12MSAxM+yMjeyd"
               "mCA267Cw6rCAIOuEmOyKteuLiOuLpC4g7J6E6rOE6rCS7J2EIEtVUkUtdjEg7IyNIOyImOyXkCDrp57strAg6rOo656Q64qU7KeA"
               "IOuztOyEuOyalCIsICLsjI0g7IiY6rCAIO2BrOqyjCDri6TrpbTrqbQg7J6E6rOE6rCS7J2EIOuIiOuMgOykkeydtOuCmCDrsLHr"
               "toTsnITsiJjroZwg6rOg66W4IOqyg+yeheuLiOuLpC4g7KCV66Cs7ZWcIGU1LXNtYWxsIOy9lOyCrOyduOyXkOyEnCBLVVJFLXYx"
               "IOyMjSDsiJgg67KI7Ke4IOqwkuydhCDslIHri4jri6QiLCAi64K0IOy9lOuTnOqwgCDri6Tsi5wg7KCV7ZWcIOyehOqzhOqwkuyd"
               "hCBLVVJFLXYxIOydmCDsjI0g7IiY7JeQIOunnuy3hOuKlOyngCJdLCAicmVsX3BjdCI6IFsi44CM6rGw7J2YIOqwmeydgCDquIDs"
               "nbQg7J6I64qUIOumrOu3sCA9IGsg6rG0IChyICUp44CNIOyXkOyEnCByIOydtCBrIMO3IOumrOu3sCDsiJggw5cgMTAwIOqzvCDq"
               "sJnslYTslbwg7ZWp64uI64ukLiDrqqjrk6Ag7IyN7J2YIOyImChDUFUgNDk5LDUwMCnroZwg64KY64iE66m0IDAuMCAlIOyymOuf"
               "vCDsnpHsnYAgJSDqsIAg64KY7Ji164uI64ukIiwgIuu2hOuqqOuKlCDrpqzrt7Ag7IiYIGBsZW4oUkVWKWAg7J6F64uI64ukIiwg"
               "IuuCtCDsvZTrk5zqsIAg6rGw7J2YIOqwmeydgCDquIAg67mE7Jyo7J2YIOu2hOuqqOuhnCDrrLTsl4fsnYQg7I2864qU7KeAIl0s"
               "ICJmb3JiaWQiOiBbIuyghOyehCDri7Tri7nsnpDsnZgg44CM7YyQ64uoIDog4oCm44CNIO2WieydtCDstpzroKXsl5Ag64Ko7JWE"
               "IOyeiOyKteuLiOuLpC4g7J20IO2MkOuLqOydgCBlNS1zbWFsbCDsl5Ag66ee7KeAIOyViuuKlCDsnoTqs4TqsJLsnLzroZwg7IS8"
               "IOu5hOycqOyXkOyEnCDrgpjsmKgg66y47J6l7J206528IOyngOybgeuLiOuLpCIsICLsoITsnoQg7L2U65Oc7JeQ7IScIOyngOyb"
               "jOyVvCDtlaAg7ZaJ7J20IOy2nOugpeyXkCDrgqjslYQg7J6I64qU7KeAIl0sICJwbG90IjogWyLqt7jrnpjtlITrpbwg7J20IOyF"
               "gOyXkOyEnCDqt7jroLjripTsp4AsIOqwgOuhnOy2leqzvCDshLjroZzstpXsl5Ag7J2066aEKOOAjOyehOqzhOqwkuOAjcK344CM"
               "7J6E6rOE6rCSIOydtOyDgeyduCDsjI0g7IiY44CNKeydhCDri6zslZjripTsp4Ag67O07IS47JqULiDqt7jrnpjtlITripQg6re4"
               "66a87J20IOyeiOuKlOyngOyZgCDrkZAg7LaV7J2YIOydtOumhOunjCDssYTsoJDtlanri4jri6QiLCAiYGF4LnNldF94bGFiZWwo"
               "XCLsnoTqs4TqsJJcIilgIOqzvCBgYXguc2V0X3lsYWJlbChcIuyehOqzhOqwkiDsnbTsg4Hsnbgg7IyNIOyImFwiKWAg7J2EIOq3"
               "uOumrOuKlCDsvZTrk5zsl5Ag64Sj7Jy865286rOgIFBST01QVCDsl5Ag7KCB7Jy87IS47JqULiDshKDsnZgg7IiYIMK3IOuhnOq3"
               "uCDsiqTsvIDsnbwgwrcg7IS466Gc7ISgIMK3IOuylOuhgOuKlCDssYTsoJDtlZjsp4Ag7JWK7Iq164uI64ukIiwgIuuCtCDqt7jr"
               "npjtlITsnZgg6rCA66Gc7LaVwrfshLjroZzstpUg7J2066aE7J20IOyWtOuUlOyEnCDruaDsoYzripTsp4AiXSwgImVycm9yIjog"
               "WyJOYW1lRXJyb3Ig66m0IOuwlOuhnCDsnIQg44CM66+47IWYIDIg7KSA67mEIOyFgOOAjeydhCDrqLzsoIAg7Iuk7ZaJ7ZaI64qU"
               "7KeA67aA7YSwIO2ZleyduO2VmOyEuOyalC4g7Iuk7ZaJ7ZaI64qU642w64+EIOyYpOulmOqwgCDrgpjrqbQg7Jik66WYIOusuOye"
               "pSDssqsg7ZaJ7J2EIOq3uOuMgOuhnCBBSSDtipzthLDsl5Ag67O07JesIOyjvOqzoCwg44CM66+47IWYIDIg7KSA67mEIOyFgOOA"
               "jeydtCDrp4zrk6Ag7J2066aEKFJFViDCtyBMT1cgwrcgRU5DT0RFUlMgwrcgZW1iZWQgwrcgT0xEX1RIUkVTSE9MRCDCtyBOX1JF"
               "ViDCtyBERVZJQ0Ugwrcgcm5kIMK3IENPTCDCtyBGSUdfVynsnYAg64uk7IucIOunjOuTpOyngCDrp5Dqs6Ag6riA7J6QIOq3uOuM"
               "gOuhnCDsk7Drnbzqs6Ag7KCB7Jy87IS47JqULiDrqqjrjbgg7J2066aE7J2AIGBFTkNPREVSU2Ag7J2YIO2CpCDjgIxLVVJFLXYx"
               "44CNIOOAjGU1LXNtYWxs44CNIOq3uOuMgOuhnCBgZW1iZWRgIOyXkCDrhJjquYHri4jri6QiLCAi7J20IOyYpOulmOqwgCDslrTr"
               "ipAg7ZaJ7JeQ7IScIOyZnCDrgqzripTsp4AiXX19")


_ASK = {"m1": "full", "m2": "blank"}   # 미션 2 는 함정이라 빈칸을 내가 채우는 질문 틀입니다


_ASK_HEAD = {"full": "◼ 튜터에 그대로 보낼 질문 (코드는 고치지 말라고 분명히 적습니다)",
             "blank": "◼ 튜터에 보낼 질문 (빈칸은 내가 채웁니다)"}


_ASK_FORM = {"full": "내 PROMPT 는 「{prompt}」 였는데 「{name}」 채점이 ❌ 였습니다."
                     " AI 코드 표시선 아래 코드를 보고 「{ask}」 에 3~5문장으로 답해 주세요. 코드는 고치지 마세요.",
             "blank": "내 PROMPT 는 ____ 였는데 「{name}」 채점이 ❌ 였습니다. AI 코드 표시선 아래 코드에서"
                      " 「{ask}」 부터 짚고, 어느 행을 무엇으로 고칠지 한 문장으로 답해 주세요."}


def _blocks(key, reason):
    """PROMPT 칸이 비었을 때(reason) 값·그래프 채점 전에 잠그는가.

    함정 미션(m2)은 PROMPT 칸이 비어도 잠그지 않고 표시선 아래 전임 담당자 코드를 진단합니다
    """
    if reason and key != "m2":
        return True
    return False


import numpy as np                        # 미션 1 의 다시 세기가 쓴다


# ── 이 실습에만 있는 채점 — 값끼리의 관계(rels) · 준비 셀의 배열로 다시 세기(custom m1)
#    값(values) · 실행 경로별 기준(by_device) · 그래프(plot)는 아래 공통 채점 코드가 보고, 여기 항목은 `_BOOK["checks"]` 로 이어 붙습니다.
_NLP3_PRE = {"spy": None, "calls": []}     # 미션 1 — 이번 실행의 encode 기록
_NLP3_NUM = r"(?<![\w.])(-?\d[\d,]*(?:\.\d+)?)"     # 숫자 하나(쉼표 · 소수) — 건수와 비율을 읽을 때 씁니다
_NLP3_PAREN = r"\([^()]*\)|\[[^\[\]]*\]"            # 괄호 설명


def _nlp3_unspy():
    """`_nlp3_pre` 가 감싼 encode 를 되돌립니다"""
    saved, _NLP3_PRE["spy"] = _NLP3_PRE.get("spy"), None
    if not saved:
        return
    cls, mine, orig = saved
    try:
        if vars(cls).get("encode") is mine:
            cls.encode = orig
    except Exception:
        pass


def _nlp3_pre(tag):
    """셀 실행 전 — 미션 1 의 PROMPT 셀이면 문장 인코더의 encode 가 받은 글과 돌려준 벡터를 적어 둡니다.

    벡터를 이름(AIR_EMB · q)으로 찾으면 앞 실행이 남긴 벡터로 판정하거나(이름을 바꿔 다시 쓴 맞는 코드가 ❌), 이름이 다르면
    다시 세기가 통째로 빠집니다(틀린 건수가 통과). 그래서 이 셀이 도는 동안 encode 가 실제로 받은 글과 돌려준 벡터를 적습니다.
    다음 셀이 돌기 전과 채점이 끝난 뒤에 원래 encode 로 되돌립니다. 코드에 encode 가 없으면 감싸지 않습니다.
    """
    _nlp3_unspy()
    _NLP3_PRE["calls"] = []
    body = str(_STATE.get("src") or "").split("여기부터 AI 코드", 1)[-1]
    if tag != "m1" or not _re.search(r"encode|SentenceTransformer", body):
        return
    try:
        import sentence_transformers as _st
        cls = _st.SentenceTransformer
        orig, calls = cls.encode, _NLP3_PRE["calls"]

        def encode(self, sentences, *a, **k):
            out = orig(self, sentences, *a, **k)
            try:                           # 앞머리를 prompt 인자로 준 호출도 「앞머리 + 글」 로 적습니다
                lead = k.get("prompt")
                name = k.get("prompt_name") or getattr(self, "default_prompt_name", None)
                if lead is None and name:
                    lead = (getattr(self, "prompts", None) or {}).get(name)
                one = isinstance(sentences, str)
                calls.append(((sentences,) if one else tuple(str(x) for x in sentences), str(lead or ""), out, one))
            except Exception:
                pass
            return out

        cls.encode = encode
        _NLP3_PRE["spy"] = (cls, encode, orig)
    except Exception:
        pass


def _nlp3_vec(out):
    """encode 가 돌려준 값 → 행마다 벡터 하나인 float32 배열(못 바꾸면 None)"""
    try:
        if hasattr(out, "detach"):
            out = out.detach().cpu().numpy()
        elif isinstance(out, (list, tuple)) and len(out) and hasattr(out[0], "detach"):
            out = [v.detach().cpu().numpy() for v in out]
        arr = np.asarray(out, dtype=np.float32)
        return arr.reshape(1, -1) if arr.ndim == 1 else arr if arr.ndim == 2 else None
    except Exception:
        return None


def _nlp3_unit(v):
    """길이 1 로 맞춘 벡터(행렬이면 행마다)"""
    v = np.asarray(v, dtype=np.float32)
    n = np.linalg.norm(v, axis=-1, keepdims=True)
    return v / np.where(n > 0, n, 1.0)


def _nlp3_encoded():
    """이번 실행에서 인코딩한 민원 벡터와 질의 벡터 — {"E", "q", "bare_titles", "bare_first", "bare_query"} · 다 찾지 못하면 None.

    encode 가 받은 글마다 벡터를 적어 두었다가 AIR 의 제목 차례로 다시 모읍니다 — 벡터 이름을 바꿔도(emb · qv), 제목을 묶음으로
    나눠 인코딩해도, 셀 끝에서 모델을 지워도 같습니다. 글은 「PREFIX + 글」 을 먼저, 없으면 앞머리 없는 글을 찾아 앞머리를
    두었는지도 함께 돌려줍니다.
    """
    air, query, prefix = _get("AIR"), _get("QUERY"), str(_get("PREFIX") or "")
    calls = _NLP3_PRE.get("calls") or []
    if air is None or query is None or not calls:
        return None
    try:
        seen = {}
        for sents, lead, out, _one in calls:
            arr = _nlp3_vec(out)
            if arr is None or arr.shape[0] != len(sents):
                continue
            for i, text in enumerate(sents):
                seen[lead + text] = arr[i]
        rows, bare = [], []
        for t in (str(x) for x in air["사건제목"]):
            v = seen.get(prefix + t)
            if v is None:
                v = seen.get(t)
                bare.append(len(rows))
            if v is None:
                return None
            rows.append(v)
        qv, bare_q = seen.get(prefix + str(query)), False
        if qv is None:
            qv, bare_q = seen.get(str(query)), True
        if qv is None or not rows:
            return None
        return {"E": _nlp3_unit(np.vstack(rows)), "q": _nlp3_unit(qv).reshape(-1), "bare_titles": len(bare),
                "bare_first": _bi.bool(bare) and bare[0] == 0, "bare_query": bare_q}
    except Exception:
        return None


def _nlp3_bare(text):
    """괄호 설명을 걷은 글"""
    return _re.sub(_NLP3_PAREN, " ", str(text or "").replace("−", "-"))


def _nlp3_count_pct(tail):
    """「k 건 (p %)」 줄의 건수와 비율 — 건수는 % 가 붙지 않은 첫 수(괄호 밖 먼저), 비율은 % 가 붙은 첫 수입니다.

    「92 건 (3.1 %)」 · 「92 건 (리뷰 3,000건 가운데 3.1 %)」 · 「92 건 / 3,000 건 (3.1 %)」 를 같게 읽습니다.
    % 가 붙은 수가 없으면 괄호 밖의 수가 꼭 둘일 때만 둘째 수를 비율로 봅니다(수가 여럿인 다른 줄을 값 줄로 읽지 않게).
    두 수를 못 읽으면 None.
    """
    full = str(tail or "").replace("−", "-").replace(",", "")
    outer = _nlp3_bare(tail).replace(",", "")
    pct = [float(m.group(1)) for m in _re.finditer(_NLP3_NUM + r"\s*%", full)]
    plain_out = [float(m.group(1)) for m in _re.finditer(_NLP3_NUM + r"(?!\s*%)(?![\d.])", outer)]
    plain_all = [float(m.group(1)) for m in _re.finditer(_NLP3_NUM + r"(?!\s*%)(?![\d.])", full)]
    k = plain_out[0] if plain_out else plain_all[0] if plain_all else None
    p = pct[0] if pct else plain_out[1] if len(plain_out) == 2 else None
    return None if k is None or p is None else (k, p)


def _rel_check(rel, ctx):
    """값 둘의 관계 하나 — (통과 여부, 설명) · 이름이 있는 줄이 없으면 None (그 값의 ❌ 가 이미 찍혔습니다)

    값은 값 칸에서 읽은 값을 씁니다. 「k 건 (r %)」 꼴의 건수만 이름 줄의 표지 뒤 글에서 읽습니다(비율은 값 칸이 읽은 값).
    """
    def _first(label):                                  # 값 칸에서 읽은 값 — 값 칸이 아닌 이름은 정본의 읽기로
        got = ctx["read"](label)
        return ctx["values"].get(label, got[0] if got else None)

    a, a_tails = _first(rel["a"]), ctx["tails"](rel["a"])
    b = _first(rel["b"]) if rel.get("b") else 1.0
    if a is None or b is None:
        return None
    kind = rel["kind"]
    if kind == "ratio_min":
        return (b > 0 and a / b >= rel["min"]), (f"{rel['a']} ÷ {rel['b']} = {a / b:,.4g}" if b else f"{rel['b']} 가 0 이다")
    if kind == "gt":
        return a > b, f"{rel['a']} {a:g} · {rel['b']} {b:g}"
    if kind == "near":
        return (b > 0 and _bi.abs(a / b - 1) <= rel["tol"]), f"{rel['a']} {a:,.6g} · {rel['b']} {b:,.6g}"
    if kind == "pct_of_n":
        n_all = globals().get(rel.get("n_name", "N_REV"))
        if not n_all:
            return None
        k_p = _nlp3_count_pct(a_tails[0]) if a_tails else None
        if k_p is None:
            return False, "건수와 비율 가운데 하나만 있다 — 「k 건 (r %)」 꼴로 한 행에 함께 적는다"
        k, r = k_p[0], a                                 # 비율은 값 칸이 읽은 값 — 같은 줄을 두 가지로 읽지 않습니다
        return _bi.abs(100 * k / n_all - r) <= 0.06, f"{k:,.0f} ÷ {n_all:,} × 100 = {100 * k / n_all:.2f}, 출력한 비율 {r:g} %"
    return None


def _nlp3_cap(spec, ctx):
    """미션 2 — 거의 같은 글이 있는 리뷰 건수는 다시 정한 임계값의 쌍 수 × 2 를 넘을 수 없습니다(한 쌍은 리뷰 2건).

    다시 정한 임계값이 아닌 0.95 · 0.90 에서 센 건수(78 건 · 892 건)는 비율이 기준 범위 안이어도 여기서 걸립니다.
    """
    rels = {rel["kind"]: rel for rel in spec.get("rels", ())}
    near, pct = rels.get("near"), rels.get("pct_of_n")
    if not near or not pct:
        return []
    pairs, tails = ctx["values"].get(near["a"]), ctx["tails"](pct["a"])
    k_p = _nlp3_count_pct(tails[0]) if tails else None
    if pairs is None or k_p is None or pairs < 0 or k_p[0] <= 2 * pairs:
        return []
    return [("거의 같은 글 건수와 쌍 수", False,
             f"거의 같은 글이 있는 리뷰 {k_p[0]:,.0f} 건이 「{near['a']}」 {pairs:,.0f} 쌍 × 2 = {2 * pairs:,.0f} 건보다 많습니다."
             " 한 쌍은 리뷰 2건이라 이 수를 넘을 수 없습니다 — 건수를 다시 정한 임계값이 아닌 다른 임계값에서 셌는지 봅니다",
             pct["a"])]


def _nlp3_recount(E, qq, got, air, where):
    """민원 벡터와 질의 벡터로 이웃 20건을 다시 골라 「환불」 없는 민원 수를 대조합니다 — (통과 여부, 이름, 설명, 힌트 키) 목록"""
    key, top_k = str(_get("KEYWORD") or "환불"), int(_get("TOP_K") or 20)
    cos = E @ qq
    nouns = air["명사"].reset_index(drop=True)
    titles = air["사건제목"].reset_index(drop=True)

    def _no_kw(idx):
        return int(_bi.sum(key not in nouns.iloc[i] for i in idx))

    top, low = np.argsort(-cos)[:top_k], np.argsort(cos)[:top_k]
    right = _no_kw(top)
    if got == right:
        return [(True, "이웃 20건 다시 세기", f"{where} AIR_EMB @ q 의 큰 순서 {top_k}건에서 다시 세어도 {right} 건입니다", "recount")]
    why = ""
    if got == _no_kw(low):
        why = " — 출력한 값은 코사인이 작은 순서로 고른 20건의 값입니다"
    elif got == top_k - right:
        why = " — 출력한 값은 「환불」이 있는 민원을 셌을 때의 값입니다"
    elif got == int(_bi.sum(key not in str(titles.iloc[i]).split() for i in top)):
        why = " — 출력한 값은 제목을 공백으로 자른 어절에서 「환불」을 찾았을 때의 값입니다"
    return [(False, "이웃 20건 다시 세기",
             f"내 코드의 AIR_EMB @ q 로 큰 순서 {top_k}건을 다시 고르면 `명사` 에 「{key}」이 없는 민원은 {right} 건인데"
             f" 출력은 {got:g} 건입니다{why}", "recount")]


def _m1_prefix():
    """미션 1 — 이름 공간의 `q` 와 `AIR_EMB[0]` 이 「PREFIX + 글」을 인코딩한 벡터인지 봅니다(encode 기록이 없을 때만 씁니다).

    내 코드가 불러 둔 문장 인코더로 두 글을 앞머리 있음 · 없음으로 한 번씩 인코딩해 코사인으로 비교합니다. 인코더를 찾지 못했거나
    두 벡터 어느 쪽과도 맞지 않으면(다른 처리를 한 벡터) 이 대조는 건너뜁니다.
    """
    prefix = str(_get("PREFIX") or "")
    emb, qv, air, query = _get("AIR_EMB"), _get("q"), _get("AIR"), _get("QUERY")
    if not prefix or emb is None or qv is None or air is None or query is None:
        return []
    encs = [v for k, v in globals().items() if not k.startswith("_") and type(v).__name__ == "SentenceTransformer"]
    try:
        title = str(air["사건제목"].iloc[0])
        mine = {"질의 q": np.asarray(qv, dtype=np.float32).reshape(-1),
                "첫 제목 AIR_EMB[0]": np.asarray(emb, dtype=np.float32)[0].reshape(-1)}
        for enc in encs:
            V = np.asarray(enc.encode([prefix + query, query, prefix + title, title], normalize_embeddings=True,
                                      convert_to_numpy=True, show_progress_bar=False), dtype=np.float32)
            if V.shape[1] != mine["질의 q"].size:
                continue
            ref = {"질의 q": (V[0], V[1]), "첫 제목 AIR_EMB[0]": (V[2], V[3])}
            bare, ok_all = [], True
            for name, x in mine.items():
                x = x / (float(np.linalg.norm(x)) or 1.0)
                with_p, without_p = float(x @ ref[name][0]), float(x @ ref[name][1])
                if with_p >= 0.999:
                    continue
                ok_all = False
                if without_p >= 0.999:
                    bare.append(name)
            if ok_all or bare:
                return _nlp3_prefix_rows(prefix, bare)
        return []
    except Exception:
        return []


def _nlp3_prefix_rows(prefix, bare):
    """앞머리 대조의 채점 줄 — bare 는 앞머리 없이 인코딩한 쪽의 이름 목록"""
    if not bare:
        return [(True, "접두어 PREFIX", f"질의와 첫 제목 모두 접두어 「{prefix.strip()}」 를 더해 인코딩한 벡터입니다", "prefix")]
    return [(False, "접두어 PREFIX",
             f"{' · '.join(bare)} 가 접두어 「{prefix.strip()}」 없이 인코딩한 벡터와 같습니다."
             " e5-small 은 질의와 제목 모두 앞에 같은 `PREFIX` 를 더합니다", "prefix")]


def _m1_checks(ctx):
    """미션 1 의 다시 세기와 앞머리 대조 — 이번 실행에서 encode 가 받은 글과 돌려준 벡터로 봅니다.

    encode 기록으로 민원 벡터와 질의 벡터를 다 모으지 못했으면(제목을 고쳐서 인코딩한 코드, 인코딩 없이 앞 실행의 벡터를
    다시 쓴 코드) 이름 공간의 `AIR_EMB` · `q` 로 봅니다. 그것도 없으면 다시 세지 못했다고 ❌ 로 알립니다 — 값 범위(1～20)만으로는
    작은 순서 · 「환불」 있는 쪽 · 어절로 센 값이 걸러지지 않습니다.
    """
    rec = _nlp3_encoded()
    _nlp3_unspy()
    _NLP3_PRE["calls"] = []
    air, prefix = _get("AIR"), str(_get("PREFIX") or "")
    got = ctx["values"].get("이웃 20건 중 「환불」 없는 민원")
    if air is None:
        return []
    out = []
    try:
        if rec is not None:
            if got is not None:
                out += _nlp3_recount(rec["E"], rec["q"], got, air, "이 셀이 인코딩한 벡터로")
            if prefix:
                bare = (["질의 q"] if rec["bare_query"] else []) + (
                    ["첫 제목 AIR_EMB[0]"] if rec["bare_first"] else [f"민원 제목 {rec['bare_titles']:,}건"] if rec["bare_titles"] else [])
                out += _nlp3_prefix_rows(prefix, bare)
            return out
        emb, qv = _get("AIR_EMB"), _get("q")
        E = np.asarray(emb, dtype=np.float32) if emb is not None else None
        qq = np.asarray(qv, dtype=np.float32).reshape(-1) if qv is not None else None
        if E is None or qq is None or E.ndim != 2 or E.shape[0] != len(air) or E.shape[1] != qq.size:
            if got is None:
                return []
            return [(False, "이웃 20건 다시 세기",
                     "이 셀이 인코딩한 민원 벡터와 질의 벡터를 찾지 못해 다시 세지 못했습니다. `AIR` 의 사건제목과 `QUERY` 를"
                     " SentenceTransformer 의 `encode` 로 인코딩하고, 벡터는 `AIR_EMB` · `q` 에 두세요", "recount")]
        if got is not None:
            out += _nlp3_recount(E, qq, got, air, "이름 공간의")
        return out + _m1_prefix()
    except Exception:
        return out


def _nlp3_spec(key, spec):
    """이 세션의 기준 — 「k 건 (r %)」 꼴로 출력하는 값(rels 의 pct_of_n 이 가리키는 이름)은 괄호 안의 % 수를 읽게 켭니다.

    함정 미션은 이름 줄이 없는 칸에 출력 꼴 힌트만 냅니다(`form_only`) — 이 권의 값 힌트 첫 단은 읽은 값을 전제로 쓴 글이고
    기준값을 예로 들어, 전임 코드를 그대로 ▶ 한 첫 화면에 내면 PROMPT 를 쓰기 전에 답이 보입니다.
    """
    out = dict(spec)
    own = [rel["a"] for rel in spec.get("rels", ()) if rel.get("kind") == "pct_of_n" and rel.get("a") in spec.get("values", {})]
    if own:
        out["paren_unit"] = tuple(own)
    if not _blocks(key, "PROMPT 입력란이 비어 있습니다"):
        out["form_only"] = True
    return out


def _nlp3_checks(key, spec, ctx):
    """값·그림 밖의 채점 — 값끼리의 관계와 준비 셀의 배열로 다시 센 값. 출력 글 · 읽은 값 · ❌ 횟수는 ctx 로 받습니다"""
    out = []
    for rel in spec.get("rels", ()):               # 값끼리의 관계 — DEVICE 마다 값이 달라도 관계는 같습니다
        got_rel = _rel_check(rel, ctx)
        if got_rel is None:
            continue
        ok_rel, note = got_rel
        out.append((rel["name"], ok_rel, note + ("" if ok_rel or not ctx["attempt"] else f" (기준: {rel['rule']})"), rel["hint"]))
    out += _nlp3_cap(spec, ctx)                    # 거의 같은 글 건수의 상한(쌍 수 × 2)
    custom = spec.get("custom")
    rows = _m1_checks(ctx) if custom == "m1" else ()
    for row in rows:                               # 준비 셀의 배열로 다시 센 값과 맞는지
        out.append((row[1], row[0], row[2], row[3]) + tuple(row[4:5]))
    return out


# 권 설정 · pre: 미션 1 의 encode 기록 · spec: 「k 건 (r %)」 꼴로 찍는 칸, 함정 미션의 첫 화면 힌트 · checks: 값끼리의 관계, 다시 세기
_BOOK = {"pre": _nlp3_pre, "spec": _nlp3_spec, "checks": (_nlp3_checks,)}
# ━━ 채점 기준 끝 — 아래는 모든 실습이 함께 쓰는 채점 코드입니다 ━━


# ───────────────────────── 출력에서 값 읽기 ─────────────────────────
#  숫자 하나 — 천 단위 쉼표(8,760) · 지수 표기 · 부호. 영문·숫자·밑줄·점 바로 뒤에 붙은 숫자는
#  이름의 일부라 읽지 않습니다 — 「float64」·「int64」 의 64, 「C23」 의 23, 「v2.3」 의 3. 영문에 붙임표로
#  이은 숫자도 이름입니다 — 「Chronos-2」 의 2, 「t-1」 의 1, 「top-3」 의 3.
_NUM = (r"(?<![A-Za-z_\d.])(?<![A-Za-z]-)[-+]?(?:\d{1,3}(?:,\d{3})+(?!\d)|\d+)(?:\.\d*)?(?:[eE][-+]?\d+)?")
_NUM_RE = _re.compile(_NUM)
_KW_RE = _re.compile(r"[A-Za-z_][A-Za-z0-9_]*=(?!=)")       # 「period=24」 같은 인자 설명
_KW_VAL = _re.compile(r"\s*(?:'[^']*'|\"[^\"]*\"|\([^()]*\)|\[[^\[\]]*\]|[^\s,;)\]]+)")
#  Series 를 통째로 찍은 출력의 끝 줄 — 「Name: coef, dtype: float64」 · 「Freq: h, Name: cnt, Length: 168, dtype: float64」 ·
#  「dtype: int64」. f-string 안에 Series 를 넣으면 뒤의 글자(단위)가 이 줄 끝에 붙으므로(「… dtype: float64 대/도」)
#  dtype 뒤의 꼬리 글자는 가리지 않습니다.
_SERIES_RE = _re.compile(r"^(?:(?:Name|Length|Freq): .*, )?dtype: [A-Za-z]\w*", _re.M)
_GLYPH = {**{c: "-" for c in "—–―‒‐‑-−"}, **{c: "·" for c in "·ㆍ•‧・∙⋅"}}   # 겉보기가 같은 대시·가운뎃점은 같게 봅니다
_SEP = set(",;|") | set("·ㆍ•‧・∙⋅") | set("—–―")          # 값 뒤에서 다른 말로 넘어가는 자리


def _num_tokens(text):
    """출력 텍스트의 숫자 — (찍힌 글자, 값) 목록 (유니코드 마이너스 − 는 - 로 읽습니다)"""
    picked = []
    for m in _NUM_RE.finditer((text or "").replace("−", "-")):
        try:
            picked.append((m.group(0), float(m.group(0).replace(",", ""))))
        except ValueError:
            continue
    return picked


def _nums(text):
    """출력 텍스트에서 숫자만 뽑습니다"""
    return [v for _, v in _num_tokens(text)]


def _unwrap_np(txt):
    """numpy·pandas 의 값 표기를 숫자로 폅니다 — 화면에는 맞는 수가 보이는데 채점이 못 읽는 일이 없게.

    np.float64(40.2) · np.int64(3) · float64(40.2) → 40.2 · 3 (numpy 2 가 dict·list 출력에 남기는 꼴),
    array([40.2]) · [40.2] · [80.] → 40.2 · 80. (원소 하나인 배열 — `.values` 뒤 `[0]` 을 빠뜨린 꼴).
    원소가 둘 이상인 배열은 펴지 않습니다 — 어느 값인지 모릅니다(채점 표가 「숫자 하나로」 를 안내합니다).
    """
    txt = _re.sub(r"\b(?:np|numpy)\.[A-Za-z_]\w*\(\s*([^()]*?)\s*\)", r"\1", txt or "")
    txt = _re.sub(r"\b(?:float|int|uint|complex)(?:8|16|32|64|128)\(\s*([^()]*?)\s*\)", r"\1", txt)
    txt = _re.sub(r"\barray\(\s*(\[[^\[\]()]*\])(?:\s*,\s*dtype=[\w.]+)?\s*\)", r"\1", txt)
    return _re.sub(r"\[\s*([-−+]?\d[\d,]*\.?\d*(?:[eE][-+]?\d+)?)\s*\]", r" \1 ", txt)


def _bare_gap(gap):
    """이름과 값(또는 「=」) 사이가 따옴표·굵게 표시(**)·표의 세로줄·빈칸·괄호 설명·줄표뿐인가 — 그러면 그 이름의 값 줄입니다"""
    gap = _re.sub(r"\([^()]*\)|\[[^\[\]]*\]", "", gap or "")
    return not _re.sub(r"[\s'\"`*」』”’|│┃—–―\-:=,·：＝]", "", gap)


def _paren_open(t, unit):
    """괄호 밖에 숫자가 하나도 없는 꼬리에서 괄호 안의 값 하나만 드러낸 글 — 그런 꼴이 아니면 None.

    「명사 + 용언 1위 = 하다 (18.74 %)」 처럼 단어 뒤 괄호에 비율을 넣거나 `print("이름 =", (단어, 비율))` 로 튜플을 그대로
    찍으면 값이 괄호 안에만 있습니다. 괄호 안은 설명으로 보고 건너뛰므로 맞는 출력이 「숫자를 찾지 못했습니다」 가 됩니다.
    괄호 밖에서 값을 하나도 못 읽었을 때만 봅니다 — 읽은 값이 있으면 그 값이 먼저입니다(「확신 = 26 (전체 168 시간 가운데)」).

    - 그릇: 값 표지(또는 이름) 바로 뒤가 여는 괄호이고, 안이 쉼표로 가른 단어(숫자 없는 글)와 숫자(짧은 단위까지)뿐이며
      숫자가 하나면 그 숫자입니다 — `('하다', 18.74)` · `(하다, 18.74 %)` · `(18.74 %)`. 숫자가 둘 이상인 그릇은 어느 값인지
      모르므로 읽지 않습니다(`[0.57 0.09]` · `(1, 0, 1)`).
    - 기준 단위: 괄호 안에 기준 단위가 바로 붙은 수가 있으면 그 수입니다 — 「하다 (18.74 %)」 · 「하다(18.74%)」 ·
      「하다 (문서 비율 18.74 %)」. 단위가 없는 칸의 괄호 안 숫자(「(기준: 2 미만이면 비슷)」)는 지금처럼 설명입니다.
    """
    groups = _spans(t)
    if not groups:
        return None
    pick = None
    a, b = groups[0]
    before = t[:a].rstrip()
    if not before or before[-1] in "=:→：＝":            # 그릇 — 표지(또는 이름) 바로 뒤의 괄호
        at, spots, clean = a + 1, [], True
        for piece in t[a + 1:b].split(","):
            body = _re.sub(r"[()\[\]'\"`\s]", "", piece)
            m = _NUM_RE.match(body)
            if m and len(body) - m.end() <= 4 and not _re.search(r"\d", body[m.end():]):
                spots.append((at, at + len(piece)))        # 숫자 하나(짧은 단위까지)
            elif _re.search(r"[\d=:→：＝]", body):
                clean = False                              # 숫자가 섞인 말 · 다른 표지 — 그릇이 아니라 설명입니다
            at += len(piece) + 1
        if clean and len(spots) == 1:
            m = _NUM_RE.search(t, spots[0][0], spots[0][1])
            end = spots[0][1]
            while m and end > m.end() and t[end - 1] in ")]'\"` \t":
                end -= 1
            pick = (m.start(), end) if m else None
    if pick is None and unit:
        for a, b in groups:
            for m in _NUM_RE.finditer(t, a + 1, b):
                rest = t[m.end():b]
                gap = len(rest) - len(rest.lstrip())
                if rest[gap:].startswith(unit) or (unit.startswith("%") and rest[gap:].startswith("%")):
                    pick = (m.start(), m.end() + gap + (len(unit) if rest[gap:].startswith(unit) else 1))
                    break
            if pick is not None:
                break
    if pick is None:
        return None
    hide = {i for a, b in groups for i in range(a, _bi.min(b, len(t) - 1) + 1)} - set(range(pick[0], pick[1]))
    return "".join(" " if i in hide else ch for i, ch in enumerate(t))


def _read_value(tail, series=False, inner=0, unit=""):
    """이름 뒤 꼬리에서 그 이름의 값 하나 — (값, 찍힌 글자, 값 뒤 글자, 이름 바로 뒤 값인가, 꼴, 읽은 길) · 없으면 None.

    꼴은 "plain"·"series"(Series 의 인덱스를 건너뛴 값)·"kw"(인자 설명의 값), 읽은 길은 "marker"(값 표지 뒤)·
    "bare"(표지 없이 이름 뒤 첫 숫자)·"kw" 입니다.

    - 괄호·대괄호 안(설명 · 「(기준: 2 미만이면 비슷)」 · 「(m=7)」)은 건너뜁니다.
    - 「period=24, robust=True」 같은 인자 설명은 값이 아닙니다. 그 뒤의 「→」·「=」 뒤를 읽고, 인자 설명뿐이면 마지막 인자 값입니다.
    - 값 표지(「=」·「:」·「→」, 전각 「＝」·「：」)가 숫자보다 먼저 나오면 그 뒤 첫 숫자가 값입니다. 시각 23:00 의 「:」 는 표지가 아닙니다.
    - 숫자가 표지보다 먼저 나오면 — 그 숫자와 표지 사이에 쉼표·가운뎃점·줄표가 있으면 그 숫자가 값이고
      (「기온 계수 -40.20 대/도, 크기 = 40.20」 의 -40.20), 없으면 숫자는 이름의 설명이고 표지 뒤가 값입니다
      (「…, 원점 5 11-16 23:00 까지 = 385.06」 의 385.06). 「→」 는 앞에 숫자가 있으면 해석으로 넘어가는 자리입니다.
      이름 바로 뒤(빈칸·괄호 설명만 사이에 둔)의 소수는 구분 기호가 없어도 값입니다 — 「이름 4.837 대 기준값 = 0.271」 의
      4.837. 이름 바로 뒤의 정수(「이름 2개 원점 평균 = 0.271」 의 2)는 개수를 적은 설명일 수 있어 표지 뒤를 읽습니다.
    - series=True(출력에 「dtype:」 행이 있음)면 「= 24    277.46」 의 인덱스 24 대신 값 277.46 을 읽습니다.
    - inner 는 이름이 괄호 안에서 잡혔을 때의 괄호 깊이입니다 — 그 괄호가 닫히기 전의 숫자는 읽지 않습니다.
    - 괄호 밖에서 값을 하나도 못 읽었으면 괄호 안의 값 하나를 읽습니다 — 값이 그릇(튜플)에 담겼거나, 괄호 안에 기준 단위
      (unit)가 붙은 수가 있을 때입니다(`_paren_open`). 이렇게 읽은 줄은 약한 줄입니다(`_read_paren` · `_value_use`, 0.4.2).
    """
    return _read_paren(tail, series, inner, unit)[0]


def _read_paren(tail, series=False, inner=0, unit=""):
    """`_read_value` 의 값과 그 값을 괄호 안에서만 읽었는가 — (값 또는 None, 괄호 안에서만 읽음).

    괄호 안에서만 읽은 값은 약한 줄의 값입니다 — 「이름 (제로샷 41.6 % 보다 높음)」 처럼 덧붙인 설명 줄의 수가 맨 이름
    「이름 = 값」 줄과 부딪치지 않게, 다른 값 줄이 있으면 `_value_use` 가 쓰지 않습니다(그런 줄뿐이면 0.4 의 53번대로 읽습니다).
    """
    t = "(" * inner + _unwrap_np(tail).replace("−", "-")
    got = _scan_value(t, 0, series, True)
    if got is None and not inner and t.count("(") + t.count("[") > t.count(")") + t.count("]"):
        got = _scan_value(_re.sub(r"[()\[\]]", " ", t), 0, series, True)   # 닫히지 않은 괄호 — 괄호 기호만 떼고 다시
    if got is None and not inner:
        opened = _paren_open(t, unit)                     # 「하다 (18.74 %)」 · `('하다', 18.74)` — 값이 괄호 안에만 있습니다
        if opened is not None:
            got = _scan_value(opened, 0, series, True)
            return got, got is not None
    return got, False


def _scan_value(t, i, series, top):
    """`_read_value` 의 한 단계 — top 이면 이름 바로 뒤, 아니면 값 표지 뒤에서 읽습니다"""
    depth, cand, sep, kws, start, n = 0, None, False, [], i, len(t)
    while i < n:
        ch = t[i]
        if ch in "([":
            depth += 1
            i += 1
            continue
        if ch in ")]":
            depth = _bi.max(0, depth - 1)
            i += 1
            continue
        if depth:
            i += 1
            continue
        kw = _KW_RE.match(t, i)
        if kw and (i == 0 or not (t[i - 1].isalnum() or t[i - 1] == "_")):
            val = _KW_VAL.match(t, kw.end())
            toks = _num_tokens(val.group(0)) if val else []
            if toks:
                kws.append(toks[0])
            i = val.end() if val else kw.end()
            continue
        if ch in "=:→：＝" or t.startswith("->", i):
            step = 2 if t.startswith("->", i) else 1
            if ch == ":" and 0 < i < n - 1 and t[i - 1].isdigit() and t[i + 1].isdigit():
                i += 1                                            # 시각·비율(23:00 · 1:2)
                continue
            if ch == "=" and t[i + 1:i + 2] == "=":
                i += 2
                continue
            arrow = ch == "→" or step == 2
            if cand is not None and (sep or arrow or not top or (cand[3] and "." in cand[1])):
                return cand                                       # 값 뒤의 다른 말(「크기 = …」·「→ 해석」)
            got = _scan_value(t, i + step, series, False)
            if got is not None:
                if top:
                    got = got[:3] + (cand is None and _bare_gap(t[start:i]),) + got[4:5] + ("marker",)
                return got
            if cand is not None:
                return cand
            i += step
            continue
        m = _NUM_RE.match(t, i)
        if m:
            if cand is None:
                raw, end, kind = m.group(0), m.end(), "plain"
                if series:                                        # 「= 24    277.46」 — Series 의 인덱스와 값
                    pair = _re.match(r"(-?\d+)\s{2,}(" + _NUM + ")", t[i:])
                    if pair:
                        raw, end, kind = pair.group(2), i + pair.end(), "series"
                try:
                    cand = (float(raw.replace(",", "")), raw, t[end:], top and _bare_gap(t[start:i]), kind,
                            "bare" if top else "marker")
                except ValueError:
                    cand = None
                if cand is not None and not top:
                    return cand
            i = m.end()
            continue
        if cand is not None and (ch in _SEP or (ch == "-" and t[i - 1:i] == " " and t[i + 1:i + 2] == " ")):
            sep = True
        i += 1
    if cand is not None:
        return cand
    if kws:                                                       # 인자 설명뿐 — 마지막 인자의 값(「fs_hour=0.5715」)
        return (kws[-1][1], kws[-1][0], "", False, "kw", "kw")
    return None


def _depth(txt):
    """txt 를 지난 뒤의 괄호 깊이"""
    depth = 0
    for ch in txt or "":
        if ch in "([":
            depth += 1
        elif ch in ")]":
            depth = _bi.max(0, depth - 1)
    return depth


def _spans(txt):
    """맨 바깥 괄호·대괄호 묶음의 (여는 자리, 닫는 자리) 목록 — 닫히지 않은 묶음은 글 끝까지입니다"""
    out, depth, start = [], 0, 0
    for i, ch in enumerate(txt or ""):
        if ch in "([":
            if depth == 0:
                start = i
            depth += 1
        elif ch in ")]" and depth:
            depth -= 1
            if depth == 0:
                out.append((start, i))
    if depth:
        out.append((start, len(txt)))
    return out


def _squeeze(txt):
    """빈칸과 괄호 기호를 걷고 대시·가운뎃점 글리프를 하나로 맞춘 글자"""
    return "".join(_GLYPH.get(ch, ch) for ch in (txt or "") if not ch.isspace() and ch not in "()[]")


def _tighten(txt, own=None):
    """빈칸과 괄호 기호를 걷어낸 글자 · 원래 자리 — 대시·가운뎃점 글리프는 하나로 맞춥니다.

    own 이 None 이면 괄호 안의 말은 둡니다(기호만 걷음). own 이 집합이면 안의 말이 own 에 있는 묶음만 두고 나머지 묶음은
    안의 말까지 걷습니다. 줄에 끼어든 괄호 설명(「…입력(512일)에 있는…」)은 이름의 일부가 아니지만, 이름에 원래 있는
    괄호 안의 말(「…MASE(Chronos-2 + 공휴일)」)은 이름을 가르는 부분이라 걷지 않습니다.
    """
    txt = txt or ""
    drop = set()
    if own is not None:
        for a, b in _spans(txt):
            if _squeeze(txt[a + 1:b]) not in own:
                drop.update(range(a, b + 1))
    keep = [i for i, ch in enumerate(txt) if i not in drop and not ch.isspace() and ch not in "()[]"]
    return "".join(_GLYPH.get(txt[i], txt[i]) for i in keep), keep


def _canon(txt):
    """숫자 표기를 맞춥니다 — 5.0 → 5 · 3.50 → 3.5 · 1,234 → 1234"""
    def _one(m):
        s = m.group(0).replace(",", "")
        return s.rstrip("0").rstrip(".") if "." in s else s
    return _re.sub(r"\d[\d,]*(?:\.\d+)?", _one, txt or "")


def _label_spot(line, label):
    """줄에서 이름이 있는 자리 (시작, 끝, 느슨하게 찾았는가) — 없으면 None.

    차례로 찾습니다 — 정확히 같은 글자(괄호 밖 자리 먼저) → 빈칸과 줄에 끼어든 괄호 설명을 걷어낸 꼴(대시 — – ― - 와
    가운뎃점 · ㆍ • ‧ ・ 는 같게) → 괄호 기호만 뗀 꼴(「횟수(D)」 와 「횟수 D」) → 숫자 표기까지 맞춘 꼴은 부르는 쪽이 봅니다.

    이름에 괄호가 있으면 그 안의 말도 이름입니다. 「공휴일 있는 세 주 MASE(Chronos-2 + 공휴일)」 을 찾을 때
    「공휴일 있는 세 주 MASE(Chronos-2 제로샷) = 2.763」 은 다른 이름의 줄입니다 — 괄호 안까지 걷고 찾으면 나란히 찍은
    다른 모델의 줄이 같은 이름 줄이 되어 맞는 코드가 「줄이 2개이고 값이 서로 다릅니다」 로 막힙니다. 괄호 묶음을 아예
    적지 않은 줄(「공휴일 있는 세 주 MASE = 1.022」)만 느슨하게 찾은 것으로 돌려주고, 부르는 쪽은 제대로 찾은 줄이 하나라도
    있으면 느슨하게 찾은 줄을 쓰지 않습니다(`_occurrences`).
    """
    if label and label in line:
        spots = [m.start() for m in _re.finditer(_re.escape(label), line)]
        top = [p for p in spots if _depth(line[:p]) == 0]
        # 「{'상태별 리뷰 수': 178, '리뷰 수': 200}」 — 앞에 다른 단어가 붙은 자리(더 긴 이름)보다 이름으로 시작하는 자리가 먼저입니다
        at = next((p for p in top if not line[:p].rstrip()[-1:].isalpha()), top[0] if top else spots[0])
        return at, at + len(label), False
    full = _tighten(label)[0]
    if not full:
        return None
    groups = [_squeeze(label[a + 1:b]) for a, b in _spans(label)]
    for own in (set(groups), None):
        tight_line, keep = _tighten(line, own)
        at = tight_line.find(full)
        if at >= 0:
            return keep[at], keep[at + len(full) - 1] + 1, False
    if not groups:
        return None
    # 이름의 괄호 묶음을 적지 않은 줄 — 뼈대(괄호 묶음을 뺀 이름)만으로 찾습니다. 이름의 괄호 자리에 다른 말의 괄호가
    # 있으면(「MASE(Chronos-2 제로샷)」) 이 이름의 줄이 아닙니다
    base, base_keep = _tighten(label, set())
    base_line, keep = _tighten(line, set())
    at = base_line.find(base) if base else -1
    if at < 0:
        return None
    spot_of = {_bi.sum(1 for k in base_keep if k < a): g for (a, _b), g in zip(_spans(label), groups)}
    for a, b in _spans(line):
        rel = _bi.sum(1 for k in keep if k < a) - at
        if rel in spot_of and _squeeze(line[a + 1:b]) != spot_of[rel]:
            return None
    return keep[at], keep[at + len(base) - 1] + 1, True


_OP_SURE = set("−+×÷*")                    # 식의 연산 기호
_OP_MAYBE = set("-–‒/><≥≤→")               # 붙임표 · 빗금 · 부등호 · 화살표 — 식일 수도, 말을 가르는 자리일 수도 있습니다


def _operand_head(head):
    """이름 앞의 글이 「다른 이름 + 연산 기호」 로 끝나는가 — "expr"(이름이 식의 뒤 항) · "maybe"(식인지 모름) · "".

    「명사만 1위 − 명사 + 용언 1위 : −7.82 %p」 에서 「명사 + 용언 1위」 는 뺄셈의 뒤 항이고 −7.82 는 두 값의 차입니다.
    이 줄을 뒤 이름의 값 줄로 읽으면 제 값 줄(「명사 + 용언 1위 = 하다 18.74 %」)과 값이 달라 맞는 출력이
    「줄이 2개이고 값이 서로 다릅니다」 가 됩니다. 이름에 원래 든 기호(「명사 + 용언 1위」 의 +)는 이름의 일부라 보지 않습니다.
    기호 앞의 토막이 숫자로 시작하면(「A = 0.271 / 이름 = 0.281」 · 「A = 40.2 대/도 - 이름 = …」) 그 기호는 항목을 가르는
    자리이고, 기호 앞에 말이 없으면(「- 이름 = 값」 · 「* 이름」 · 「> 이름」) 글머리표입니다 — 둘 다 식이 아닙니다.
    붙임표 · 빗금 · 부등호 · 화살표(「A - 이름」 · 「A / 이름」 · 「A > 이름 : a > b」 · 「A → 이름 : a → b」)는 말을 가르는 자리로도
    쓰여 식인지 알 수 없습니다 — "maybe" 로 돌려주고, 부르는 쪽이 이름으로 시작하는 줄을 먼저 씁니다.
    """
    h = (head or "").rstrip()
    op = h[-1:]
    if not op or (op not in _OP_SURE and op not in _OP_MAYBE):
        return ""
    before = h[:-1].rstrip()
    if not before or not (before[-1].isalnum() or before[-1] in ")]%\0」』”’\"'"):
        return ""
    if op in "-–‒" and h[-2:-1].strip():           # 앞말에 붙은 붙임표는 단어를 잇는 글자입니다
        return ""
    cut = _bi.max(before.rfind(ch) for ch in "=:→：＝,;|")
    seg = before[cut + 1:].strip()
    if not seg or _NUM_RE.match(seg.replace("−", "-")):
        return ""
    return "expr" if op in _OP_SURE else "maybe"


def _mark_at(tail):
    """꼬리에서 괄호 밖 첫 값 표지(「=」·「:」, 전각 포함)의 자리 — 없으면 -1. 시각 23:00 의 「:」 는 표지가 아닙니다"""
    depth, n = 0, len(tail or "")
    for i, ch in enumerate(tail or ""):
        if ch in "([":
            depth += 1
        elif ch in ")]":
            depth = _bi.max(0, depth - 1)
        elif depth == 0 and ch in "=:：＝":
            if ch == ":" and 0 < i < n - 1 and tail[i - 1].isdigit() and tail[i + 1].isdigit():
                continue
            return i
    return -1


#  강한 값 줄(0.4.2) — 이름이 줄 머리에 있고, 이름 바로 뒤가 값 표지(「=」·「:」·「→」, 전각 포함)이며 표지 바로 뒤가 숫자인 줄.
#  줄 머리: 이름 앞에는 빈칸 · 글머리표 · 번호(「1.」·「1)」·「(1)」·「①」) · 여는 따옴표 · 굵게 표시 · 사전의 「{」 만 옵니다
_LEAD_OK = _re.compile(r"(?:\s|[-*•‧·+>▪◦▶■□○●※→]|\d{1,3}[.)](?=\s)|\(\d{1,3}\)|[①-⑳⑴-⒇]|['\"`「『“‘{]|[(\[](?=['\"]))*")
#  이름과 표지 사이에는 닫는 따옴표 · 굵게 표시 · 빈칸만 — 숫자가 numpy 꼴(np.float64(…) · [ … ] · array([…]))로 싸여도 됩니다
_STRONG_TAIL = _re.compile(r"[\s'\"`*」』”’]*(?:=(?!=)|:|：|＝|→|->)\s*"
                           r"(?:(?:np|numpy)\.[A-Za-z_]\w*\(\s*|(?:float|int|uint)(?:8|16|32|64|128)\(\s*|array\(\s*\[\s*|\[\s*)?"
                           r"[-+−]?\.?\d")
#  한 줄에 값 짝을 이어 적는 구분 기호 — 쉼표(숫자 사이의 천 단위 쉼표는 아님) · 쌍반점 · 세로줄 · 빗금 · 가운뎃점 · 줄표 · 화살표 · 빈칸 둘 이상
_CHAIN_SEP = _re.compile(r"(?<!\d),|,(?!\d)|[;|/·ㆍ•‧・∙⋅—–―→]|->|\s-\s|\s{2,}")
_NAME_RE = {}                              # 이름 → 빈칸 · 대시 · 가운뎃점 글리프를 가리지 않고 찾는 정규식(한 번만 짓습니다)


def _name_len_at(text, i, names):
    """text 의 i 자리에서 시작하는 이름(names 가운데 가장 긴 것)의 글자 수 — 없으면 0. 가려 둔 더 긴 이름(\\0)도 이름입니다"""
    if text[i:i + 1] == "\0":
        j = i
        while j < len(text) and text[j] == "\0":
            j += 1
        return j - i
    best = 0
    for name in names:
        if not name:
            continue
        if text.startswith(name, i):
            best = _bi.max(best, len(name))
            continue
        pat = _NAME_RE.get(name)
        if pat is None:
            same = {}
            for k, v in _GLYPH.items():
                same.setdefault(v, set()).add(k)
            pat = _NAME_RE[name] = _re.compile(r"\s*".join(
                "[" + _re.escape("".join(sorted(same[_GLYPH[ch]] | {_GLYPH[ch]}))) + "]" if ch in _GLYPH else _re.escape(ch)
                for ch in name if not ch.isspace()))
        m = pat.match(text, i)
        if m:
            best = _bi.max(best, m.end() - i)
    return best


def _chain_head(lead, names):
    """이름 앞의 글(lead)이 「이름 = 숫자」 짝을 구분 기호로 이어 적은 값 줄의 앞부분인가 — 「확신 = 26 시간 · 보류 = 142 시간」 의 「보류」.

    맨 앞 짝은 줄 머리에서 시작하고, 짝마다 이름(값 칸의 이름) 바로 뒤가 값 표지이고 표지 바로 뒤가 숫자이며, 짝 사이에는 값 표지가 없는
    글과 구분 기호만 있습니다. 「(참고) 제로샷: A = 85.1 % · B = 41.6 %」 의 B 는 맨 앞 짝이 줄 머리에서 시작하지 않아 아닙니다 —
    비교하려고 덧붙인 줄이라 그 줄의 수가 맨 이름 「B = 값」 줄과 부딪치지 않게 합니다.
    """
    m = _LEAD_OK.match(lead)
    i = m.end() if m else 0
    while True:
        k = _name_len_at(lead, i, names)
        if not k:
            return False
        mt = _STRONG_TAIL.match(lead, i + k)
        if not mt:
            return False
        j, nxt = mt.end(), None
        for sm in _CHAIN_SEP.finditer(lead, j):
            if _re.search(r"[=＝：]|(?<!\d):|:(?!\d)", lead[j:sm.start()]):
                return False                                  # 짝 사이의 다른 값 표지 — 이름을 모르는 짝입니다
            q = sm.end()
            while q < len(lead) and lead[q] in " \t'\"`*「『“‘":
                q += 1
            if q >= len(lead):
                return True                                   # 지금 이름 바로 앞의 구분 기호
            if _name_len_at(lead, q, names):
                nxt = q
                break
        if nxt is None:
            return False
        i = nxt


def _lead_kind(lead):
    """이름 앞 글(lead)의 꼴 — "top"(줄 머리) · "num"(날짜 · 숫자 · % 뒤) · "sep"(구분 기호 뒤) · ""(그 밖)"""
    if _LEAD_OK.fullmatch(lead or ""):
        return "top"
    body = (lead or "").rstrip(" \t'\"`*「『“‘")
    last = body[-1:]
    if last.isdigit() or last == "%":
        return "num"
    if last and (last in ",;|/·ㆍ•‧・∙⋅—–―" or (last in "-–‒" and body[-2:-1].isspace())):
        return "sep"
    return ""


def _occurrences(lines, label, others=(), series=False, unit=""):
    """이름이 있는 줄마다 하나씩 — {"tail": 이름 뒤 꼬리, "seg": 이 항목의 토막, "val": `_read_value` 결과, "boxed": 배열인가}.

    값줄의 숫자를 한 주머니에 담아 기준 범위와 대조하면 **다른 항목 줄의 값**이 이 항목을 헛통과시킵니다.
    그래서 이름이 있는 줄의 이름 뒤만 봅니다. 꼬리에 다른 항목의 이름(`others`)이 나오면 거기서 끊고
    (표 머리 줄에서 옆 칸 이름 안의 「365일」·「168」 을 이 항목의 값으로 읽지 않게), 이 이름을 품은 더 긴
    이름(「공식 차수」 에 대한 「공식 차수 AIC」)이 있는 자리는 이 이름으로 치지 않습니다.
    이름이 괄호 안에서만 잡히면(「… (명목 포함률 80% 기준)」) 그 괄호가 닫히기 전의 숫자는 읽지 않습니다. 따옴표로
    싼 이름(튜플·리스트를 그대로 찍은 `('이름', 0.281)`)은 괄호가 값을 담은 그릇이라 그 안의 값을 읽습니다.
    "seg" 는 이 이름부터 다른 항목 이름 앞까지입니다 — 줄의 첫 이름이면 이름 앞의 말까지 넣습니다.
    "line" 은 줄 원문, "loose" 는 이름의 괄호 묶음을 적지 않은 줄인가입니다 — 이름을 제대로 적은 줄이 하나라도 있으면
    느슨하게 찾은 줄은 돌려주지 않습니다(`_label_spot`). "after" 는 다른 항목 이름에서 끊지 않은 이름 뒤의 글입니다
    (값이 단어 · 문장인 항목을 보는 권 검사가 씁니다 — `_text_tails`). "pre" 는 이름 바로 앞에 다른 단어가 붙어 있는가입니다
    (「상태별 리뷰 수 : …」 의 「리뷰 수」 — 더 긴 이름의 줄일 수 있어 `_value_use` 가 이름으로 시작하는 줄을 먼저 씁니다.
    붙임표 · 빗금 뒤의 이름도 여기에 넣습니다).
    이름 앞에 「다른 이름 + 연산 기호」 가 있는 줄은 이름이 식의 뒤 항이라 이 이름의 줄이 아닙니다. 기호가 붙임표 · 빗금이면
    식인지 알 수 없어 느슨하게 찾은 줄로 둡니다(`_operand_head`).
    "tag" 는 이름 앞에 꼬리표가 붙은 줄인가입니다 — 「(전임 코드) 이름 = …」 · 「고치기 전: 이름 = …」 처럼 이름 앞의 말이
    닫는 괄호나 쌍점으로 끝납니다. "noted" 는 이름과 값 표지 사이에 괄호 설명이 낀 줄인가입니다 — 「이름(50만 건 전체, 이론) = …」.
    둘 다 비교하려고 덧붙인 줄일 수 있어, 맨 이름 「이름 = 값」 줄이 따로 있으면 `_value_use` 가 쓰지 않습니다(그런 줄뿐이면 그 값을 읽습니다).
    "strong" 은 강한 값 줄인가입니다(0.4.2) — 이름이 줄 머리에 있고(앞에는 글머리표 · 번호 · 따옴표만, `_LEAD_OK`) 이름 바로 뒤가
    값 표지이며 표지 바로 뒤가 숫자인 줄(`_STRONG_TAIL`), 그리고 그런 「이름 = 숫자」 짝을 구분 기호로 이어 적은 줄의 뒤 짝(`_chain_head`)입니다.
    강한 값 줄이 하나라도 있으면 `_value_use` 가 그 줄끼리만 값을 읽고 부딪침을 봅니다. "aside" 는 이름 앞이 날짜 · 숫자로 끝나거나
    (「2025-08-13 이름: 78.9%」) 구분 기호 바로 뒤에 이름이 온 줄(「명목 80% 와의 차: A -0.5 %p · 이름 -13.6 %p」), 꼬리표 뒤의 이름에
    표지 없이 숫자가 붙은 줄(같은 줄의 「명목 80% 와의 차: 이름 -0.5 %p」)입니다 — 강한 값 줄이 없을 때 이름으로 시작하는 줄이 따로
    있으면 `_value_use` 가 쓰지 않습니다. "paren" 은 값을 괄호 안에서만 읽은 줄입니다(`_read_paren`) — 다른 값 줄이 있으면 쓰지 않습니다.
    unit 은 이 칸의 기준 단위입니다 — 값이 괄호 안에만 있는 줄에서 기준 단위가 붙은 수를 읽습니다(`_paren_open`).
    """
    rivals = [o for o in others if o and o != label]
    longer = [o for o in rivals if label in o]
    # 빠른 거르기 — 이름 뼈대의 글자(한글·영문)가 다 있지 않은 줄에는 이름이 없습니다(출력이 수만 줄일 때 채점이 늦지 않게)
    need = {ch for ch in _tighten(label, set())[0] if ch.isalpha()}
    out = []
    for line in lines:
        if not need.issubset(line):
            continue
        masked = line
        for o in longer:
            masked = masked.replace(o, "\0" * len(o))
        spot = _label_spot(masked, label)
        plain = line
        if spot is None:
            masked = _canon(masked)
            spot = _label_spot(masked, _canon(label))
            plain = masked
        if spot is None:
            continue
        start, end, loose = spot
        role = _operand_head(masked[:start])
        if role == "expr":
            continue                                              # 「다른 이름 − 이름 : 차」 — 이 이름의 값 줄이 아닙니다
        loose = loose or role == "maybe"
        tail = masked[end:]
        stops = []
        for o in rivals:
            at = tail.find(o) if o not in longer else -1
            while at >= 0 and _depth(tail[:at]):                  # 괄호 설명 안에 나온 다른 항목 이름에서는 끊지 않습니다 —
                at = tail.find(o, at + 1)                         # 「이상치 비율 (분모 N) = 37.68 %」 의 「분모」
            if at >= 0:
                stops.append(at)
        if stops:
            tail = tail[:_bi.min(stops)]
        inner = _depth(masked[:start])
        quote = masked[start - 1:start]
        if inner and quote in ("'", '"') and masked[end:end + 1] == quote:
            inner = 0                                             # ('이름', 값) · ['이름', 값] — 그릇 안의 값
        head = masked[:end]
        seg = (tail if _bi.any(o in head for o in rivals if o not in longer) else head + tail).replace("\0", "")
        val, in_paren = _read_paren(tail, series, inner, unit)
        if val is not None and inner:
            val = val[:3] + (False,) + val[4:]
        boxed = val is None and _bi.bool(_re.search(r"\[[^\[\]]*\d[^\[\]]*\]", _unwrap_np(tail)))
        lead = masked[:start].rstrip()
        cut = _mark_at(tail)
        kind = _lead_kind(masked[:start])                         # 줄 머리 · 날짜와 숫자 뒤 · 구분 기호 뒤
        chain = kind != "top" and not inner and _chain_head(masked[:start], [label] + rivals)
        tag = lead[-1:] in tuple(":：)]」』") and _bi.any(ch.isalpha() for ch in lead)
        out.append({"tail": tail.replace("\0", ""), "seg": seg, "val": val, "boxed": boxed, "line": line, "loose": loose,
                    "after": plain[end:].replace("\0", ""),
                    "pre": role == "maybe" or lead[-1:].isalpha(),
                    "tag": tag,
                    "noted": cut >= 0 and _bi.bool(_re.search(r"[(\[]", tail[:cut])) and _bare_gap(tail[:cut]),
                    "strong": (val is not None and val[3] and val[5] == "marker" and not inner and not in_paren
                               and (kind == "top" or chain) and _bi.bool(_STRONG_TAIL.match(tail))),
                    "paren": in_paren,                            # 값을 괄호 안에서만 읽은 줄(약한 줄 — 0.4.2)
                    # 이름 앞이 날짜 · 숫자 · 구분 기호로 끝나는 줄(값 짝을 이은 줄은 아님), 꼬리표 뒤 이름에 표지 없이 숫자가 붙은 줄
                    # (「명목 80% 와의 차: 이름 -0.5 %p」) — 보조 줄일 수 있습니다
                    "aside": (kind in ("num", "sep") and not chain) or (tag and val is not None and val[5] == "bare")})
    if _bi.any(not o["loose"] for o in out):
        out = [o for o in out if not o["loose"]]
    return out


def _unit_value(val, unit):
    """「k 건 (r %)」 꼴에서 기준 단위가 붙은 괄호 안의 수 — 그런 꼴이 아니면 읽은 값 그대로.

    기준 단위가 % 인 칸에 「이름 = 14 건 (1.4 %)」 로 찍으면 표지 뒤 첫 숫자는 건수 14 이고 비율은 괄호 안에 있습니다.
    읽은 값 바로 뒤에 기준 단위가 없고, 값(과 짧은 단위 단어) 바로 뒤의 괄호 안에 기준 단위가 붙은 수가 있으면 그 수가 값입니다.
    값과 괄호 사이에 분모를 적은 꼴(「14 건 / 1,000 건 (1.4 %)」 · 「14 건 ÷ 1,000 건 (1.4 %)」)도 같습니다.
    칸마다 켜는 규칙입니다(기준의 `paren_unit`) — 「확신 = 26 (전체 168 시간 가운데)」 처럼 괄호 안의 수가 설명인 칸에도
    켜면 설명의 수를 값으로 읽습니다.
    """
    after = val[2]
    if not unit or after.lstrip().startswith(unit):
        return val
    word = r"[^\s\d()\[\],;:=→·|/÷]{0,6}"                          # 값 뒤의 짧은 단위 단어(건 · 개 · 명)
    m = _re.match(r"\s*" + word + r"\s*(?:[/÷]\s*[\d,]+(?:\.\d+)?\s*" + word + r"\s*)?\(", after)   # (분모) 여는 괄호
    if not m:
        return val
    close = after.find(")", m.end())
    close = len(after) if close < 0 else close
    inner = after[m.end():close]
    for tok in _NUM_RE.finditer(inner):
        if inner[tok.end():].lstrip().startswith(unit):
            try:
                return (float(tok.group(0).replace(",", "")), tok.group(0), inner[tok.end():] + after[close:]) + val[3:]
            except ValueError:
                continue
    return val


def _value_use(occ, val_lines):
    """이름이 있는 줄들(`_occurrences`) 가운데 채점에 쓸 값 줄.

    같은 이름 줄이 여럿이면 이름 바로 뒤 값 줄만 봅니다. 바로 뒤 값 줄은 「이름 = 값」·「이름: 값」(어느 행이든), 그리고
    단위·「=」 가 붙은 행(val_lines)의 「이름 값」 입니다. 표지도 단위도 없는 행의 「이름 숫자」(「결과 요약: 이름 A, 이름 B 2개 값」
    의 2)는 값 줄이 달리 없을 때만 씁니다. 이름 바로 앞에 다른 단어가 붙은 줄은 더 긴 이름의 줄일 수 있습니다 — 도우미가 찍는
    「상태별 리뷰 수 : 통과 178 · …」 을 「리뷰 수」 의 값 줄로 읽으면 맞는 출력이 「줄이 2개이고 값이 서로 다릅니다」 가 됩니다.
    이름으로 시작하는 값 줄이 따로 있으면 그런 줄은 쓰지 않고, 그런 줄뿐이면 지금처럼 그 값을 읽습니다.
    비교하려고 덧붙인 줄도 같습니다 — 이름 앞에 꼬리표가 붙은 줄(「(전임 코드) 이름 = 892 건」)과 이름 뒤 괄호에 설명을 넣은 줄
    (「이름(50만 건 전체, 이론) = 17.39 %」)은 맨 이름 「이름 = 값」 줄(이름 바로 뒤 값 줄)이 따로 있으면 쓰지 않습니다. 설명을 이름
    앞에 붙인 「원본 이름 = …」 은 통과하고 이름 뒤 괄호에 넣은 줄만 「줄이 2개이고 값이 서로 다릅니다」 가 되지 않게 합니다.

    강한 값 줄(「이름 = 숫자」 — 이름이 줄 머리에 있고 표지 바로 뒤가 숫자인 줄)이 하나라도 있으면 그 줄끼리만 씁니다(0.4.2).
    원점별 행 · 명목 대비 줄 · 제로샷 비교 줄 · 날짜가 앞에 붙은 줄처럼 같은 이름이 든 보조 줄의 수가 맨 이름 줄과 부딪쳐 맞는 출력이
    「줄이 2개이고 값이 서로 다릅니다」 가 되지 않게 합니다. 옛 print 와 새 print 를 함께 남긴 셀, 전임 코드의 줄은 둘 다 강한 값 줄이라
    지금처럼 부딪침으로 잡힙니다. 강한 값 줄이 없으면 아래 차례(0.4.1 과 같고, 이름 앞에 날짜 · 숫자가 오거나 구분 기호 바로 뒤에 이름이 온
    줄과 꼬리표 뒤 이름에 표지 없이 숫자가 붙은 줄 — "aside" — 은 이름으로 시작하는 줄이 따로 있을 때, 값을 괄호 안에서만 읽은 줄 —
    "paren" — 은 다른 값 줄이 있을 때 쓰지 않습니다)로 고릅니다.
    """
    found = [o for o in occ if o["val"]]
    strong = [o for o in found if o.get("strong")]
    if strong:                                              # 강한 값 줄끼리만 — 보조 줄의 수는 이 항목의 값으로 읽지 않습니다
        return strong
    if _bi.any(not o["pre"] for o in found):               # 이름으로 시작하는 줄(구분 기호 뒤의 이름 포함)이 있으면, 이름 앞에
        found = [o for o in found if not o["pre"]]          # 다른 단어가 붙은 줄(「상태별 리뷰 수 : 통과 178 …」)은 쓰지 않습니다
    if _bi.any(not o.get("aside") for o in found):         # 이름 앞이 날짜 · 숫자이거나 구분 기호 바로 뒤에 이름이 온 줄도 같습니다
        found = [o for o in found if not o.get("aside")]
    if _bi.any(not o.get("paren") for o in found):         # 값을 괄호 안에서만 읽은 줄(「이름 (제로샷 41.6 % 보다 높음)」)은
        found = [o for o in found if not o.get("paren")]    # 다른 값 줄이 있으면 쓰지 않습니다 — 그런 줄뿐이면 0.4 의 53번대로 읽습니다
    # 비교하려고 덧붙인 줄(꼬리표 · 이름 뒤 괄호 설명)은 맨 이름 「이름 = 값」 줄 — 이름 바로 뒤 값 줄 — 이 따로 있을 때만 뺍니다.
    # 문장 속의 이름이나 다른 줄의 이름(「… (분모 2,000건)」 · 「결론 : … 이름은 9월의 0.933배 …」)이 있다고 괄호 설명 줄을 버리면
    # 「분모(보낸 리뷰 수) = 2,000 건」 처럼 설명을 붙인 맞는 값 줄을 잃습니다
    if _bi.any(o["val"][3] and not o["tag"] and not o["noted"] for o in found):
        found = [o for o in found if not o["tag"] and not o["noted"]]
    direct = [o for o in found if o["val"][3]]
    return ([o for o in direct if o["line"] in val_lines or o["val"][5] != "bare"]
            or [o for o in found if o["line"] in val_lines] or direct or found)


def _text_tails(lines, label, others=()):
    """이름이 있는 줄마다 값 표지(「=」·「:」) 뒤의 글 — 값이 단어 · 문장인 항목(권 검사)이 읽습니다.

    이름 바로 뒤가 표지인 줄(「이름 = …」 · 「이름 (설명): …」)이 있으면 그 줄만, 없으면 이름이 있는 모든 줄입니다
    (표지가 없는 줄은 이름 뒤 전체). 글은 다른 항목 이름에서 끊지 않습니다 — 내가 적은 문장에 다른 항목의 이름이 나와도 그대로입니다.
    others 는 다른 항목의 이름입니다 — 이 이름을 품은 더 긴 이름(「1위 토픽」 에 대한 「1위 토픽 비율」)의 줄을 가려냅니다.
    이름 앞에 다른 단어가 붙은 줄(「전임 판단 : …」 — 다른 이름의 줄)과 꼬리표가 붙은 줄(「(전임 코드) 판단 : …」)은
    같은 급의 줄 가운데 이름으로 시작하는 줄이 따로 있으면 쓰지 않습니다(값 칸의 `_value_use` 와 같은 규칙). 「고른 품목 = 헬스장」
    처럼 그런 줄뿐이면 그 줄을 읽습니다 — 표지 없는 줄(「… · 품목 민원 4,000건」)이 있다고 버리지 않습니다.
    """
    whole, rest = [], []
    for o in _occurrences(lines, label, others):
        cut = _mark_at(o["after"])
        if cut >= 0 and _bare_gap(o["after"][:cut]):
            whole.append((o["after"][cut + 1:], o))
        else:
            rest.append((o["after"][cut + 1:] if cut >= 0 else o["after"], o))
    pick = whole or rest
    for mark in ("pre", "tag"):
        if _bi.any(not o[mark] for _, o in pick):
            pick = [(t, o) for t, o in pick if not o[mark]]
    return [t for t, _ in pick]


def _value_lines(txt, units):
    """값 줄 — 단위가 붙은 행, 또는 「이름 = 값」 꼴의 행.

    이름은 출력의 모든 행에서 찾습니다(「이름: 값」 꼴 포함). 이름 바로 뒤 값 줄이 하나도 없을 때만 이 행들의 값을
    먼저 씁니다 — 이 행들에서 값이 읽혔다고 「이름: 값」 줄을 보지 않으면, 틀린 값을 「:」 로 찍고 맞는 값을 「=」 로 찍은
    출력의 충돌을 놓칩니다. 이름이 있는 행이 아예 없으면 숫자를 읽지 않고 ❌ 로 알립니다(아래 `judge`).
    """
    return [line for line in (txt or "").splitlines()
            if line.strip() and ("=" in line or _bi.any(unit and unit in line for unit in units))]


def _short_digits(raw, value, lo, hi):
    """찍힌 자릿수가 모자라 범위 밖으로 읽혔으면 그 자릿수, 아니면 None.

    「0.10」 · 「205」 처럼 찍힌 자리로 반올림하면 기준 범위와 겹치는데 값 자체는 범위 밖이면, 계산이 아니라
    출력 자릿수를 짚어야 학생이 엉뚱한 원인을 찾지 않습니다. 소수 한 자리(「0.1」)는 자릿수를 줄인 값인지
    조회표 끝의 p 값인지 가를 수 없어 보지 않습니다.

    겹침은 경계를 빼고 봅니다 — 반올림 구간의 끝이 기준 범위의 끝에 닿기만 한 값은 자릿수 문제가 아닙니다.
    개수처럼 정수인 칸은 기준 범위를 n±0.5 로 잡는데, 바로 옆 정수(n−1 · n+1)의 반올림 구간이 그 끝에 딱 닿습니다.
    닿은 것까지 겹침으로 치면 틀린 개수(167 · 169)가 「자릿수 모자람」 이 되고 같은 이름 줄의 충돌에서도 빠집니다.
    """
    raw = (raw or "").replace(",", "").lstrip("+-")
    if not raw or "e" in raw.lower() or lo <= value <= hi:
        return None
    places = len(raw.split(".", 1)[1]) if "." in raw else 0
    if places == 1:
        return None
    half = 0.5 * 10 ** (-places)
    edge = half * 1e-6                      # 부동소수 오차로 「닿음」 이 「겹침」 으로 읽히지 않게 두는 여유
    return places if value - half < hi - edge and value + half > lo + edge else None


# ───────────────────────── 문구 · 축 이름 ─────────────────────────
_NEG_TAIL = ("하지못", "하지않", "하지는못", "하지는않", "되지않", "되지못", "할수없", "못", "아님", "아니",
             "안되", "안됐", "안돼", "안됨", "안됩", "안함", "안했", "불가", "에실패", "실패",
             "에이르지못", "에이르지않", "까지는못", "까지못")
_NEG_MARK = r"(?:False|false|FALSE|No|no|NO|N|X|x|×|✗|✘|❌)(?![A-Za-z])"
_YES_MARK = r"[=:]?(?:True|true|TRUE|Yes|yes|YES|O|예|네|있음|함)(?![A-Za-z])"


def _flat(txt):
    """빈칸을 걷고 겉보기가 같은 대시·가운뎃점을 맞춘 글자"""
    return "".join(_GLYPH.get(ch, ch) for ch in _re.sub(r"\s+", "", txt or ""))


def _word_in(row, word):
    """빈칸을 걷은 줄 `row` 에 단어가 **긍정으로** 있는가.

    「비기각」·「미기각」·「기각하지 못함」·「기각 안 됨」·「기각에 이르지 못함」·「기각 X」·「기각 = False」 는
    「기각」 을 글자로 품지만 뜻은 반대입니다. 「기각 여부」 뒤에 답(True·예)이 없으면 무엇도 말하지 않은 것이고,
    「비정상 쪽 증거 없음」 도 반대입니다. 그 꼴을 지운 뒤에도 단어가 남아야 채운 것으로 봅니다
    (「비정상 쪽」 안의 「정상 쪽」 도 같은 방식으로 지웁니다). 금지 문구(forbid)는 이 규칙을 쓰지 않습니다 —
    머리말·옛 이름에 단어용 부정 꼴을 대면 문구째 지워집니다(`_forbid_in`).
    """
    w = _flat(word)
    if not w:
        return True
    e = _re.escape(w)
    neg = "|".join((
        "(?:비|미|불)" + e,                                            # 비기각 · 미기각 · 불기각
        e + "(?:" + "|".join(_NEG_TAIL) + ")",                          # 기각하지 못함 · 기각 안 됨 · 기각에 이르지 못함
        e + r"(?:여부)?[=:]?" + _NEG_MARK,                              # 기각 X · 기각 = False · 기각 여부 = False
        e + r"의?증거(?:는|가|도)?(?:없|부족|불충분|아님|아니|못)",           # 비정상 쪽 증거 없음
        e + r"여부(?!" + _YES_MARK + ")",                               # 「기각 여부」 뒤에 답이 없다
    ))
    return w in _re.sub(neg, "", _flat(row))


#  금지 문구(forbid) 뒤에서 그 문구를 서술로 부정한 꼴 — 「…넣어도 되지 않습니다」·「운영 가능하지 않다」·「…이 아니다」
_FORBID_NOT = (r"(?:은|는|이|가|도|고|라고|다고)?(?:(?:하|되)지(?:는|도)?(?:않|못)|할수(?:는|도)?없|안(?:되|된|됩|돼|됨|됐)"
               r"|아니|아님|아닙)")


#  줄 머리말(「판단 :」) 바로 뒤에서 그 줄을 지웠다고 알리는 말 — 「판단 : 행은 삭제했습니다」·「판단 : 줄은 지웠습니다」
_FORBID_GONE = (r"(?:행|줄|문장|결론)(?:은|을|도|이|만)?(?:모두|전부|이제)?"
                r"(?:지웠|지움|지우|지운|삭제|뺐|뺌|빼고|뺀|없앴|없앰|없애|제거|남기지않|출력하지않|찍지않)")


def _forbid_in(row, phrase):
    """빈칸을 걷은 줄 `row` 에 금지 문구가 남아 있는가 — 함정 미션의 `forbid`.

    「판단 :」 처럼 「:」 로 끝나는 문구는 줄 머리말입니다. 줄의 맨 앞이나 쉼표·화살표 같은 구분 기호 바로 뒤에 올 때만
    남은 줄입니다 — 「누수 여부 판단: …」 처럼 다른 말 뒤에 이어 쓴 것, 따옴표 안에 옮겨 적은 것은 머리말이 아닙니다.
    머리말 뒤에 오는 말은 가리지 않습니다. 「판단 : 불가」·「판단 : X」·「판단 : 아니다」 도 판단 줄이고, 전각 「：」 도
    같은 표지입니다. 하나만 가립니다 — 머리말 바로 뒤가 그 줄을 지웠다는 말(「(판단 : 행은 삭제했습니다)」)이면 PROMPT 의
    「판단 : 행은 지워」 를 되받아 알린 글이라 남은 줄이 아닙니다(`_FORBID_GONE`).

    그 밖의 문구(옛 이름 · 전임 담당자의 결론)는 글자로 대조합니다. 문구 바로 뒤에서 서술로 부정한 인용(「…넣어도 되지
    않습니다」)과 답 없는 「… 여부」 만 남은 줄이 아닙니다. 단어용 부정 꼴(`_word_in` 의 비·미·불 접두, 못·불가·실패,
    X·No·N)은 쓰지 않습니다 — 「계절 나이브 대비 제로샷 평균 MASE」 의 「비」, 「제로샷 평균 MASE: N/A」 의 「N」,
    「판단 : 불가」 의 「불가」 가 문구를 지워 남은 줄이 통과하던 자리입니다.
    """
    w = _flat(phrase).replace("：", ":")
    if not w:
        return False
    row = _re.sub(r"[*`]", "", row or "").replace("：", ":")
    e = _re.escape(w)
    if w.endswith(":") and len(w) > 1:
        # 「(판단 : 행은 삭제했습니다)」 — 머리말 바로 뒤가 그 줄을 지웠다는 말이면 남은 줄이 아니라 지웠다고 알리는 글입니다
        return _bi.bool(_re.search(r"(?:^|[^0-9A-Za-z가-힣_'\"「『“‘])" + e + "(?!" + _FORBID_GONE + ")", row))
    return w in _re.sub(e + "(?:" + _FORBID_NOT + "|여부(?!" + _YES_MARK + "))", "", row)


def _axes(fig):
    """Figure · Axes 어느 쪽을 받아도 축 목록으로 폅니다"""
    if fig is None:
        return []
    axs = getattr(fig, "axes", None)
    if isinstance(axs, list):
        return axs
    return [fig] if hasattr(fig, "get_xlabel") else []


def _sup_label(ax, which):
    """그림 수준 축 이름(fig.supxlabel · fig.supylabel)이 있는가 — 패널이 여럿인 그림에서 축마다 이름을 다는 대신 씁니다"""
    fig, seen = getattr(ax, "figure", None), set()
    while fig is not None and id(fig) not in seen:   # SubFigure 면 바깥 그림까지 올라갑니다
        seen.add(id(fig))
        text = getattr(fig, "_sup" + which + "label", None)
        try:
            if text is not None and text.get_text().strip():
                return True
        except Exception:
            pass
        fig = getattr(fig, "figure", None)
    return False


def _mate_label(ax, which):
    """가로축(which="x")·세로축("y")을 나눠 쓰는 다른 축에 그 이름이 있는가.

    `ax.twinx()` 로 겹친 두 축은 가로축 이름을 한쪽에, 세로축 이름을 다른 쪽에 다는 일이 흔하고, `sharex=True` 로 쌓은
    그림은 맨 아래 축에만 가로축 이름을 답니다. 화면에는 두 이름이 다 보이므로 짝의 이름도 이 축의 이름으로 봅니다.
    나눠 쓰지 않는 옆 칸의 이름(`subplots(1, 2)` 의 한 칸에 가로축 이름, 다른 칸에 세로축 이름)은 보지 않습니다.
    """
    mates = []
    for find in (lambda: getattr(ax, "get_shared_" + which + "_axes")().get_siblings(ax),
                 lambda: ax._twinned_axes.get_siblings(ax)):
        try:
            mates += [a for a in find() if a is not ax]
        except Exception:
            pass
    for a in mates:
        try:
            if getattr(a, "get_" + which + "label")().strip():
                return True
        except Exception:
            pass
    return False


def _plot_gaps(ax, spec):
    """축 하나를 기준과 대조해 어긋난 곳을 문장 목록으로 돌려줍니다 (빈 목록이면 통과).

    그래프 종류(막대·가로 막대·선·점, pandas·seaborn 으로 그린 것)는 가리지 않고 가로축·세로축 이름만 봅니다.
    `fig.supxlabel`·`fig.supylabel` 로 그림 전체에 단 이름, 축을 나눠 쓰는 짝(twinx · sharex)에 단 이름도 축 이름입니다.
    기준이 `가로축라벨` 이면(세로 눈금에 막대 이름이 붙는 가로 막대 그림) 가로축 이름만 봅니다.
    """
    x_name = ax.get_xlabel().strip() or _sup_label(ax, "x") or _mate_label(ax, "x")
    y_name = ax.get_ylabel().strip() or _sup_label(ax, "y") or _mate_label(ax, "y")
    if spec.get("축라벨") and not (x_name and y_name):
        return ["가로축·세로축 이름 가운데 빈 것이 있습니다"]
    if spec.get("가로축라벨") and not x_name:
        return ["가로축 이름이 비어 있습니다"]
    return []


def _order_of(tail):
    """「공식 차수」 줄 꼬리의 비계절 차수 — ((p, d, q), 이름 바로 뒤 값 줄인가) · 못 읽으면 None.

    값 표지(괄호 밖 첫 「=」·「:」) 뒤에서 **정수 3개가 든 첫 괄호 묶음**을 먼저 찾습니다(계절 차수까지 한 묶음에 적은
    7개도). 「= 1위 조합 (1,0,1)(0,1,0,24)」·「= AIC 1위 (1, 0, 1) x (0, 1, 0, 24)」 처럼 괄호 앞에 숫자가 든 말이 와도
    괄호 안을 읽습니다. 그런 묶음이 없으면 표지 뒤 앞 숫자 3개입니다 — 「= 1, 0, 1」. (1,0,1) · (1.0, 0, 1.0) ·
    SARIMA(1,0,1)(…) 를 같게 읽고, 이름 뒤 괄호 설명(「(p=1, d=0, q=1 로 적합)」)은 건너뜁니다.
    진행 막대 줄(「17%|█▋ | 1/6 [00:00<…」)은 차수 줄이 아닙니다.
    """
    t, depth, cut = _unwrap_np(tail), 0, -1
    if _re.search(r"\d+%\|", t):
        return None
    for i, ch in enumerate(t):
        if ch in "([":
            depth += 1
        elif ch in ")]":
            depth = _bi.max(0, depth - 1)
        elif ch in "=:：＝" and depth == 0:
            cut = i
            break
    rest = t[cut + 1:] if cut >= 0 else t
    direct = cut >= 0 and _bare_gap(t[:cut])
    num = r"(?<![A-Za-z_\d.])-?\d+(?:\.\d+)?"
    picks = [_re.findall(num, a or b) for a, b in _re.findall(r"\(([^()\[\]]*)\)|\[([^()\[\]]*)\]", rest)]
    nums = next((p for p in picks if len(p) in (3, 7)), None) or _re.findall(num, rest)
    nums = nums[:3]
    if len(nums) < 3 or _bi.any(float(x) != int(float(x)) for x in nums):
        return None
    return tuple(int(float(x)) for x in nums), direct


# ───────────────────────── 채점 상태 · 그림 ─────────────────────────
_FIGS = []                                 # 「PROMPT 셀」이 만든 그림을 모으는 자리
_STATE = {"tag": None, "buf": None, "out": None, "ids0": set(), "src": "", "prompt": None, "pub": None,
          "lost": None, "imgs": 0}
_TRY = {}                                  # 미션마다 ❌ 가 몇 번째인지 — 2단 힌트·기준 범위·빠진 단어가 이 수를 봅니다
_orig_figure = getattr(_plt.figure, "_orig", _plt.figure)   # 이 셀을 다시 실행해도 겹겹이 싸이지 않게


def _figure(*a, _orig=_orig_figure, _figs=_FIGS, **k):
    # 학생 코드가 그림을 만들 때 불립니다 — 채점 코드의 이름을 기본값으로 묶어 둡니다(학생 셀이 같은 이름을 만들어도 그대로)
    f = _orig(*a, **k)
    _figs.append(f)
    return f


_figure._orig = _orig_figure
_plt.figure = _figure                      # subplots · pandas · seaborn 도 이 경로를 거칩니다


def _get(name):
    """이름이 살아 있으면 그 값, 없으면 None — 아직 만들지 않은 이름에도 채점이 죽지 않게"""
    return globals().get(name)


def _device_spec(spec):
    """실행 경로(CPU · GPU)마다 기준이 다른 미션 — 기준의 `by_device` 에서 지금 DEVICE 의 한 벌을 덮어씁니다.

    준비 셀이 DEVICE 를 다시 정해도 채점 시점의 값을 따릅니다("cuda" 로 시작하면 GPU 기준, 그 밖은 CPU 기준).
    """
    per = (spec or {}).get("by_device")
    if not per:
        return spec
    dev = "cuda" if str(_get("DEVICE") or "cpu").startswith("cuda") else "cpu"
    return {**{k: v for k, v in spec.items() if k != "by_device"}, **dict(per.get(dev, {}))}


def _spec_of(key):
    """이 미션의 채점 기준 — 실행 경로별 기준을 맞추고(`_device_spec`), 권 설정의 `spec` 으로 이 세션의 기준을 다시 계산합니다"""
    spec = _device_spec(_ANS.get(key))
    hook = _BOOK.get("spec")
    if hook is not None and spec is not None:
        try:
            spec = hook(key, spec) or spec      # 이 세션에서 다시 계산한 기준(권 설정)
        except Exception:
            pass
    return spec


def _takes_ctx(fn):
    """권 검사 함수가 셋째 인자(출력 글 · 그림 · 실행 상태)를 받는 꼴인가 — `fn(key, spec, ctx)`. 옛 꼴은 `fn(key, spec)` 입니다"""
    code = getattr(fn, "__code__", None)
    if code is None:
        return False
    n = code.co_argcount - (1 if getattr(fn, "__self__", None) is not None else 0)
    return n >= 3 or _bi.bool(code.co_flags & 0x04)


def _ai_lines(src):
    """셀 원문에서 「여기부터 AI 코드」 표시선 아래의 코드 줄 — 주석은 뗍니다(표시선이 없으면 빈 목록).

    셀이 표시선 줄에서 끝나면(줄바꿈 없이 — 코드를 받기 전의 PROMPT 셀) 표시선 줄의 나머지 글자(「 ---」)를 코드 줄로 돌려주지 않습니다.
    """
    src = str(src or "")
    if "여기부터 AI 코드" not in src:
        return []
    rest = src.split("여기부터 AI 코드", 1)[1]
    body = rest.split("\n", 1)[1] if "\n" in rest else ""
    return [ln.split("#", 1)[0] for ln in body.splitlines()]


def _open_figs():
    """지금 열려 있는 그림들 — 래퍼를 거치지 않고 읽습니다"""
    try:
        return [_plt._pylab_helpers.Gcf.figs[n].canvas.figure for n in _plt.get_fignums()]
    except Exception:
        return []


def _num_txt(v, fmt=",.6g"):
    """채점 문장에 찍을 수 — 음수 부호는 권 설정(`minus`)을 따릅니다(본문이 − U+2212 를 쓰는 권)"""
    return format(v, fmt).replace("-", _BOOK.get("minus", "-"))


def _val_txt(val):
    """읽은 값(`_read_value` 결과)을 채점 문장에 찍을 글자 — 찍힌 자리까지 보여 줍니다.

    유효숫자 6자리로만 찍으면 학생이 출력한 「25,209.34」 가 「25,209.3」 으로, 「1,052,176.7」 이 「1.05218e+06」 으로 보여
    자리가 잘린 값으로 읽힙니다. 6자리로 줄였을 때 찍힌 자리의 숫자가 사라지면 찍힌 소수 자릿수대로 찍습니다.
    끝의 0 은 뗍니다(91.0 → 91). 소수 7자리 넘게 찍힌 값(반올림하지 않은 0.2812250064155164)은 지금처럼 유효숫자 6자리이고
    (지수 표기가 되는 큰 수만 소수 6자리), 지수 표기로 찍힌 값은 지수 표기입니다.
    """
    v, raw = val[0], str(val[1] or "")
    short = _num_txt(v)
    places = len(raw.split(".", 1)[1]) if "." in raw else 0
    if "e" in raw.lower() or v != v or v in (float("inf"), float("-inf")) or (places > 6 and "e" not in short):
        return short
    places = _bi.min(places, 6)
    full = format(v, ",." + str(places) + "f")
    full = full.rstrip("0").rstrip(".") if "." in full else full
    try:
        same = float(format(v, ",.6g").replace(",", "")) == float(full.replace(",", ""))
    except ValueError:
        return short
    return short if same else full.replace("-", _BOOK.get("minus", "-"))


_ANSI_RE = _re.compile(r"\x1b\[[0-9;?]*[ -/]*[@-~]")       # 굵게·색을 켜고 끄는 글자 — 화면에는 보이지 않습니다


def _clean_out(text, prompt=""):
    """채점이 읽을 출력 — 화면에 보이지 않는 글자를 지우고 PROMPT 를 그대로 찍은 글을 뺍니다.

    굵게 찍으려고 넣은 ANSI 코드(ESC[1m · ESC[0m)는 「[」 가 있어 열린 괄호로 읽히고, `print(PROMPT)` 로 찍은 지시문의
    「"이름 = 값" 한 행으로 출력하고, Q_ZS 의 0.5 분위수로 …」 는 이름 줄이 되어 0.5 가 값으로 읽힙니다.
    PROMPT 글은 값 줄이 아닙니다.
    """
    txt = _ANSI_RE.sub("", text or "")
    echo = str(prompt or "").strip()
    if len(echo) >= _PROMPT_MIN and echo in txt:
        txt = txt.replace(echo, "\n")
    return txt


# ───────────────────────── PROMPT 칸 ─────────────────────────
def _one_line(e, n=120):
    """예외를 한 줄로 — 트레이스백 대신 학생이 읽을 문장 하나만 남깁니다.

    문법 오류는 임시 파일 이름 대신 셀의 몇째 줄인지 적고, 예외의 문장을 읽다가 다시 오류가 나도
    채점 표가 닫히도록 그 자리는 안내 문구로 채웁니다.
    """
    try:
        if isinstance(e, SyntaxError) and getattr(e, "msg", None):
            txt = str(e.msg) + (f" (셀 {e.lineno}번째 줄)" if getattr(e, "lineno", None) else "")
        else:
            txt = str(e)
    except Exception:
        txt = "(오류 문장을 읽지 못했습니다)"
    txt = txt.replace("\n", " ").strip()
    return (txt[:n] + "…") if len(txt) > n else txt


def _prompt_in_cell(src):
    """셀 원문의 PROMPT 따옴표 안 글자 — 셀을 실행하기 전에 읽어 둡니다.

    이름 공간의 PROMPT 를 읽으면 PROMPT 블록을 지운 셀이 앞 미션에서 채운 PROMPT 로 통과하고, 문법 오류로
    셀이 돌지 않은 날은 채운 PROMPT 도 앞 값으로 보입니다. 셀에 PROMPT 대입이 없으면 "" 이고, 따옴표 꼴이 아니라
    읽지 못하면 None 입니다. 셀이 돌았으면 따옴표 안 글자 대신 실행된 PROMPT 값을 씁니다(`_post`).
    """
    src = str(src or "")
    if not _re.search(r"(?m)^[ \t]*PROMPT[ \t]*=(?!=)", src):
        return ""
    lit = _re.search(r"^[ \t]*PROMPT[ \t]*=[ \t]*[rRuU]?(\"\"\"|'''|\"|')(.*?)\1", src, _re.M | _re.S)
    return lit.group(2) if lit else None


def _prompt_now(given=None):
    """채점할 PROMPT 글자 — 셀 원문에서 읽은 것(given)이 있으면 그것, 없으면 이름 공간의 PROMPT"""
    if given is not None:
        return str(given)
    try:
        return str(_ip.user_ns.get("PROMPT", _get("PROMPT")) or "")
    except Exception:
        return str(_get("PROMPT") or "")


def _prompt1(n=70, given=None):
    """지금 셀의 지시 첫 줄 — 튜터에 보낼 질문에 그대로 싣습니다"""
    head_lines = [l.strip() for l in _prompt_now(given).split("\n") if l.strip()]
    return (head_lines[0][:n] + "…") if head_lines else "____"


_PROMPT_MIN = 30
_PROMPT_PLACEHOLDERS = ("", "목표", "쓸 변수", "출력", "목표 —", "쓸 변수 —", "출력 —", "TODO",
                        "여기에 내 말로 적습니다")


def _strip(txt):
    """빈칸·줄표·가운뎃점을 떼어 낸 알맹이 — 자리표시자 대조는 이 꼴로 합니다"""
    return _re.sub(r"[\s—·:,\-]+", "", str(txt or ""))


_PLACEHOLDERS_STRIPPED = {_strip(t) for t in _PROMPT_PLACEHOLDERS}


def _prompt_missing(given=None):
    """지시 입력란이 아직 내 말로 채워지지 않았으면 사유 한 문장, 채워졌으면 None"""
    txt = _prompt_now(given).strip()
    stripped = _strip(txt)                          # 길이보다 먼저 본다 — 견본을 그대로 둔 자리가 있다
    if not stripped:
        return "PROMPT 입력란이 비어 있습니다"
    if stripped in _PLACEHOLDERS_STRIPPED:
        return "PROMPT 입력란이 처음 적혀 있던 견본 그대로입니다"
    if len(txt) < _PROMPT_MIN:
        return f"PROMPT 입력란이 {len(txt)}자뿐입니다 (최소 {_PROMPT_MIN}자)"
    return None


def _reveal_show(txt):
    """완성본 코드와 해설 — **평문 그대로** 찍는다.

    LMS 렌더러 계약: 「🔓 정답이 열렸습니다 …」 줄부터 채점 표를 닫는 `─` 구분선 줄까지를
    출력 영역이 「🔓 완성본 풀이 보기」 상자로 접는다. 접힘의 표지가 이 **평문 마커**라
    HTML `details` 로 싣지 않는다 — 마커도 이 함수도 바꾸지 않는다.
    """
    print(txt)


# ───────────────────────── 오류 · 흔한 실수 ─────────────────────────
def _error_hint(key, error):
    """오류에 맞는 (힌트, 물을 것).

    문법 오류(SyntaxError·IndentationError)는 셀이 한 줄도 돌지 않은 것이라 붙여 넣은 글자를 짚습니다.
    노트북에 있는 이름을 잘못 옮겨 적은 NameError 면 그 이름을 짚고, 권 설정의 `error_hint(key, error)` 가 (힌트, 물을 것)을
    돌려주면 그것, 권의 힌트에 오류 종류별 힌트(`error_kinds`)가 있으면 그것, 없으면 미션 공통 힌트입니다
    (준비 셀·설치 셀이 만든 이름과 다시 실행할 셀).
    """
    kind, msg = type(error).__name__, _one_line(error, 400)
    tbl = _HINT.get(key, {})
    if isinstance(error, SyntaxError):
        line_no = getattr(error, "lineno", None)
        at = f"셀 {line_no}번째 줄" if line_no else "오류 문장이 가리키는 줄"
        head = "문법 오류라 이 셀의 코드가 한 줄도 실행되지 않았습니다. "
        if "```" in (_STATE.get("src") or ""):
            return (head + "AI 답에서 코드를 옮길 때 「```python」·「```」 같은 코드 블록 표시 행(펜스 줄)이 함께 들어왔습니다."
                    " 표시 행을 지우고 그 안쪽 코드만 남긴 뒤 다시 실행하세요",
                    "붙여 넣은 코드의 첫 행과 끝 행에 파이썬이 아닌 행이 있는지")
        if isinstance(error, (IndentationError, TabError)):
            return (head + f"들여쓰기가 어긋났습니다 — {at}과 바로 위 행의 들여쓰기 칸 수를 비교하고,"
                    " 같은 블록의 행은 같은 칸 수로 맞추세요",
                    "이 행의 들여쓰기가 위 행과 어떻게 어긋났는지")
        return (head + f"{at}에서 괄호·따옴표가 닫혔는지, 코드가 아닌 설명 문장이 섞였는지,"
                ' PROMPT 에 적을 문장을 PROMPT = """ """ 따옴표 밖에 두지 않았는지 보세요',
                "이 문법 오류가 어느 행에서 왜 났는지")
    if kind == "NameError":                # 노트북에 있는 이름을 잘못 옮겨 적은 경우(FIG_W → FGW)는 그 이름을 짚습니다
        import difflib
        bad = _re.search(r"name '(\w+)'", msg)
        near = bad and difflib.get_close_matches(
            bad.group(1), [n for n in globals() if not n.startswith("_") and n != "FIGW"], n=1, cutoff=0.7)
        if near:
            return (f"「{bad.group(1)}」 이름은 이 노트북에 없습니다. 비슷한 이름은 「{near[0]}」 입니다."
                    f" 잘못 옮겨 적었다면 코드에서 「{bad.group(1)}」 를 「{near[0]}」 로 고쳐 다시 실행하세요",
                    "이 오류가 어느 행에서 왜 났는지")
    hook = _BOOK.get("error_hint")          # 권 설정 — 셀 위쪽에 내가 적은 값의 꼴이 어긋나 난 오류처럼 권만 아는 까닭
    if hook is not None:
        try:
            own = hook(key, error)
        except Exception:
            own = None
        if own and own[0]:
            return own[0], (own[1] if len(own) > 1 and own[1] else "이 오류가 어느 행에서 왜 났는지")
    for rule in tbl.get("error_kinds", ()):
        if _bi.any(kind == k and _re.search(pat, msg) for k, pat in rule.get("match", ())):
            return rule.get("hint", ""), rule.get("ask") or "이 오류가 어느 행에서 왜 났는지"
    hint, question = tuple(tbl.get("error", ("", "이 오류가 어느 행에서 왜 났는지")))[:2]
    return hint, question


def _note_for(rules, values, tail_text):
    """값이 흔한 실수 하나의 값과 같으면 그 원인 문장 — 규칙은 (하한, 상한, 문장) · ("nan", 문장) (`_CAUSE` 블롭)"""
    for rule in rules or ():
        if rule[0] == "nan":
            if "nan" in (tail_text or "").lower():
                return rule[1]
        elif _bi.any(float(rule[0]) <= v <= float(rule[1]) for v in values):
            return rule[2]
    return ""


def _stale_code(rule, src, lo, hi):
    """값은 기준 범위 안인데 그 값을 코드에 손으로 적었고 shift 인자는 전임 코드의 틀린 값뿐이면 그 사실 한 문장.

    함정 미션에서 전임 코드의 `shift(24)` 를 그대로 둔 채 기준 범위의 숫자를 코드에 적어 출력하면 값만 맞고
    고친 곳이 없습니다. 그 꼴만 거릅니다(`_CAUSE["code"]` 규칙이 있는 미션만).
    """
    if not rule or not src:
        return ""
    body = src.split("여기부터 AI 코드", 1)[-1]           # PROMPT 안의 말은 보지 않습니다
    body = "\n".join(l.split("#", 1)[0] for l in body.splitlines())
    args = [_re.sub(r"\s+", "", a) for a in _re.findall(r"\.shift\(([^()]*)\)", body)]
    if not (args and _bi.all(a in rule[0] for a in args)):
        return ""
    if not _bi.any(lo <= v <= hi for v in _nums(body)):
        return ""
    return rule[1]


def _ask_line(form, name, prompt, question):
    """튜터에 보낼 질문 한 줄 — 권의 `_ASK_FORM` 틀에 채웁니다"""
    tmpl = _ASK_FORM.get(form) or _ASK_FORM.get("full") or "{name} — {ask}"
    return "  " + tmpl.format(name=name, prompt=_prompt1(given=prompt), ask=question)


_FORM_HINT = (                              # 이름·숫자를 못 찾은 ❌ — 값 계산이 아니라 출력 꼴을 짚습니다
    "값 계산보다 출력 꼴이 먼저입니다 — 이름과 값을 「이름 = 숫자」 꼴로 한 줄에 찍고, 이름은 「만들 것」 의 굵은 이름과"
    " 글자까지 같게 적도록 PROMPT 의 출력 형식에 적으세요. 표(DataFrame)·딕셔너리·배열로 띄웠다면 항목마다 print 한 줄로 바꿉니다",
    "이 값을 출력하는 행이 이름과 값을 어떤 꼴로 찍는가")
_DIGITS_HINT = (                            # 반올림하면 기준 범위에 닿는 값 — 계산보다 자릿수를 짚습니다
    "값은 맞았을 수 있습니다 — 출력 형식의 자릿수(:.2f 같은 서식)가 「만들 것」 보다 적어 반올림된 값이 찍혔습니다."
    " PROMPT 의 출력 형식에 「만들 것」 의 소수점 아래 자릿수를 적고 다시 실행하세요",
    "이 값을 소수점 아래 몇 자리로 출력하도록 코드가 정했는가")
_DUP_HINT = (                               # 같은 이름 줄이 여럿이고 값이 다르면 — 어느 줄이 최종 답인지 모릅니다
    "같은 이름으로 값을 두 번 이상 찍었습니다 — 전임 코드나 시험 삼아 찍은 옛 print 를 지우고, 최종 값 한 줄만 남기도록"
    " PROMPT 에 적으세요",                  # 함정 미션의 힌트(전임 담당자 코드가 있는 미션)
    "같은 이름을 찍는 print 가 코드 어디 어디에 있는가")
_DUP_HINT_PLAIN = (                         # 함정이 아닌 미션 — 비교 · 중간 결과를 출력하며 같은 이름을 쓴 경우를 짚습니다(0.4.2, 분야 단어 없이)
    "비교 결과나 중간 결과를 출력할 때도 이 이름을 쓰면, 함께 출력한 수도 이 항목의 값으로 읽습니다. 이름은 최종 값을 출력할 때만"
    " 쓰도록 PROMPT 에 적으세요",
    _DUP_HINT[1])


def _hint_key(key, kind, label):
    """❌ 의 종류에 맞는 힌트 열쇠 — 권의 힌트(`_HINT`)에 그 종류의 열쇠가 있으면 그것, 없으면 공통 힌트"""
    own = _HINT.get(key, {})
    pick = {"form": ("이름_형식", "label"), "dup": ("줄_중복", "dup"), "digits": ("round",), "unit": ("unit",)}
    for cand in pick.get(kind, ()):
        if cand in own:
            return cand
    return {"form": "_form", "dup": "_dup", "digits": "_digits"}.get(kind, label)


def _is_trap(key):
    """함정 미션인가 — 빈 PROMPT 에도 잠그지 않고 표시선 아래 전임 담당자 코드를 진단하는 미션(권 토막의 `_blocks`)"""
    try:
        return not _blocks(key, "PROMPT 입력란이 비어 있습니다")
    except Exception:
        return False


def _hint_parts(tiers):
    """권의 힌트 한 칸 → (힌트 단 목록, 튜터에 물을 것) — 원소가 둘이면 (힌트, 물을 것), 셋 이상이면 마지막이 물을 것"""
    tiers = (tiers,) if isinstance(tiers, str) else tuple(tiers or ())
    steps = tiers[:-1] if len(tiers) > 2 else tiers[:1]
    return [t for t in steps if t], (tiers[-1] if len(tiers) > 1 else "")


# ───────────────────────── 채점 표 ─────────────────────────
def judge(key, text="", figs=(), error=None, prompt=None, shown=None):
    """출력 텍스트와 그림을 기준과 대조해 채점 표를 찍습니다 — 통과하면 완성본 풀이를 엽니다.

    prompt 는 셀 원문에서 읽은 PROMPT(없으면 이름 공간의 PROMPT), shown 은 그림이 화면에 떴는지
    (True·False, 알 수 없으면 None — inline 백엔드에서만 셉니다)입니다.
    """
    spec = _spec_of(key)                   # 실행 경로별 기준(by_device) · 이 세션에서 다시 계산한 기준(권 설정)
    if spec is None:                       # 기준이 아직 없는 미션은 조용히 넘어갑니다
        return None
    head_label = ("미션 " + key[1:]) if key[:1] == "m" and key[1:].isdigit() else key
    failed, passed = [], True
    print("\n" + "─" * 54)
    print(f"◼ {head_label} 채점 결과")
    if spec.get("path_note"):              # 실행 경로(CPU · GPU)마다 기준이 다른 미션 — 어느 기준으로 채점하는지 먼저 알립니다
        print(f"ⓘ 채점 기준: {spec['path_note']}")
    reason = _prompt_missing(prompt)                  # 값·그래프보다 먼저 — 내 말로 시킨 지시가 있는가
    _locked = _bi.bool(reason)
    if _blocks(key, reason):                       # 권 토막의 `_blocks` — 함정 미션은 빈 PROMPT 에도 진단합니다
        print(f"❌ {reason} — 짧게 시켜도 됩니다. 다만 무엇을 어떤 단위로 출력할지까지 적으면 한 번에 통과할 확률이 높습니다")
        print("   PROMPT 입력란을 채우기 전에는 값·그래프 채점도, 완성본 풀이도 열지 않습니다")
        print('   AI 튜터 대화창이 아니라 이 셀의 PROMPT = """ """ 따옴표 안에 적었는지 확인하세요')
        print("─" * 54)
        return False
    elif reason:                                   # 함정 미션은 PROMPT 가 비어도 표시선 아래 코드를 진단합니다(지시서 §9-10 · ⓘ 한 줄)
        print(f"ⓘ {reason} — 아래 채점 결과는 전임 담당자 코드의 진단입니다. 완성본 풀이는 PROMPT 를 채운 뒤에 열립니다")
        print("   (표시선 아래를 이미 고쳤다면 아래는 AI 코드 표시선 아래 코드의 진단입니다)"
              ' 고칠 지시는 AI 튜터 대화창이 아니라 이 셀의 PROMPT = """ """ 따옴표 안에 적습니다')
    form = _ASK.get(key, "full")
    if error is not None:                   # 오류가 나면 값·그래프 채점은 뜻이 없습니다
        kind = type(error).__name__                  # 괄호가 안 닫힌 셀은 3.13 에서 _IncompleteInputError 입니다
        kind = "SyntaxError" if isinstance(error, SyntaxError) and kind.startswith("_") else kind
        msg = _one_line(error)
        print(f"❌ 실행 오류 — {kind}" + (f": {msg}" if msg else ""))
        hint, question = _error_hint(key, error)          # 비슷한 이름이 있으면 그 이름, 없으면 미션 공통 힌트
        if hint:
            print(f"   힌트 : {hint}")
        print(_ASK_HEAD.get(form, ""))
        print(_ask_line(form, "실행 오류", prompt, question))
        print("─" * 54)
        return False
    txt = _clean_out(text, _prompt_now(prompt))
    trap = _is_trap(key)
    cell_src = str(_STATE.get("src") or "")
    if ("여기부터 AI 코드" in cell_src and not _bi.any(l.strip() for l in _ai_lines(cell_src))
            and not txt.strip() and not _bi.any(_axes(f) for f in figs)):
        # PROMPT 만 적고 ▶ 한 셀 — 틀린 것이 아니라 아직 코드를 받지 않은 것입니다. 값 · 그래프 ❌ 와 힌트를 늘어놓지 않고
        # 다음 할 일 한 줄만 알립니다(❌ 횟수에 넣지 않습니다)
        print("ⓘ 출력이 없습니다 — 「여기부터 AI 코드」 표시선 아래에 코드가 없습니다. "
              + ("전임 담당자 코드를 지웠다면 실행 취소로 되돌린 뒤, " if trap else "")
              + 'AI 튜터에 "위 PROMPT 대로 코드를 써 줘" → 「이 셀에 넣기」 → ▶ 하면 그 코드의 출력과 그림을 채점합니다')
        print("─" * 54)
        return False
    if not txt.strip():
        print("❌ 출력이 없습니다 — 값을 print 하도록 프롬프트의 출력 형식에 적으세요")
    values = spec.get("values", {})
    units = [str(ref[2]) for ref in values.values() if len(ref) > 2]
    val_lines = set(_value_lines(txt, units))               # 단위·「=」 가 붙은 행 — 이름 바로 뒤 값 줄이 없을 때 먼저 씁니다
    all_lines = [l for l in txt.splitlines() if l.strip()]
    series = _bi.bool(_SERIES_RE.search(txt))               # Series 를 통째로 찍은 흔적
    need_words = spec.get("need_words", {})        # 함정 미션 — {이름: [단어, …]} 그 항목의 토막에 꼭 있어야 하는 단어
    cause = _get("_CAUSE") or {}
    fail_notes = cause.get("fail", {}).get(key, {})            # 흔한 실수의 값 → 원인 문장
    pass_notes = cause.get("pass", {}).get(key, {})
    code_notes = cause.get("code", {}).get(key, {})
    find = {l: (r[3] if len(r) > 3 and r[3] else l) for l, r in values.items()}   # 출력에서 찾을 이름(짧은 이름 포함)
    names = list(find.values()) + (["공식 차수"] if spec.get("차수_표기") else [])
    skip = _BOOK.get("skip", {})
    own_too = []                                   # own_too: 출력 꼴 힌트에 그 항목의 권 힌트를 함께 낼 이름(아래 `not occ`)
    form_only = spec.get("form_only")              # 이름 줄이 없는 칸에 출력 꼴 힌트만 내는 미션(권 힌트 첫 단이 읽은 값을 전제로 쓴 글)
    forms = spec.get("forms", {})                  # {이름: "단어 숫자 %"} 값과 단어를 함께 찍는 칸의 출력 꼴
    paren_unit = spec.get("paren_unit", ())        # 「k 건 (r %)」 꼴로 찍는 칸 — 괄호 안의 기준 단위 수를 읽습니다(`_unit_value`)
    traps = spec.get("traps", {})                  # {이름: [(하한, 상한, 문장), …]} 흔한 실수의 값이면 그 실수를 짚는 힌트
    seen = {}                                      # 값 칸에서 읽은 값 — 권 검사가 값끼리 맞물리는지 볼 때 씁니다(ctx["values"])
    made = "고칠 것" if trap else "만들 것"          # 채점할 이름이 적힌 본문 소제목 — 함정 미션은 「고칠 것」, 다른 미션은 「만들 것」
    for label, ref in values.items():
        skip_if, skip_msg = skip.get((key, label), (None, ""))
        if skip_if is not None and skip_if():           # 대체 탐색을 쓴 날의 AIC 처럼 채점하지 않는 칸
            print(f"ⓘ {label} — {skip_msg}")
            continue
        lo, hi = float(ref[0]), float(ref[1])
        unit = ref[2] if len(ref) > 2 else ""
        name = find[label]
        want_form = f"「{name} = {forms.get(label) or ('숫자' + ((' ' + unit) if unit else ''))}」"
        where = "이 이름이 있는 줄" if name == label else f"이 이름(「{name}」)이 있는 줄"
        # 이 이름이 있는 줄에서 이름 뒤의 값 하나만 읽습니다. 이름이 있는 줄이 없을 때 출력 전체의 숫자로
        # 물러나면 다른 행의 값(부호를 뒤집은 차, 다른 모델의 값)으로 틀린 값이 ✅ 가 됩니다
        occ = _occurrences(all_lines, name, names, series, unit)   # 「이름 = 값」·「이름: 값」·「이름 값」 — 모든 행에서
        # 같은 이름 줄이 여럿이면 이름 바로 뒤 값 줄만 봅니다(`_value_use`). 그 줄들의 값이 모두 맞아야 통과입니다 — 하나만
        # 맞아도 통과시키면 옛 print 를 남긴 채 덧붙인 줄·부호를 뒤집은 줄이 헛통과합니다.
        use = _value_use(occ, val_lines)
        vals = [o["val"] for o in use]
        if label in paren_unit:                         # 「이름 = 14 건 (1.4 %)」 — 괄호 안의 기준 단위 수가 값입니다
            vals = [_unit_value(v, unit) for v in vals]
        in_range = [v for v in vals if lo <= v[0] <= hi]
        clash = [v for v in vals if not (lo <= v[0] <= hi) and _short_digits(v[1], v[0], lo, hi) is None]
        # 이름 바로 뒤 값 줄이 하나도 없어 「이름 = 값」 꼴이 아닌 줄에서 읽은 값 — 이름 앞에 다른 말이 붙은 줄(더 긴 이름일 수 있는 줄:
        # 「상태별 리뷰 수 : 통과 178 …」)이나, 값 표지 없이 이름과 숫자 사이에 다른 말이 있는 줄(결론 문장 「…하루 평균은 9월의
        # 0.933배로…」 의 9). 범위 안이면 그 값을 받고(「전체 리뷰 수 = 200 건」 · 「리뷰 수는 200 건입니다」), 범위 밖이면 이 이름의
        # 값으로 읽지 않습니다 — 다른 항목의 값이나 문장 속의 수로 「읽은 값은 … 범위 밖」 과 계산 힌트를 내면 거짓 원인이 됩니다.
        # 이름과 값 표지 사이에 설명이 낀 줄(「이름 — period=24, robust=False → 0.8115」)은 지금처럼 그 값을 읽습니다
        weak = _bi.bool(use) and (use[0]["pre"] or (not use[0]["val"][3] and use[0]["val"][5] == "bare"))
        foreign = use[0]["line"].strip() if weak and not in_range else ""
        if vals and not foreign:
            seen[label] = (in_range or vals)[0][0]
        # 개수 자리에 찍은 비율 — 기준 단위가 % 로 시작하는 칸(% · %p)은 값 뒤의 % 가 제 단위입니다
        pct = (next((v for v in in_range if v[2].lstrip().startswith("%")), None)
               if unit and not unit.startswith("%") else None)
        stale = _stale_code(code_notes.get(label), _STATE.get("src"), lo, hi) if in_range else ""
        words = need_words.get(label, ())
        words = (words,) if isinstance(words, str) else tuple(words)
        missing = []
        if in_range and words:          # 값은 기준 범위 안 — 그 항목의 토막에 필요한 단어가 있는가
            # 값 칸 하나만 고치고 전임자의 틀린 결론 문구(「→ 정상」)를 그대로 둔 줄이 통과하지 않게 합니다.
            # 줄 전체가 아니라 이 항목의 토막 — 같은 줄의 다른 항목 단어는 보지 않습니다
            rows = []
            for o in _occurrences(all_lines, name, names, series):
                flat = _re.sub(r"\s+", "", o["seg"])
                rows += [flat, _re.sub(r"\([^()]*\)", "", flat)]
            missing = [w for w in words if not _bi.any(_word_in(row, w) for row in rows)]
        if in_range and pct is not None:              # 개수 자리에 비율을 찍은 판 — 범위가 넓어 수만으로는 못 거릅니다
            passed = False
            failed.append((label, _hint_key(key, "unit", label)))
            print(f"❌ {label} — {where}에서 읽은 값 {_val_txt(pct)} 뒤에 % 가 붙어 있습니다."
                  f" 비율이 아니라 {want_form} 처럼 개수로 출력하세요")
        elif in_range and clash:                      # 같은 이름, 다른 값
            passed = False
            failed.append((label, _hint_key(key, "dup", label)))
            shown_vals = " · ".join(_val_txt(v) for v in vals)
            print(f"❌ {label} — {where}이 {len(vals)}개이고 값이 서로 다릅니다({shown_vals})."
                  f" 틀린 값을 찍는 옛 줄을 지우고 {want_form} 한 줄만 남기세요")
        elif in_range and stale:                      # 값만 손으로 적고 코드는 그대로 둔 판
            passed = False
            failed.append((label, label))
            print(f"❌ {label} — 출력에서 {_val_txt(in_range[0])} 를 찾았지만 {stale}")
        elif in_range and missing:
            passed = False
            failed.append((label, label))
            _prev_fail = _TRY.get(key, 0)              # 첫 ❌ 에는 빠진 단어를 감춥니다
            print(f"❌ {label} — 값 {_val_txt(in_range[0])} 는 기준 범위 안이지만, 이 이름이 있는 줄의 문구가 기준과 다릅니다"
                  + (" (그 줄에 있어야 할 단어 : " + " · ".join(f"「{w}」" for w in missing) + ")"
                     if _prev_fail else " — 무엇을 기각했고 어느 쪽 증거인지 같은 줄에 적었는지 보세요"))
        elif in_range:
            print(f"✅ {label} — 출력에서 찾은 값: {_val_txt(in_range[0])}")
            note = _note_for(pass_notes.get(label, ()), [in_range[0][0]], "")
            if note:
                print(f"   ⓘ {note}")
            if in_range[0][4] == "series":           # f-string 안의 Series 는 뒤에 적은 단위가 dtype 줄 끝으로 밀립니다 —
                # 단위 ⓘ 까지 겹치면 적은 단위를 안 적었다고 읽히므로, 출력 꼴을 고치라는 ⓘ 하나만 냅니다
                print("   ⓘ 표(Series)를 그대로 넣어 인덱스와 값이 함께 찍혔습니다 — 값만 읽었습니다."
                      " float(…) 로 숫자 하나만 출력하면 보고서에 그대로 옮길 수 있습니다")
            elif unit and unit not in " ".join(o["tail"] for o in use):   # 단위는 그 이름 줄에서 봅니다
                print(f"   ⓘ 단위 「{unit}」 도 값과 함께 출력하면 보고서에 그대로 옮길 수 있습니다")
        elif not occ or foreign:
            passed = False
            # 계산 힌트가 아니라 출력 꼴 힌트 — 함정 미션의 단어 항목은 행 이름·문구를 함께 안내하는 항목 힌트 그대로
            form_key = _hint_key(key, "form", label)
            failed.append((label, label if words else form_key))
            # 함정 미션의 전임 담당자 코드가 이 이름의 줄을 아예 찍지 않는 권이 있습니다 — 공통 출력 꼴 힌트만 내면 그 함정을
            # 짚는 항목 힌트와 튜터에 물을 것이 사라지므로, 권이 그 항목에 적어 둔 힌트 첫 단과 물을 것을 함께 냅니다.
            # 두 경우는 지금처럼 출력 꼴 힌트만 냅니다 — 권에 출력 꼴 힌트(이름_형식 · label)가 따로 있으면 그 힌트와 물을
            # 것이 이미 권의 것이고, 단어를 채점하는 함정 미션은 전임 담당자 코드가 이름 줄을 이미 찍으며 단어 없는 항목의
            # 힌트가 그 줄의 결론 문구를 짚는 글이라 이름 줄이 없을 때 내면 사실과 다를 수 있습니다
            # 권 힌트 첫 단이 읽은 값을 전제로 쓴 글이면(기준값을 예로 든 글) 이름 줄이 없는 화면에 내지 않습니다 — 기준의 `form_only`
            if (trap and not need_words and not form_only and form_key == "_form"
                    and _hint_parts(_HINT.get(key, {}).get(label))[0]):
                own_too.append(label)
            print(f"❌ {label} — {where}을 출력에서 찾지 못했습니다. {want_form} 한 줄로 출력하세요")
            if foreign:
                print(f"   ⓘ 「{foreign[:60]}」 은 「이름 = 값」 꼴의 줄이 아니라(이름 앞뒤에 다른 말이 있습니다) 이 이름의 값으로 읽지 않았습니다")
        elif not vals:
            passed = False
            note = _note_for(fail_notes.get(label, ()), [], " ".join(o["tail"] for o in occ))   # 「= nan」 같은 흔한 실수
            failed.append((label, label if words or note else _hint_key(key, "form", label)))
            if _bi.any(o["boxed"] for o in occ):     # 배열로 찍힌 값 — 값이 아니라 출력 꼴의 문제입니다
                print(f"❌ {label} — {where}에서 숫자를 찾지 못했습니다 — 값이 대괄호 [ ] 안의 배열이라 숫자 하나로 읽지 못했습니다")
                print(f"   ⓘ 값이 배열·Series 꼴로 찍혔습니다 — float(값) 이나 .item() 으로 꺼낸 숫자 하나를"
                      f" {want_form} 한 줄로 출력하세요")
            else:
                print(f"❌ {label} — {where}에서 숫자를 찾지 못했습니다. {want_form} 한 줄로 출력하세요")
                if _bi.any(_re.search(r"[(\[][^()\[\]]*\d", o["tail"]) for o in occ):
                    print("   ⓘ 이 줄의 숫자는 괄호 안에만 있습니다 — 괄호 안은 설명으로 보고 값으로 읽지 않습니다."
                          f" 값을 괄호 밖에 두고 {want_form} 꼴로 출력하세요")
            if note:
                print(f"   원인 : {note}")
        else:
            passed = False
            first = vals[0]
            _prev_fail = _TRY.get(key, 0)              # 첫 ❌ 에는 기준 범위를 감춥니다
            short = _short_digits(first[1], first[0], lo, hi)
            if short is not None:
                done = "정수로만" if short == 0 else f"소수점 아래 {short}자리까지만"
                got_txt = (f"{where}에서 읽은 값은 {_val_txt(first)} 인데 기준 범위 밖입니다 — 반올림한 자리가 모자라"
                           f" 범위 밖으로 읽혔습니다({done} 찍힘). 「{made}」 에 적힌 소수점 아래 자릿수대로 출력하세요")
            else:
                got_txt = f"{where}에서 읽은 값은 {_val_txt(first)} 인데 기준 범위 밖입니다"
            ref_txt = spec.get("range_text", {}).get(label)      # 한쪽이 열린 기준(「0 보다 큼」)은 범위 대신 말로 적습니다
            band = (f"기준: {ref_txt}" if ref_txt else f"기준 {_num_txt(lo, 'g')}" if lo == hi
                    else f"기준 범위 {_num_txt(lo, 'g')} ~ {_num_txt(hi, 'g')}")
            print(f"❌ {label} — {got_txt}" + (f" ({band})" if _prev_fail else ""))
            note = "" if short is not None else _note_for(fail_notes.get(label, ()), [v[0] for v in vals],
                                                          " ".join(o["tail"] for o in use))
            if note:                                 # 흔한 실수 하나와 같은 값이면 그 원인을 바로 말합니다
                print(f"   원인 : {note}")
            trap_note = "" if short is not None else _note_for(traps.get(label, ()), [v[0] for v in vals], "")
            if trap_note:                            # 기준에 적어 둔 흔한 실수의 값(traps)이면 그 실수를 짚습니다
                print(f"   힌트 : {trap_note}")
            if short is None and _bi.any(_re.search(r"\[|\barray\(|dtype|\bnp\.\w+\(", o["tail"]) for o in use):
                print("   ⓘ 값이 배열·Series 꼴로 찍혔습니다 — float(값) 이나 .item() 으로 숫자 하나만 출력하세요")
            elif short is None and first[5] == "bare":
                print(f"   ⓘ 이 줄은 이름 바로 뒤에 「=」·「:」 가 없어 이름 뒤의 숫자를 값으로 읽었습니다."
                      f" {want_form} 꼴로 출력하면 「=」 뒤 값을 읽습니다")
            elif short is None and first[5] == "marker" and _bi.any(lo <= v <= hi for v in _nums(first[2])):
                # 「= 182/200 = 91.0%」·「= 512일 가운데 0 일」·「= 2개 원점 평균 0.271」 — 값은 같은 줄 뒤쪽에 있습니다.
                # 계산이 아니라 출력 꼴의 문제일 수 있다는 것을 알려 계산부터 의심하지 않게 합니다. 뒤 숫자가 우연히 범위 안일
                # 수도 있어(「차 = -0.210 (MSE 0.536, MAE 0.326)」 의 0.326) 「값이 그 뒤에 있다면」 으로 적습니다
                print(f"   ⓘ 이 줄은 「=」·「:」 뒤의 첫 숫자({first[1]})를 값으로 읽었습니다 — 같은 줄의 뒤 숫자는 읽지 않습니다."
                      f" 값이 그 뒤에 있다면 계산식·설명에 든 숫자는 빼고 {want_form} 꼴로 값 하나만 출력하세요")
            failed.append((label, _hint_key(key, "digits", label) if short is not None else label))
    order_tag = spec.get("차수_표기")               # 적합만 고치고 출력 문자열을 그대로 두면 ✅ 가 열리던 자리
    if order_tag:
        order_label = "공식 차수"                   # 이 이름이 있는 줄의 「=」·「:」 뒤에서 고친 차수 표기를 찾습니다
        order_occ = _occurrences(all_lines, order_label, names)   # 「공식 차수 AIC」 줄은 건너뜁니다
        want_order = tuple(int(x) for x in _re.findall(r"\d+", order_tag))
        got = [r for r in (_order_of(o["tail"]) for o in order_occ) if r]
        # 이름 바로 뒤 값 줄(「공식 차수 = …」)이 있으면 그 줄만 — 「공식 차수 탐색: …」 같은 줄의 숫자는 차수가 아닙니다
        read = [r[0] for r in ([r for r in got if r[1]] or got)]
        many = list(dict.fromkeys(read))
        order_form = f"「{order_label} = (p,d,q)(P,D,Q,m)」 한 줄로 출력하세요"
        if len(many) == 1 and many[0] == want_order:
            print(f"✅ 차수 표기 — 「{order_label}」 줄이 고친 차수를 적고 있습니다")
        else:
            passed = False
            if len(many) > 1:
                failed.append(("차수 표기", _hint_key(key, "dup", "차수 표기")))
                print(f"❌ 차수 표기 — 「{order_label}」 줄이 {len(read)}개이고 적힌 차수가 서로 다릅니다"
                      f" ({' · '.join(str(o) for o in many)}). 최종 차수 한 줄만 남기세요")
            elif many:
                failed.append(("차수 표기", "차수_표기"))
                print(f"❌ 차수 표기 — 「{order_label}」 줄에 적힌 비계절 차수 {many[0]} 가 고친 차수와 다릅니다")
            else:
                failed.append(("차수 표기", _hint_key(key, "form", "차수 표기")))
                print("❌ 차수 표기 — " + (f"「{order_label}」 줄에서 숫자 3개로 된 차수를 읽지 못했습니다. {order_form}"
                                        if order_occ else f"이 이름이 있는 줄을 출력에서 찾지 못했습니다. {order_form}"))
    flat_lines = [_flat(l) for l in all_lines]
    forbid_all = [str(p) for p in spec.get("forbid", ())]
    forbid_seen = set()
    for phrase in forbid_all:                        # 함정 미션 — 전임자의 결론 줄이 남으면 통과시키지 않습니다
        # 빈칸만 다른 문구(「판단 :」·「판단:」)는 빈칸을 걷으면 같은 문구입니다 — 한 번만 알립니다
        same = _flat(phrase)
        if not same or same in forbid_seen:
            continue
        # 「…넣어도 되지 않습니다」 처럼 서술로 부정해 인용한 줄은 남은 줄이 아닙니다(`_forbid_in`)
        if _bi.any(_forbid_in(row, phrase) for row in flat_lines):
            forbid_seen.add(same)
            passed = False
            if ("남은 줄", "forbid") not in failed:
                failed.append(("남은 줄", "forbid"))
            # 알리는 글자는 출력에 찍힌 꼴 — 같은 문구의 여러 꼴 가운데 출력에 글자 그대로 있는 것, 없으면 먼저 적힌 것
            shown_as = next((p for p in forbid_all if _flat(p) == same and p in txt), phrase)
            print(f"❌ 남은 줄 — 출력에 「{shown_as}」 문구가 남아 있습니다")
    plot_spec = spec.get("plot")
    if plot_spec:
        axs = [a for f in figs for a in _axes(f)]
        if not axs:
            passed = False
            failed.append(("그래프", "plot"))
            print("❌ 그래프 — 그림을 찾지 못했습니다 (그래프 한 장도 같은 프롬프트에 함께 시킵니다."
                  " 채점은 matplotlib 으로 그린 그림을 읽습니다)")
        else:
            gaps = None
            for ax in axs:                 # 축이 여럿이면 기준에 맞는 축이 하나만 있어도 통과입니다
                try:
                    this_gaps = _plot_gaps(ax, plot_spec)
                except Exception as e:
                    this_gaps = [f"그림의 축 이름을 읽지 못했습니다 ({type(e).__name__})"]
                if not this_gaps:
                    gaps = []
                    break
                if gaps is None or len(this_gaps) < len(gaps):
                    gaps = this_gaps
            axis_words = "가로축·세로축 이름이" if plot_spec.get("축라벨") or not plot_spec.get("가로축라벨") else "가로축 이름이"
            if gaps:
                passed = False
                failed.append(("그래프", "plot"))
                print("❌ 그래프 — " + " · ".join(gaps))
            elif shown is False:                    # inline 백엔드인데 이 셀이 그림을 한 장도 화면에 내보내지 않았습니다
                print(f"✅ 그래프 — 그림 객체에 {axis_words} 있습니다")
                print("   ⓘ 그림이 화면에 뜨지 않았습니다 — plt.close() 나 savefig 로만 끝냈다면 plt.show() 로 띄워"
                      " 눈으로도 확인하세요")
            else:                                   # 그래프 종류는 가리지 않고 그림과 축 이름만 봅니다
                print(f"✅ 그래프 — 그림이 있고 {axis_words} 있습니다")
    ctx, notices = None, []
    for check in _BOOK.get("checks", ()):            # 권 설정의 검사(계획표의 열 · 단어 · 저장한 파일처럼 값·그림 밖의 결과물)
        try:
            if _takes_ctx(check):                    # fn(key, spec, ctx) — 출력 글 · 그림 · 실행 상태를 받는 꼴
                if ctx is None:
                    ctx = {"text": txt, "lines": all_lines, "figs": list(figs), "axes": [a for f in figs for a in _axes(f)],
                           "values": seen, "failed": failed, "attempt": _TRY.get(key, 0), "locked": _locked, "trap": trap,
                           "src": str(_STATE.get("src") or ""), "code": _ai_lines(_STATE.get("src")),
                           "prompt": _prompt_now(prompt), "shown": shown,
                           "read": lambda name, others=(): [o["val"][0] for o in _value_use(
                               _occurrences(all_lines, name, names + list(others), series), val_lines)],
                           "tails": lambda name, others=(): _text_tails(all_lines, name, names + list(others))}
                rows = list(check(key, spec, ctx) or ())
            else:                                    # fn(key, spec) — 옛 꼴
                rows = list(check(key, spec) or ())
        except Exception as e:                       # 검사가 예외로 끝나도 채점 표는 끝까지 찍습니다
            rows = [("채점 검사", False, f"이 미션의 검사 하나를 마치지 못했습니다 ({type(e).__name__}: {_one_line(e, 80)})."
                     " 이 미션의 준비 셀을 실행했는지 보고, 그래도 같으면 맨 위 「[채점] 이 셀을 먼저 실행하세요」 셀을 다시 실행한 뒤"
                     " 이 셀을 다시 실행하세요", "_check")]
        for row in rows:
            name, ok, why, hkey = row[:4]
            more = row[4] if len(row) > 4 else ()
            if ok is None:                           # 알림 — 틀린 것이 아니라 다음 할 일(❌ 로 세지 않고 풀이는 열지 않습니다)
                notices.append(why if not name else f"ⓘ {name} — {why}")
                continue
            if ok:
                print(f"✅ {name} — {why}")
            else:
                passed = False
                if (name, hkey) not in failed:       # 같은 항목이 두 번 ❌ 여도 힌트는 한 번만
                    failed.append((name, hkey))
                print(f"❌ {name} — {why}")
            for line in ((more,) if isinstance(more, str) else more or ()):   # 그 줄에 덧붙이는 줄(ⓘ 참고 · 힌트)
                print(f"   {line}")
    tbl = dict(_HINT.get(key, {}))
    tbl.setdefault("_form", tuple(t.replace("「만들 것」", f"「{made}」") for t in _FORM_HINT))     # 공통 힌트가 가리키는
    tbl.setdefault("_digits", tuple(t.replace("「만들 것」", f"「{made}」") for t in _DIGITS_HINT))  # 소제목을 이 미션의 것으로
    tbl.setdefault("_dup", _DUP_HINT if trap else _DUP_HINT_PLAIN)
    attempt = _TRY[key] = _TRY.get(key, 0) + (1 if failed else 0)
    groups = {}                                    # 같은 힌트(출력 꼴 등)는 한 번만 — 그 힌트가 걸린 항목 이름을 모아 찍습니다
    for name, hkey in failed:
        groups.setdefault(hkey, []).append(name)
    for hkey, members in groups.items():
        tiers = tbl.get(hkey, ("",))
        tier = tiers[:-1] if len(tiers) > 2 else tiers[:1]   # 2단 힌트는 마지막 원소가 튜터에 물을 것입니다
        hint = tier[_bi.min(attempt, len(tier)) - 1] if tier else ""
        if hint:
            print(f"   힌트({'출력 꼴' if hkey == '_form' else ' · '.join(members)}) : {hint}")
        if hkey == "_form":                        # 함정 미션 — 출력 꼴 힌트 뒤에 그 항목의 권 힌트 첫 단
            for name in members:
                if name in own_too:
                    print(f"   힌트({name}) : {_hint_parts(tbl.get(name))[0][0]}")
    if failed:
        print(_ASK_HEAD.get(form, ""))
        tmpl = _ASK_FORM.get(form) or ""
        for hkey, members in groups.items():
            # 함정 미션에서 이름 줄이 없던 항목은 권이 그 항목에 적어 둔 물을 것을 씁니다(없는 줄의 출력 꼴을 묻지 않게)
            own_ask = [m for m in members if hkey == "_form" and m in own_too and _hint_parts(tbl.get(m))[1]]
            rest = [m for m in members if m not in own_ask]
            if rest:
                question = (tbl.get(hkey, ("", ""))[-1] if len(tbl.get(hkey, ("",))) > 1 else "") \
                    or "이 항목이 왜 ❌ 로 남는지"
                joined = "」·「".join(rest) if "「{name}」" in tmpl else " · ".join(rest)
                print(_ask_line(form, joined, prompt, question))
            for name in own_ask:
                print(_ask_line(form, name, prompt, _hint_parts(tbl.get(name))[1]))
    elif notices:                                  # 틀린 것은 없고 다음 할 일이 남았습니다(두 번에 나눠 푸는 미션의 1회차)
        for line in notices:
            print(line)
    elif _locked:                                  # 값·그림은 기준에 맞지만 PROMPT 칸이 아직 비었습니다
        print('ⓘ 값과 그림은 기준에 맞습니다 — 이 셀의 PROMPT = """ """ 따옴표 안에 고칠 지시를 내 말로 쓰고'
              " 다시 실행하면 완성본 풀이가 열립니다")
    else:
        print("🔓 정답이 열렸습니다 — 아래 「🔓 완성본 풀이 보기」를 펼쳐 내 코드와 비교해 보세요")
        _reveal_show(_b64.b64decode(_REVEAL[key]).decode("utf-8"))
    print("─" * 54)
    return passed and not _locked and not notices


# ───────────────────────── PROMPT 셀 훅 ─────────────────────────
class _Tee:
    """「PROMPT 셀」의 출력을 화면과 버퍼 양쪽으로 흘려보냅니다"""

    def __init__(self, orig, buf):
        self._o, self._b = orig, buf

    def write(self, s):
        self._b.write(s)
        return self._o.write(s)

    def flush(self):
        self._o.flush()

    def __getattr__(self, n):
        return getattr(self._o, n)


def _tag(src):
    """머리 행이 `# 미션 N — 프롬프트`·`# 과제 — 프롬프트` 인 셀만 채점 대상입니다.

    머리 행은 비어 있지 않은 앞 3행 가운데 처음 나오는 `# 미션`·`# 과제` 행입니다 — 맨 앞에 빈 줄이나
    import 한 줄이 끼어도 채점 표가 찍히게 합니다. 「준비 셀」(`# 미션 N · 준비`)은 대상이 아닙니다.
    """
    lines = [l for l in (src or "").split("\n") if l.strip()]
    if lines and lines[0].lstrip().startswith("#@title"):
        return None
    for head_line in lines[:3]:
        m = _re.match(r"\s*#\s*(미션\s*(\d+)|과제)\b", head_line)
        if not m:
            continue
        if "프롬프트" not in head_line:
            return None
        return "m" + m.group(2) if m.group(2) else "과제"
    return None


def _lost_tag(src):
    """머리줄이 사라진 「PROMPT 셀」 — 미션 이름이 적힌 코드면 그 미션, PROMPT·표시선만 있으면 "?", 아니면 None.

    AI 답 전체를 셀에 덮어쓰면 첫 행의 머리줄이 사라져 채점이 아무것도 찍지 않습니다. 조용히 지나가지 않고
    「머리줄이 없어 채점하지 못했다」 고 채점 표로 알립니다(`_lost_notice`).
    """
    first = next((l for l in (src or "").split("\n") if l.strip()), "")
    if first.lstrip().startswith("#@title") or _re.match(r"\s*#\s*(미션\s*\d+|과제)\s*·\s*준비", first):
        return None                                       # 채점 셀 · 준비 셀
    flat = _flat(src)
    hits = {k: _bi.sum(_flat(l) in flat for l in (_device_spec(s) or {}).get("values", {})) for k, s in _ANS.items()}
    best = _bi.max(hits.values()) if hits else 0
    keys = [k for k, n in hits.items() if n == best and n > 0]
    if len(keys) == 1:
        return keys[0]
    if "여기부터 AI 코드" in (src or "") or _re.search(r"^[ \t]*PROMPT[ \t]*=", src or "", _re.M):
        return "?"
    return None


def _lost_notice(key):
    """머리줄이 없어 채점하지 못한 셀 — 채점 표 꼴로 알리고 닫습니다"""
    if key == "?":
        head_label, header = "채점하지 못한 셀", "「# 미션 N — 프롬프트」"
    else:
        head_label = ("미션 " + key[1:]) if key[:1] == "m" and key[1:].isdigit() else key
        trap = _is_trap(key)                               # 빈 PROMPT 에도 잠그지 않는 미션이 함정 미션입니다
        header = f"「# {head_label} — 프롬프트{' (함정)' if trap else ''}」"
    print("\n" + "─" * 54)
    print(f"◼ {head_label} 채점 결과" if key != "?" else f"◼ {head_label}")
    print(f"❌ 머리줄 — 셀 첫 행에 {header} 행이 없어 이 셀을 채점하지 못했습니다")
    print("   힌트 : AI 답을 붙여 넣으며 셀 전체를 덮어쓰면 첫 행의 머리줄, PROMPT 입력란, 「# --- 여기부터 AI 코드 ---」"
          " 표시선이 함께 사라집니다. 셀을 되돌리거나(실행 취소) 그 세 가지를 셀 맨 위에 다시 적고,"
          " AI 코드는 표시선 아래에만 붙여 넣은 뒤 다시 실행하세요")
    print("─" * 54)


def _shown_text(data):
    """display 출력 묶음에서 화면에 보이는 글자 — text/plain 이 자리표시면 markdown · html 의 글자.

    자리표시는 `<IPython.core.display.Markdown object>` · `<pandas.io.formats.style.Styler at 0x…>` 같은 객체 표기입니다
    (`display(표.style…)` 의 text/plain 에는 「object」 단어가 없습니다). HTML 표는 칸마다 줄이 바뀌어 적혀 있으므로
    한 행을 한 줄로 모으고, 화면에 보이지 않는 `<style>`·`<script>` 는 뺍니다.
    """
    import html as _html
    plain = str(data.get("text/plain", ""))
    if plain and not _re.fullmatch(r"\s*<[\w.]+(?: object)?(?: at 0x[0-9a-fA-F]+)?>\s*", plain):
        return plain
    if "text/markdown" in data:                  # **굵게** · `코드` 표시 기호는 화면에 안 보입니다
        return _re.sub(r"\*\*|__|`", "", str(data["text/markdown"]))
    if "text/html" in data:
        page = _re.sub(r"(?is)<(style|script)\b.*?</\1\s*>", "", str(data["text/html"]))
        page = _re.sub(r"(?is)<table\b.*?</table\s*>", lambda m: _re.sub(r"\s+", " ", m.group(0)), page)
        page = _re.sub(r"(?i)</t[dh]\s*>", "\t", page)
        page = _re.sub(r"(?i)<br\s*/?>|</(?:p|div|tr|li|h\d)\s*>", "\n", page)
        return _html.unescape(_re.sub(r"<[^>]+>", "", page))
    return plain


def _pub_hook():
    """display() 로 띄운 표·값도 채점이 읽게 — Colab·LMS(ipykernel)는 display 출력을 stdout 이 아닌 채널로 보냅니다.

    「PROMPT 셀」이 도는 동안만 display_pub.publish 를 감싸, 화면에 보이는 글자(text/plain · Markdown · HTML)를
    채점 버퍼에 적고 화면에 뜬 그림(image/*) 수를 셉니다. 터미널처럼 표시가 stdout 에도 쓰이는 자리에서 두 번
    담기지 않게, 원래 표시를 부르는 동안은 화면으로만 내보냅니다. `_post` 가 원래 publish 로 되돌립니다.
    """
    _pub_unhook()
    try:
        dp = _ip.display_pub
    except Exception:
        return
    had = "publish" in vars(dp)
    orig = dp.publish
    state, sysm = _STATE, _sys                 # 학생 코드가 도는 동안 불립니다 — 이름을 여기서 묶어 둡니다

    def note(d):
        if _bi.any(str(m).startswith("image/") for m in d):
            state["imgs"] = state.get("imgs", 0) + 1
        elif d and state.get("buf") is not None:
            state["buf"].write("\n" + _shown_text(d) + "\n")

    own = _get("_OWN")
    if isinstance(own, dict):                  # 학생 셀이 채점 코드와 같은 이름을 만든 뒤 display 를 불러도 읽게
        note = _guard(note, own, globals())

    def publish(data=None, *a, **k):
        try:
            note(data if isinstance(data, dict) else {})
        except Exception:
            pass
        cur = sysm.stdout
        sysm.stdout = state.get("out") or cur
        try:
            return orig(data, *a, **k)
        finally:
            sysm.stdout = cur

    dp.publish = publish
    _STATE["pub"] = (dp, had, orig, publish)


def _pub_unhook():
    """`_pub_hook` 이 감싼 publish 를 되돌립니다"""
    saved, _STATE["pub"] = _STATE.get("pub"), None
    if not saved:
        return
    dp, had, orig, mine = saved
    try:
        if vars(dp).get("publish") is mine:
            if had:
                dp.publish = orig
            else:
                del dp.publish
    except Exception:
        pass


def _pre(info=None):
    """「PROMPT 셀」이면 출력 버퍼를 걸고 그림 목록을 비웁니다"""
    try:
        src = str(getattr(info, "raw_cell", "") or "")
        _STATE.update(tag=None, lost=None, src=src, prompt=None, imgs=0)
        tag = _tag(src)
        hook = _BOOK.get("pre")
        if hook is not None:
            try:
                hook(tag if tag in _ANS else None)     # 권 설정 — 셀 실행 전 상태를 적어 둡니다
            except Exception:
                pass
        if tag is None:
            _STATE["lost"] = _lost_tag(src)
            return
        if tag not in _ANS:
            return
        _STATE["tag"] = tag
        _FIGS.clear()
        _STATE["ids0"] = {id(f) for f in _open_figs()}
        _STATE["prompt"] = _prompt_in_cell(src)    # 앞 미션의 PROMPT 가 남아 있어도 이 셀 것만
        _STATE["buf"], _STATE["out"] = _io.StringIO(), _sys.stdout
        _sys.stdout = _Tee(_sys.stdout, _STATE["buf"])
        _pub_hook()
    except Exception:
        pass


def _post(result=None):
    """「PROMPT 셀」이 끝나면 출력을 되돌리고 그 셀의 출력·그림으로 채점합니다 — 무슨 일이 있어도 채점 표를 닫습니다"""
    tag, buf, out, prompt = _STATE.get("tag"), _STATE.get("buf"), _STATE.get("out"), _STATE.get("prompt")
    _STATE.update(tag=None, buf=None, out=None, prompt=None)
    _pub_unhook()
    if out is not None:
        _sys.stdout = out
    lost, _STATE["lost"] = _STATE.get("lost"), None
    if lost:
        try:
            _lost_notice(lost)
        except Exception:
            print("─" * 54)
        return
    if tag is None or tag not in _ANS:
        return
    try:
        seen = {id(f) for f in _FIGS}
        figs_all = list(_FIGS) + [f for f in _open_figs()
                                  if id(f) not in seen and id(f) not in _STATE.get("ids0", set())]
        txt = buf.getvalue() if buf is not None else ""
        last_val = getattr(result, "result", None)       # 셀 마지막 표현식의 값 — print 없이 표로 띄운 값도 읽습니다
        if last_val is not None:
            try:
                txt = txt + "\n" + str(last_val)
            except Exception:
                pass
        # 문법 오류(SyntaxError·IndentationError)는 실행 전에 나서 error_before_exec 에만 남습니다 — 이것도 실행 오류입니다
        before = getattr(result, "error_before_exec", None)
        error = getattr(result, "error_in_exec", None) or before
        if before is None and prompt:
            prompt = None                                # 셀이 돌았으면 PROMPT 는 방금 실행된 값을 읽습니다
        elif before is not None and prompt is None:
            prompt = ""                                  # 돌지 않은 셀의 PROMPT 를 따옴표 꼴로 못 읽었습니다
        try:
            import matplotlib as _mpl
            shown = (_STATE.get("imgs", 0) > 0) if "inline" in _mpl.get_backend().lower() else None
        except Exception:
            shown = None
        judge(tag, txt, figs_all, error=error, prompt=prompt, shown=shown)
    except Exception as e:
        print(f"(채점을 마치지 못했습니다 — {type(e).__name__}: {_one_line(e)})")
        print("   맨 위 「[채점] 이 셀을 먼저 실행하세요」 셀을 다시 실행한 뒤 이 셀을 다시 실행하세요")
        print("─" * 54)


def _own_names(before):
    """이 셀이 만든 이름 → 값 — 밑줄로 시작하는 이름과 judge. IPython 이 쓰는 이름(_ · _i · _oh · _3 …)은 뺍니다.

    셀을 실행하기 전과 값이 달라졌거나 새로 생긴 이름입니다. 이 셀을 다시 실행하면 모듈 별칭처럼 값이 그대로인 이름은
    달라진 것으로 보이지 않으므로, 앞서 실행 때 적어 둔 이름도 함께 넣습니다.
    """
    prev, g = _get("_OWN"), globals()
    prev = prev if isinstance(prev, dict) else {}
    return {n: v for n, v in list(g.items())
            if (n[:1] == "_" or n == "judge") and n[:2] != "__" and n != "_G0"
            and not _re.fullmatch(r"_(?:i{0,3}|ih|oh|dh|i?\d+|exit_code)", n)
            and (n not in before or before[n] is not v or n in prev)}


def _guard(fn, own, g):
    """훅(`_pre`·`_post`)이 도는 동안만 채점 코드의 이름을 제자리에 둡니다.

    채점 코드의 도우미(`_axes`·`_flat`·`_nums`·`_get` …)는 학생 셀과 같은 이름 공간에 있습니다. 학생 코드가
    `_fig, _axes = plt.subplots(1, 2)` 처럼 같은 이름을 만들면 채점이 그 값을 함수로 부르다 멎고, 그 뒤 모든 미션이
    「채점을 마치지 못했습니다」 로 끝납니다. 그래서 훅에 들어갈 때 채점 코드의 이름을 되돌려 놓고, 나올 때 학생이 만든
    값을 돌려놓습니다 — 채점도, 그 이름을 쓰는 학생의 다음 셀도 그대로 돕니다.
    """
    def run(*a, **k):
        mine = {n: g[n] for n in own if n in g and g[n] is not own[n]}     # 학생 셀이 같은 이름에 넣어 둔 값
        g.update(own)
        try:
            return fn(*a, **k)
        finally:
            own.update({n: g[n] for n in own if n in g and g[n] is not own[n]})   # 채점 코드가 바꿔 단 값은 그대로
            g.update(mine)

    run.__name__ = fn.__name__
    return run


try:
    _OWN = _own_names(_G0)
    del _G0
    _ip = get_ipython()
    _OWN.update(_ip=_ip, _OWN=_OWN)
    for _ev, _fn in (("pre_run_cell", _pre), ("post_run_cell", _post)):
        for _old in list(_ip.events.callbacks[_ev]):
            if getattr(_old, "__name__", "") == _fn.__name__:
                _ip.events.unregister(_ev, _old)     # 이 셀을 다시 실행해도 채점은 한 번만 찍힙니다
        _ip.events.register(_ev, _guard(_fn, _OWN, globals()))
    print("채점 준비 완료 — 미션의 「PROMPT 셀」을 실행하면 실행 결과(출력)에 채점 결과가 찍힙니다")
except Exception:
    print("채점 준비 완료 (IPython 밖이라 훅 없이 judge() 만 올렸습니다)")
````

## 항공 민원과 표준답변 준비

자료는 이론과 같은 세 가지입니다.

| 자료 | 실습에서 쓰는 범위 |
|---|---|
| 공정거래위원회 소비자 민원 2020(data.go.kr 15098314) | 품목명이 항공여객운송서비스인 항공 민원 12,232건 |
| 한국소비자원 소비자상담 표준답변(data.go.kr 15144809) | 1,332개의 「질문」 열 |
| 네이버 쇼핑 리뷰(bab2min/corpus) | 미션 2 에서 표본 2만 건의 앞 1,000건(CPU) · 3,000건(GPU) |

문장 인코더와 건수는 런타임에 따라 정해집니다. 출력의 코사인이 이론과 달라도 모델이 다르면 정상입니다.

| 런타임 | 문장 인코더 | 항공 민원 |
|---|---|---|
| GPU | KURE-v1 | 12,232건 전체 |
| CPU | multilingual-e5-small(접두어 「query: 」) | 4,000건(시드 0) |

「오늘의 준비」 셀은 민원 제목과 표준답변 「질문」을 같은 문장 인코더로 바꿉니다. 그리고 민원마다 가장 가까운 답의 코사인을 구합니다. **대응 없는 민원**은 이 코사인이 임계값보다 작은 민원입니다.

- 표 1 · 항공 민원마다 가장 가까운 표준답변의 코사인 (백분위수)
- 표 2 · 임계값을 중앙값에서 0.02 씩 옮길 때 대응 없는 항공 민원 (%)
- 표 3 · 후보 쌍 40개에서 가장 가까운 답으로 많이 나온 표준답변 (쌍)
- 그림 1 · 항공 민원마다 가장 가까운 표준답변의 코사인

후보 쌍 40개는 과정이 KURE-v1 코사인의 구간 5개에서 8쌍씩 뽑은 민원과 답의 짝입니다.

````python
# 오늘의 준비 — 읽고 실행만 합니다
# [heavy] 항공 민원 제목과 표준답변 1,332개를 문장 인코더로 바꿔 표 1～3 과 그림 1 을 만듭니다. CPU 30초 · GPU 1분 안팎이고, 처음 한 번은 원자료와 가중치를 받느라 더 걸립니다
import os, re, time, hashlib, urllib.request
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt
from sentence_transformers import SentenceTransformer

# 실행 환경과 문장 인코더 — GPU 면 KURE-v1 · 항공 민원 전체, CPU 면 e5-small · 항공 민원 4,000건(시드 0)
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
if DEVICE == "cuda":
    MODEL_NAME, MODEL_REV, PREFIX, N_AIR = "nlpai-lab/KURE-v1", "8b418a58414668e75532ed045c22d9ca018ae2b2", "", None
else:
    MODEL_NAME, MODEL_REV, PREFIX, N_AIR = ("intfloat/multilingual-e5-small", "614241f622f53c4eeff9890bdc4f31cfecc418b3",
                                            "query: ", 4000)

# 원자료 — 원 출처 URL · 파일 이름 · SHA-256(2026-09-24 확인) · 데이터 페이지
RAW_FILES = {
    "complaints": ("https://www.data.go.kr/cmm/cmm/fileDownload.do?"
                   "atchFileId=FILE_000000002610054&fileDetailSn=1&insertDataPrcus=N",
                   "kftc_complaints_1.csv",
                   "50c098fea2ec1d05191d07c45cd0f75b4d0ac3b5e5ac36910fe40f2aab699ebe",
                   "https://www.data.go.kr/data/15098314/fileData.do"),
    "answers": ("https://www.data.go.kr/cmm/cmm/fileDownload.do?"
                "atchFileId=FILE_000000003197092&fileDetailSn=1&insertDataPrcus=N",
                "kca_std_answers.csv",
                "be5ad6ab3e2d6becbf6a32fe8f432c7bb2f575af10c017a42ce5e89bbffdf9d8",
                "https://www.data.go.kr/data/15144809/fileData.do"),
}
MIRROR_URL = "https://github.com/studio-js/korean-cx-text/raw/main/raw"   # 과정 복제본 — 원 출처와 같은 바이트, 먼저 받는 곳


def data_dir():
    # 자료 폴더: 환경변수 NLP_DATA_DIR 이 있으면 그 폴더, 없으면 ./nlp_data
    return os.environ.get("NLP_DATA_DIR", "nlp_data")


def find_local(name):
    # 자료 폴더 → cache/ → raw/ 순서로 찾아 처음 나온 경로, 없으면 None
    for sub in ("", "cache", "raw"):
        path = os.path.join(data_dir(), sub, name)
        if os.path.isfile(path):
            return path
    return None


def sha256_of(path, chunk=1 << 20):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for block in iter(lambda: f.read(chunk), b""):
            h.update(block)
    return h.hexdigest()


def fetch_one(url, name):
    # url 에서 받아 자료 폴더/cache/ 에 두고 경로를 반환함 — 실패하면 세 번까지 다시 시도하고, 끝내 못 받으면 None
    path = os.path.join(data_dir(), "cache", name)
    os.makedirs(os.path.dirname(path), exist_ok=True)
    for i in range(3):
        try:
            urllib.request.urlretrieve(url, path + ".part")
            os.replace(path + ".part", path)
            return path
        except Exception as e:
            print(f"받기 실패 {i + 1}/3: {e}")
            time.sleep(2 * (i + 1))
    return None


def fetch_raw(name):
    # 원자료 경로: 로컬을 먼저 보고, 없으면 과정 복제본(MIRROR_URL) → 원 출처 URL 순서로 받음
    url, fname, sha, page = RAW_FILES[name]
    path = find_local(fname)
    if path is None:
        path = fetch_one(f"{MIRROR_URL}/{fname}", fname)
        if path is not None and sha256_of(path) != sha:   # 복제본이 다르면 버리고 원 출처로
            os.remove(path)
            path = None
    if path is None:
        path = fetch_one(url, fname)
    if path is None:
        raise RuntimeError(f"원 출처에서 파일을 받지 못했습니다. 데이터 페이지: {page}")
    if sha256_of(path) != sha:
        print(f"원 출처 파일이 바뀌었습니다. 결과 수가 교안과 다를 수 있습니다(데이터 페이지: {page})")
    return path


TEST_INPUT_RE = re.compile(   # 민원 파일의 시험 입력 행
    r"(?i)^\s*(?:test|테스트)\s*(?:입니다)?\s*\.?\s*$"
    r"|IN-TEST"
    r"|(?:인바운드|아웃바운드|녹취|수화기).*테스트|테스트.*(?:인바운드|아웃바운드|녹취|수화기)"
    r"|^\s*자동차\s*(?:상담\s*)?테스트"
    r"|^\s*\d{8}\s*[-_ ].*테스트"
    r"|테스트\s*데이터|상담\s*(?:전화\s*)?테스트|전화\s*테스트|헬프데스크\s*시험|^\s*일반\s*테스트")
EMPTY_TITLE_RE = re.compile(r"^[^가-힣A-Za-z]*$")   # 한글 음절도 로마자도 없는 제목
PHONE_RE = re.compile(r"(?<!\d)(?:01[016789]|0\d{1,2})[-.)\s]{0,2}\d{3,4}[-.\s]\d{4}(?!\d)")   # 전화번호 형태

# 후보 쌍 40개 — (민원 row_id, 표준답변 번호). 과정이 KURE-v1 코사인의 구간 5개에서 8쌍씩 뽑은 짝입니다
PAIR_IDS = [(367149, 1230), (46374, 1234), (12109, 1230), (183649, 1298), (122676, 1232), (111373, 1224),
            (28723, 719), (239236, 1234), (222824, 1054), (275200, 576), (440108, 1228), (302897, 1230),
            (367779, 1226), (125915, 1229), (213952, 1226), (331999, 1237), (93797, 1234), (240528, 1230),
            (134301, 1229), (20019, 1233), (400710, 887), (54381, 1230), (58079, 1231), (409279, 1234),
            (304266, 1234), (268718, 1153), (120628, 1234), (288451, 1230), (214295, 1230), (162502, 1230),
            (352979, 1231), (187463, 1230), (379974, 1225), (65639, 1224), (322661, 1234), (401674, 1298),
            (304447, 1234), (313528, 1230), (152334, 1234), (230532, 1224)]

# 항공 민원 — 시험 입력 · 빈 제목 행을 빼고 전화번호 형태를 [전화번호] 로 바꾼 뒤 품목명이 항공여객운송서비스인 행
t0 = time.time()
raw = pd.read_csv(fetch_raw("complaints"), encoding="cp949", dtype=str)
raw = raw.rename(columns=lambda c: c.split("(")[0].strip())
raw.insert(0, "row_id", range(len(raw)))
title = raw["사건제목"].fillna("").str.strip()
is_test = title.str.contains(TEST_INPUT_RE)
keep = ~is_test & ~(title.str.match(EMPTY_TITLE_RE) & ~is_test)
raw["사건제목"] = title.map(lambda t: PHONE_RE.sub("[전화번호]", t))
air_all = raw.loc[keep & (raw["품목명"] == "항공여객운송서비스"), ["row_id", "사건제목"]].reset_index(drop=True)
del raw, title
air = (air_all if N_AIR is None else air_all.sample(n=N_AIR, random_state=0)).reset_index(drop=True)

ans = pd.read_csv(fetch_raw("answers"), encoding="cp949", dtype={"번호": int})
for col in ["품목명", "구분", "질문"]:
    ans[col] = ans[col].astype(str).str.strip()

# 민원 제목과 표준답변 「질문」 열을 같은 문장 인코더로 바꿔, 민원마다 가장 가까운 답의 코사인을 구한다
model = SentenceTransformer(MODEL_NAME, device=DEVICE, revision=MODEL_REV)
model.max_seq_length = 256


def encode(texts):
    return model.encode([PREFIX + t for t in texts], batch_size=64, normalize_embeddings=True,
                        convert_to_numpy=True, show_progress_bar=False).astype(np.float32)


best = (encode(air["사건제목"]) @ encode(ans["질문"]).T).max(axis=1)   # 길이 1 벡터끼리의 내적 = 코사인
q25, q50, q75 = (float(v) for v in np.percentile(best, [25, 50, 75]))
thresholds = [q50 - 0.02, q50, q50 + 0.02]
unmatched = [100 * float((best < t).mean()) for t in thresholds]   # 가장 가까운 답의 코사인이 임계값보다 작은 민원의 비율
print(f"문장 인코더 {MODEL_NAME.split('/')[-1]} · 항공 민원 {len(air):,}건 · 표준답변 {len(ans):,}개 · {time.time() - t0:.0f}초")
print()

print("표 1 · 항공 민원마다 가장 가까운 표준답변의 코사인 (백분위수)")
print(f"  25백분위수 {q25:.3f} · 중앙값 {q50:.3f} · 75백분위수 {q75:.3f}")
print(f"  가장 작은 값 {best.min():.3f} · 가장 큰 값 {best.max():.3f}")
print()

print("표 2 · 임계값을 중앙값에서 0.02 씩 옮길 때 대응 없는 항공 민원 (%)")
for name, t, p in zip(["중앙값 − 0.02", "중앙값", "중앙값 + 0.02"], thresholds, unmatched):
    print(f"  {name:<9}  임계값 {t:.3f}  →  {p:4.1f} %")
print()

# 후보 쌍 40개에서 가장 가까운 답으로 많이 나온 표준답변 둘 — 쌍 수가 큰 순서, 같으면 번호 순서
pairs = pd.DataFrame(PAIR_IDS, columns=["row_id", "번호"])
pair_no = pairs["번호"].value_counts().rename_axis("번호").reset_index(name="쌍")
pair_no = pair_no.sort_values(["쌍", "번호"], ascending=[False, True]).head(2).merge(ans, on="번호")
by_row = air_all.set_index("row_id")["사건제목"]
print("표 3 · 후보 쌍 40개에서 가장 가까운 답으로 많이 나온 표준답변 (쌍)")
for no, cnt, kind, q in pair_no[["번호", "쌍", "구분", "질문"]].itertuples(index=False):
    eg = by_row[pairs.loc[pairs["번호"] == no, "row_id"].iloc[0]]   # 이 답을 가장 가까운 답으로 받은 첫 민원
    print(f"  {no} · {kind} · {cnt}쌍")
    print(f"     질문 「{q[:40]}…」")
    print(f"     이 답을 받은 민원 「{eg}」")

fig, ax = plt.subplots(figsize=(7, 3.2))
ax.hist(best, bins=60, color="#B3B3B3")
for v in (q25, q75):                       # 가운데 절반(25～75백분위수)의 두 경계
    ax.axvline(v, color="#051C2C", linestyle="--")
ax.set_xlabel("가장 가까운 표준답변의 코사인")
ax.set_ylabel("항공 민원 수")
ax.set_title("그림 1 · 항공 민원마다 가장 가까운 표준답변의 코사인")
plt.show()
````

## 코사인과 최근접 이웃 확인

> **[문제] 가까운 글이 없을 때 상위 5 와 임계값**
>
> 예를 들어 질의 하나와 리뷰 1,000건의 코사인을 구했더니 가장 큰 값이 0.93 이었다고 합시다. 상위 5 와 임계값 0.95 로 각각 고르면 몇 건일까요?
>
> 1. 둘 다 0건이다
> 2. 상위 5 는 5건, 임계값 0.95 는 0건이다
> 3. 둘 다 5건이다

> **[문제] 행렬곱 한 번이 코사인이 되는 조건**
>
> 민원 제목 4,000건의 문장 임베딩을 한 행씩 모은 행렬이 `AIR_EMB` 이고, 질의 벡터가 `q` 입니다. `AIR_EMB @ q` 는 수 4,000개입니다. 이 수가 질의와 제목마다의 코사인이 되려면 벡터에 무엇이 필요할까요? 이유를 한두 문장으로 적어 보세요.

## 항공 민원과 표준답변으로 판단하기

아래 세 문항은 「오늘의 준비」 셀 출력을 읽고 답합니다. 표 1 · 그림 1 로는 코사인이 모인 구간을 보고, 표 2 로는 임계값에 따른 비율을 봅니다. 표 3 으로는 가장 가까운 답이 맞는 답인지 봅니다.

<img src="https://raw.githubusercontent.com/UKPLab/sentence-transformers/master/docs/img/SemanticSearch.png" alt="파란 점 여러 개와 주황 점 하나가 흩어진 평면 그림 — 주황 점(Query) 바로 옆의 파란 점 하나에 Relevant Document 라는 이름이 있다" width="339" height="345"/>
<p align="center"><em>▲ 파란 점이 미리 벡터로 바꿔 둔 문서, 주황 점이 질의다. 가장 가까운 파란 점이 관련 문서(Relevant Document)다. 여기서는 주황 점이 항공 민원 한 건, 파란 점이 표준답변 질문 1,332개다 (출처: sentence-transformers 문서 Semantic Search, UKPLab, Apache-2.0)</em></p>

<img src="https://upload.wikimedia.org/wikipedia/commons/7/72/Wilson_score_interval_and_logistic_example.png" alt="가로축 t, 세로축 p 인 S 자 실선 양쪽에 바깥쪽 파선 두 개(n = 10)와 안쪽 점선 두 개(n = 100)가 그려진 그림" width="560" height="347"/>
<p align="center"><em>▲ 실선은 참 비율 p, 양쪽 선은 표본 n 개로 낸 비율의 95% 신뢰구간이다. p = 0.5 에서 n = 10 은 약 0.24～0.76, n = 100 은 약 0.40～0.60 이다. 가로축 t 는 읽지 않는다 (출처: Wikimedia Commons, Wilson score interval and logistic example, CC BY 4.0)</em></p>

> **[문제] 가장 가까운 답의 코사인이 모인 범위**
>
> 「오늘의 준비」 셀을 실행하면 나오는 「표 1 · 항공 민원마다 가장 가까운 표준답변의 코사인 (백분위수)」와 「그림 1 · 항공 민원마다 가장 가까운 표준답변의 코사인」을 봅니다. 가운데 절반(25～75백분위수)이 모인 구간의 길이는 얼마쯤일까요?
>
> 1. 0.1 보다 좁다
> 2. 0.1 에서 0.3 사이다
> 3. 0.3 보다 넓다

> **[문제] 중앙값을 임계값으로 쓴 대응 없는 민원**
>
> 「오늘의 준비」 셀을 실행하면 나오는 「표 2 · 임계값을 중앙값에서 0.02 씩 옮길 때 대응 없는 항공 민원 (%)」를 봅니다. 이 표로 「항공 민원의 절반은 대응하는 표준답변이 없다」고 말할 수 있을까요?
>
> 1. 말할 수 있다: 중앙값 임계값에서 비율이 50.0 % 로 나왔다
> 2. 말할 수 없다: 쌍을 읽어 정한 임계값이 아니고, 0.02 에 크게 바뀐다
> 3. 말할 수 있다: 세 임계값 비율의 평균도 50 % 안팎이다

> **[문제] 후보 쌍 40개에 가장 많이 나온 답**
>
> 「오늘의 준비」 셀을 실행하면 나오는 「표 3 · 후보 쌍 40개에서 가장 가까운 답으로 많이 나온 표준답변 (쌍)」을 봅니다. 번호 1230 을 가장 가까운 답으로 받은 민원을 「대응하는 답이 있음」으로 집계해도 될까요?
>
> 1. 집계해도 된다: 민원과 같은 항공 품목의 표준답변이다
> 2. 집계해도 된다: 40쌍에서 가장 많이 나온 표준답변이다
> 3. 집계하지 않는다: 질문이 미성년자 결제라 민원과 다른 이야기다

## 미션 1 — 키워드 「환불」로 찾지 못한 같은 이야기

**이 미션의 질문**: 질의 「코로나로 항공권 환불을 거부당함」과 가장 가까운 항공 민원 20건 가운데 「환불」이 없는 민원은 몇 건일까요?

<img src="https://upload.wikimedia.org/wikipedia/commons/2/23/CosineSimilarity.png" alt="원점에서 뻗은 노란 화살표 두 개(car 쪽과 truck 쪽) 사이에 80° 라고 적힌 파란 호가 있고, 그 밖에 convertible · Chevrolet · Ford · towing 같은 단어 점이 흩어진 그림" width="560" height="283"/>
<p align="center"><em>▲ car 와 truck 쪽으로 뻗은 두 화살표 사이의 각이 80° 이고, 코사인은 cos 80° ≈ 0.17 이다. 미션 1 은 질의와 민원 제목마다 이 값을 구한다 (출처: Wikimedia Commons, StandingFuture, CC BY-SA 4.0)</em></p>

#### 만들 것

- **`이웃 20건 중 「환불」 없는 민원`** (건 · 정수): 질의와 코사인이 큰 20건 가운데 `명사` 목록에 `KEYWORD` 가 없는 민원 수
- **`문장 임베딩 길이`** (소수점 아래 넷째까지): 민원 제목 하나의 문장 임베딩을 `np.linalg.norm` 으로 구한 길이
- **그래프 한 장**: 가로축 이름 「코사인」, 세로축 이름 「이웃 순위」. 권장은 이웃 20건의 가로 막대 20개

**통과 기준**: 두 값이 「이름 = 값 단위」 한 행씩 기준 범위 안에 있고, 그래프에 가로축 · 세로축 이름이 있으면 통과입니다. 이웃 20건은 내 코드의 `AIR_EMB @ q` 로 다시 골라 건수를 대조합니다.

#### 프롬프트에 넣을 것

- 「미션 1 준비 셀」의 `AIR` · `QUERY` · `KEYWORD` · `TOP_K` · `DEVICE` · `MODEL_NAME` · `MODEL_REV` · `PREFIX` 는 다시 만들지 말 것
- 제목과 질의 모두 앞에 `PREFIX` 를 더해 `normalize_embeddings=True` 로 인코딩하고, 코사인은 `AIR_EMB @ q`
- 「환불」이 있는지는 제목 문자열이 아니라 `명사` 열에서 찾을 것

<details class="lms-extra"><summary>자세히 — 예시 프롬프트</summary>

AIR, QUERY, KEYWORD, TOP_K, DEVICE, MODEL_NAME, MODEL_REV, PREFIX 는 다시 만들지 마. 문장 인코더는 MODEL_NAME 을 revision=MODEL_REV, device=DEVICE 로 불러 줘. 사건제목과 QUERY 앞에 PREFIX 를 더해 normalize_embeddings=True 로 인코딩해 줘. 결과는 AIR_EMB 와 q 로 둬.

"문장 임베딩 길이 = 소수 넷째까지" 를 한 행 출력해 줘.

AIR_EMB @ q 가 큰 TOP_K 건에서 명사 열에 KEYWORD 가 없는 민원 수를 구해 줘. "이웃 20건 중 「환불」 없는 민원 = 정수 건" 으로 출력하고, 20건의 순위 · 코사인 · 제목도 출력해 줘. 가로 막대 20개에 가로축 이름 "코사인", 세로축 이름 "이웃 순위" 를 달아 줘.

</details>

````python
# 미션 1 · 준비 — 읽고 실행만 합니다. 이 셀이 미션 1 에 쓸 이름을 직접 만듭니다
#  민원 원자료를 과정 복제본에서 한 번 받아 두고(67.9MB), 항공 민원 제목을 형태소 명사로 나눕니다. 10초 안팎 걸립니다
#  문장 인코더는 불러 두지 않습니다 — 인코딩은 PROMPT 셀에서 DEVICE · MODEL_NAME · MODEL_REV · PREFIX 로 합니다
import os, re, time, math, hashlib, urllib.request
from decimal import Decimal, ROUND_HALF_UP
import numpy as np
import pandas as pd
import torch
from kiwipiepy import Kiwi
import matplotlib.pyplot as plt

COL = {"남색": "#051C2C", "파랑": "#2251FF", "하늘색": "#00A9F4",   # 맨 위 셀과 같은 색 · 캔버스 폭 —
       "회색": "#757575", "연회색": "#B3B3B3"}                        #  이 셀만 실행해도 정해집니다
FIG_W = 10.0

# 실행 환경과 문장 인코더 설정 — 맨 위 환경 확인 셀과 같은 방법(이 셀만 실행해도 정해집니다)
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
if DEVICE == "cuda":
    MODEL_NAME, MODEL_REV, PREFIX = "nlpai-lab/KURE-v1", "8b418a58414668e75532ed045c22d9ca018ae2b2", ""
    N_AIR = None                     # 항공 민원 전체
else:
    MODEL_NAME, MODEL_REV, PREFIX = "intfloat/multilingual-e5-small", "614241f622f53c4eeff9890bdc4f31cfecc418b3", "query: "
    N_AIR = 4000                     # 항공 민원 가운데 4,000건(시드 0)

# 원자료 — 원 출처 URL · 파일 이름 · SHA-256(2026-09-24 확인) · 데이터 페이지
RAW_FILES = {
    "complaints": ("https://www.data.go.kr/cmm/cmm/fileDownload.do?"
                   "atchFileId=FILE_000000002610054&fileDetailSn=1&insertDataPrcus=N",
                   "kftc_complaints_1.csv",
                   "50c098fea2ec1d05191d07c45cd0f75b4d0ac3b5e5ac36910fe40f2aab699ebe",
                   "https://www.data.go.kr/data/15098314/fileData.do"),
}
MIRROR_URL = "https://github.com/studio-js/korean-cx-text/raw/main/raw"   # 과정 복제본 — 원 출처와 같은 바이트, 먼저 받는 곳


def data_dir():
    # 자료 폴더: 환경변수 NLP_DATA_DIR 이 있으면 그 폴더, 없으면 ./nlp_data
    return os.environ.get("NLP_DATA_DIR", "nlp_data")


def find_local(name):
    # 자료 폴더 → cache/ → raw/ 순서로 찾아 처음 나온 경로, 없으면 None
    for sub in ("", "cache", "raw"):
        path = os.path.join(data_dir(), sub, name)
        if os.path.isfile(path):
            return path
    return None


def sha256_of(path, chunk=1 << 20):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for block in iter(lambda: f.read(chunk), b""):
            h.update(block)
    return h.hexdigest()


def _download(url, path, retries=3):
    # url 을 path.part 로 받아 두고 성공하면 True — 실패하면 세 번까지 다시 시도
    os.makedirs(os.path.dirname(path) or ".", exist_ok=True)
    for i in range(retries):
        try:
            urllib.request.urlretrieve(url, path + ".part")
            return True
        except Exception as e:
            print(f"받기 실패 {i + 1}/{retries}: {e}")
            time.sleep(2 * (i + 1))
    return False


def _fetch(url, name, sha=None, strict=False):
    # 로컬에 있으면 그 경로, 없으면 url 에서 받아 SHA-256 확인 뒤 자료 폴더/cache/ 에 두고 경로를 반환함(받지 못하면 None)
    path = find_local(name)
    if path is None:
        path = os.path.join(data_dir(), "cache", name)
        if not _download(url, path):
            return None
        if strict and sha and sha256_of(path + ".part") != sha:
            os.remove(path + ".part")
            raise RuntimeError(f"받은 {name} 의 SHA-256 이 맞지 않습니다. 다시 실행해 보세요.")
        os.replace(path + ".part", path)
    return path


def fetch_raw(name):
    # 원자료 경로: 로컬을 먼저 보고, 없으면 과정 복제본(MIRROR_URL) → 원 출처 URL 순서로 받음
    url, fname, sha, page = RAW_FILES[name]
    path = find_local(fname)
    if path is None:
        path = _fetch(f"{MIRROR_URL}/{fname}", fname)
        if path is not None and sha256_of(path) != sha:  # 복제본이 다르면 버리고 원 출처로
            os.remove(path)
            path = None
    if path is None:
        path = _fetch(url, fname)
    if path is None:
        raise RuntimeError(f"원 출처에서 파일을 받지 못했습니다. 데이터 페이지: {page}")
    if sha256_of(path) != sha:
        print(f"원 출처 파일이 바뀌었습니다. 결과 수가 교안과 다를 수 있습니다(데이터 페이지: {page})")
    return path


def rnd(x, digits):
    # 십진 값 기준 반올림(끝자리 5 는 올림) — 교안의 수와 같은 방식
    return float(Decimal(f"{float(x):.12g}").quantize(Decimal(1).scaleb(-digits), rounding=ROUND_HALF_UP))


# 민원 파일 읽기 — 이 과목에서 해 온 전처리 그대로: 시험 입력·빈 제목 행을 빼고, 전화번호 형태를 [전화번호] 로 바꾼다
TEST_INPUT_RE = re.compile(
    r"(?i)^\s*(?:test|테스트)\s*(?:입니다)?\s*\.?\s*$"
    r"|IN-TEST"
    r"|(?:인바운드|아웃바운드|녹취|수화기).*테스트|테스트.*(?:인바운드|아웃바운드|녹취|수화기)"
    r"|^\s*자동차\s*(?:상담\s*)?테스트"
    r"|^\s*\d{8}\s*[-_ ].*테스트"
    r"|테스트\s*데이터|상담\s*(?:전화\s*)?테스트|전화\s*테스트|헬프데스크\s*시험|^\s*일반\s*테스트")
EMPTY_TITLE_RE = re.compile(r"^[^가-힣A-Za-z]*$")   # 한글 음절도 로마자도 없는 제목(「-」「.」「0」「ㅇ」)
PHONE_RE = re.compile(r"(?<!\d)(?:01[016789]|0\d{1,2})[-.)\s]{0,2}\d{3,4}[-.\s]\d{4}(?!\d)")


def read_complaints(path, verbose=True):
    # 민원 파일(cp949) 읽기 → 열 이름에서 괄호 안 영문을 떼고, 시험 입력·빈 제목 행을 빼고, 전화번호 형태를 [전화번호] 로 바꾼다
    raw = pd.read_csv(path, encoding="cp949", dtype=str)
    df = raw.rename(columns=lambda c: c.split("(")[0].strip())
    df.insert(0, "row_id", range(len(df)))          # 원자료의 행 번호(0부터)
    title = df["사건제목"].fillna("").str.strip()
    is_test = title.str.contains(TEST_INPUT_RE)
    is_empty = title.str.match(EMPTY_TITLE_RE) & ~is_test
    df = df[~is_test & ~is_empty].copy()
    title = title[df.index]
    df["사건제목"] = title.map(lambda t: PHONE_RE.sub("[전화번호]", t))
    n_masked = int(df["사건제목"].ne(title).sum())
    df["접수일자"] = pd.to_datetime(df["접수일자"], format="%Y-%m-%d")
    df = df.reset_index(drop=True)
    if verbose:
        print(f"민원 원본 {len(raw):,}행 → 시험 입력 {int(is_test.sum())}행·빈 제목 {int(is_empty.sum())}행 제거"
              f" → {len(df):,}행, 전화번호 형태 가림 {n_masked}행")
    return df


def sample_rows(df, n, seed=0):
    # 고정 표본: 같은 입력과 같은 seed 면 같은 행이 같은 순서로 나온다
    return df.sample(n=n, random_state=seed)


def kiwi_nouns(texts, kiwi):
    # 문서마다 명사(NNG·NNP) 목록 — 순서와 반복을 그대로 둔다
    return [[t.form for t in toks if t.tag in ("NNG", "NNP")] for toks in kiwi.tokenize(list(texts))]


AIRLINE = "항공여객운송서비스"          # 민원 품목명 — 항공여객운송서비스

COMPLAINTS_PATH = fetch_raw("complaints")
_c = read_complaints(COMPLAINTS_PATH)
AIR_ALL = _c.loc[_c["품목명"] == AIRLINE, ["row_id", "접수일자", "사건제목"]].reset_index(drop=True)
del _c                                               # 민원 전체 표는 더 쓰지 않는다
AIR = (AIR_ALL if N_AIR is None else sample_rows(AIR_ALL, N_AIR, seed=0)).reset_index(drop=True)   # 쓸 변수 — 항공 민원
import os
os.environ.setdefault("KIWI_ARCH_TYPE", "balanced")  # CPU 종류(Mac·Colab)와 상관없이 형태소 분석 결과를 같게
KIWI = Kiwi(num_workers=2)
AIR["명사"] = kiwi_nouns(AIR["사건제목"], KIWI)          # 제목마다 형태소 명사 목록

QUERY = "코로나로 항공권 환불을 거부당함"             # 쓸 변수 — 질의 문장
KEYWORD, TOP_K = "환불", 20                          # 쓸 변수 — 키워드(형태소 명사) · 이웃 수

print(f"항공 민원 {len(AIR_ALL):,}건 가운데 {len(AIR):,}건 준비"
      + (" (전체)" if N_AIR is None else " (시드 0)")
      + f" · 명사 「{KEYWORD}」 가 있는 제목 {int(AIR['명사'].map(lambda ns: KEYWORD in ns).sum()):,}건")
print(f"DEVICE : {DEVICE} · 문장 인코더 : {MODEL_NAME} · 접두어 PREFIX = {PREFIX!r}")
print("쓸 수 있는 이름 : AIR · QUERY · KEYWORD · TOP_K · DEVICE · MODEL_NAME · MODEL_REV · PREFIX · rnd · KIWI · COL · FIG_W")
````

## 미션 2 — 모델만 바꾼 중복 리뷰 기준 고치기 (함정 미션)

**이 미션의 질문**: e5-small 의 임계값을 다시 정하면 거의 같은 글이 있는 리뷰는 몇 % 일까요?

미션 1 과 같은 순서로 합니다. 이 미션은 함정입니다. 「PROMPT 셀」의 표시선 아래 코드는 오류 없이 실행되지만, 「리뷰 대부분이 복사한 글」이라는 결론은 믿을 수 없습니다. PROMPT 가 비어 있어도 ▶ 하면 그 코드의 채점이 나옵니다.

**고칠 것**: e5-small 에 그대로 쓴 임계값 0.90. `REV` · `LOW` · `embed` 는 그대로 둡니다.

#### 만들 것

- **`KURE-v1 임계값 0.90 쌍`** · **`e5-small 임계값 0.90 쌍`** (쌍 · 정수): i < j 인 위 삼각 쌍 가운데 코사인 0.90 이상인 쌍 수
- **`별점 갈린 비율 KURE-v1`** · **`별점 갈린 비율 e5-small`** (% · 소수 첫째까지): 그 쌍 가운데 한쪽이 1 · 2점, 다른 쪽이 4 · 5점인 쌍의 비율
- **`e5-small 다시 정한 임계값`** (소수 넷째까지): e5-small 코사인을 큰 순서로 정렬했을 때 KURE-v1 0.90 쌍 수 번째 값
- **`다시 정한 임계값에서 e5-small 쌍`** (쌍 · 정수): 그 값 이상인 e5-small 쌍 수
- **`거의 같은 글이 있는 리뷰`** (건 · %): 다시 정한 임계값 이상인 쌍에 한 번 이상 나온 리뷰 수와 그 수 ÷ 리뷰 수. 「k 건 (r %)」 한 행으로
- **그래프 한 장**: 가로축 이름 「임계값」, 세로축 이름 「임계값 이상인 쌍 수」

**통과 기준**: 값 7개가 기준 범위 안이고, 값끼리의 관계 4가지가 맞고, 그래프에 가로축 · 세로축 이름이 있으면 통과입니다. 관계는 쌍 수 비 1.3배 이상과, e5-small 의 별점 갈린 비율이 더 큰 것입니다. 또 다시 정한 쌍 수가 KURE-v1 쌍 수의 ± 10% 안이고, 비율이 건수 ÷ 리뷰 수 × 100 이어야 합니다. 「판단 :」 행이 남으면 통과하지 못합니다.

#### 프롬프트에 넣을 것

- 0.90 을 e5-small 에 그대로 쓰면 안 되는 이유 한 문장
- e5-small 임계값은 KURE-v1 0.90 의 쌍 수에 맞춰 다시 정한다는 것
- 「미션 2 준비 셀」의 `REV` · `LOW` · `embed` 는 그대로 쓰고 「판단 :」 행은 지울 것

> **[예측] 표시선 아래 코드의 결론을 믿을 수 없는 이유는 무엇일까요?**
>
> 1. KURE-v1 에서 정한 0.90 을 e5-small 에 그대로 썼다. 두 모델은 코사인 분포가 다르다
> 2. e5-small 에 접두어 「query: 」를 더했다. 접두어 때문에 코사인이 커진다
> 3. 리뷰를 N_REV 건만 썼다. 20만 건 전체로 다시 계산해야 한다
>
> 답을 정한 뒤 아래를 읽어 보세요(정답 채점은 없습니다).

````python
# 미션 2 · 준비 — 읽고 실행만 합니다. 이 셀이 미션 2 에 쓸 이름을 직접 만듭니다
#  리뷰 원자료를 과정 복제본에서 한 번 받아 두고(20.6MB), 고정 표본의 앞 N_REV 건을 고릅니다. 몇 초면 끝납니다
#  문장 인코더는 불러 두지 않습니다 — embed 를 처음 호출할 때 그 모델을 DEVICE 로 옮깁니다
import os, re, time, math, hashlib, urllib.request
from decimal import Decimal, ROUND_HALF_UP
import numpy as np
import pandas as pd
import torch
from sentence_transformers import SentenceTransformer
import matplotlib.pyplot as plt

COL = {"남색": "#051C2C", "파랑": "#2251FF", "하늘색": "#00A9F4",   # 맨 위 셀과 같은 색 · 캔버스 폭 —
       "회색": "#757575", "연회색": "#B3B3B3"}                        #  이 셀만 실행해도 정해집니다
FIG_W = 10.0

# 실행 환경 — 맨 위 환경 확인 셀과 같은 방법(이 셀만 실행해도 정해집니다)
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
N_REV = 3000 if DEVICE == "cuda" else 1000          # 리뷰 건수 — GPU 3,000건 · CPU 1,000건

# 문장 인코더 둘 — 허브 이름 · 버전(revision) · 문장 접두어
ENCODERS = {
    "KURE-v1": ("nlpai-lab/KURE-v1", "8b418a58414668e75532ed045c22d9ca018ae2b2", ""),
    "e5-small": ("intfloat/multilingual-e5-small", "614241f622f53c4eeff9890bdc4f31cfecc418b3", "query: "),
}
OLD_THRESHOLD = 0.90          # 전임 담당자가 KURE-v1 에서 코사인이 높은 쌍을 읽어 정한 임계값

# 원자료 — 원 출처 URL · 파일 이름 · SHA-256(2026-09-24 확인) · 데이터 페이지
RAW_FILES = {
    "reviews": ("https://raw.githubusercontent.com/bab2min/corpus/master/sentiment/naver_shopping.txt",
                "naver_shopping.txt",
                "dc4d1aca0a148671cbe80bcb81962eee297370acab42be93c1617ce9336479c0",
                "https://github.com/bab2min/corpus/tree/master/sentiment"),
}
MIRROR_URL = "https://github.com/studio-js/korean-cx-text/raw/main/raw"   # 과정 복제본 — 원 출처와 같은 바이트, 먼저 받는 곳


def data_dir():
    # 자료 폴더: 환경변수 NLP_DATA_DIR 이 있으면 그 폴더, 없으면 ./nlp_data
    return os.environ.get("NLP_DATA_DIR", "nlp_data")


def find_local(name):
    # 자료 폴더 → cache/ → raw/ 순서로 찾아 처음 나온 경로, 없으면 None
    for sub in ("", "cache", "raw"):
        path = os.path.join(data_dir(), sub, name)
        if os.path.isfile(path):
            return path
    return None


def sha256_of(path, chunk=1 << 20):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for block in iter(lambda: f.read(chunk), b""):
            h.update(block)
    return h.hexdigest()


def _download(url, path, retries=3):
    # url 을 path.part 로 받아 두고 성공하면 True — 실패하면 세 번까지 다시 시도
    os.makedirs(os.path.dirname(path) or ".", exist_ok=True)
    for i in range(retries):
        try:
            urllib.request.urlretrieve(url, path + ".part")
            return True
        except Exception as e:
            print(f"받기 실패 {i + 1}/{retries}: {e}")
            time.sleep(2 * (i + 1))
    return False


def _fetch(url, name, sha=None, strict=False):
    # 로컬에 있으면 그 경로, 없으면 url 에서 받아 SHA-256 확인 뒤 자료 폴더/cache/ 에 두고 경로를 반환함(받지 못하면 None)
    path = find_local(name)
    if path is None:
        path = os.path.join(data_dir(), "cache", name)
        if not _download(url, path):
            return None
        if strict and sha and sha256_of(path + ".part") != sha:
            os.remove(path + ".part")
            raise RuntimeError(f"받은 {name} 의 SHA-256 이 맞지 않습니다. 다시 실행해 보세요.")
        os.replace(path + ".part", path)
    return path


def fetch_raw(name):
    # 원자료 경로: 로컬을 먼저 보고, 없으면 과정 복제본(MIRROR_URL) → 원 출처 URL 순서로 받음
    url, fname, sha, page = RAW_FILES[name]
    path = find_local(fname)
    if path is None:
        path = _fetch(f"{MIRROR_URL}/{fname}", fname)
        if path is not None and sha256_of(path) != sha:  # 복제본이 다르면 버리고 원 출처로
            os.remove(path)
            path = None
    if path is None:
        path = _fetch(url, fname)
    if path is None:
        raise RuntimeError(f"원 출처에서 파일을 받지 못했습니다. 데이터 페이지: {page}")
    if sha256_of(path) != sha:
        print(f"원 출처 파일이 바뀌었습니다. 결과 수가 교안과 다를 수 있습니다(데이터 페이지: {page})")
    return path


def rnd(x, digits):
    # 십진 값 기준 반올림(끝자리 5 는 올림) — 교안의 수와 같은 방식
    return float(Decimal(f"{float(x):.12g}").quantize(Decimal(1).scaleb(-digits), rounding=ROUND_HALF_UP))


# 리뷰 파일 읽기 — 이 과목에서 써 온 잡음 정리 그대로: 반복 자모는 두 글자로 줄이고, 자음 넷 이상의 자모 나열은 지운다
JAMO_RUN_RE = re.compile(r"[ㄱ-ㅣㆍ]{4,}")                 # 낱자모가 네 글자 이상 이어진 자리


def _consonants(run):
    # 서로 다른 자음 개수
    return len(set(re.findall(r"[ㄱ-ㅎ]", run)))


def clean_review(text):
    # 반복 자모는 두 글자로 줄이고 자모 나열은 지우며 띄어쓰기는 쓴 대로 둔다
    text = JAMO_RUN_RE.sub(lambda m: "" if _consonants(m.group()) >= 4 else m.group(), text)
    text = re.sub(r"([ㄱ-ㅎㅏ-ㅣ])\1{2,}", r"\1\1", text)
    return re.sub(r"\s{2,}", " ", text).strip()


def read_reviews(path):
    # 리뷰 파일(탭 구분: 별점 · 본문) 읽기 → 파일의 행 번호 row_id 와 본문_정리 열
    rows = [line.split("\t", 1) for line in open(path, encoding="utf-8").read().split("\n") if line]
    df = pd.DataFrame(rows, columns=["별점", "본문"])
    df.insert(0, "row_id", range(len(df)))
    df["별점"] = df["별점"].astype(int)
    df["본문_정리"] = df["본문"].map(clean_review)
    return df


def sample_rows(df, n, seed=0):
    # 고정 표본: 같은 입력과 같은 seed 면 같은 행이 같은 순서로 나온다
    return df.sample(n=n, random_state=seed)


_ENCODER_CACHE = {}                                    # 한 번 부른 문장 인코더를 다시 받지 않게 둔다


def embed(name, texts):
    # ENCODERS[name] 의 문장 인코더로 문장마다 길이 1 벡터 한 행(float32) — DEVICE 에서 인코딩한다
    hub, rev, prefix = ENCODERS[name]
    if name not in _ENCODER_CACHE:
        _ENCODER_CACHE[name] = SentenceTransformer(hub, device=DEVICE, revision=rev)
        _ENCODER_CACHE[name].max_seq_length = 256
    vecs = _ENCODER_CACHE[name].encode([prefix + t for t in texts], batch_size=64, normalize_embeddings=True,
                                       convert_to_numpy=True, show_progress_bar=False)
    return vecs.astype(np.float32)


REVIEWS_PATH = fetch_raw("reviews")
_r = read_reviews(REVIEWS_PATH)
REV = sample_rows(_r, 20000, seed=0).head(N_REV)[["row_id", "별점", "본문_정리"]].reset_index(drop=True)   # 쓸 변수 — 리뷰
del _r                                               # 리뷰 20만 건 전체 표는 더 쓰지 않는다
LOW = (REV["별점"] <= 2).to_numpy()                  # 쓸 변수 — 별점 1 · 2점이면 True

print(f"리뷰 {len(REV):,}건 준비 (표본 2만 건의 앞 {N_REV:,}건, 시드 0) · 별점 1 · 2점 {int(LOW.sum()):,}건"
      f" · 모든 쌍 {len(REV) * (len(REV) - 1) // 2:,}쌍")
print(f"DEVICE : {DEVICE} · 문장 인코더 둘 : " + " · ".join(ENCODERS) + f" · 옛 임계값 OLD_THRESHOLD = {OLD_THRESHOLD}")
print("쓸 수 있는 이름 : REV · LOW · ENCODERS · embed · OLD_THRESHOLD · N_REV · DEVICE · rnd · COL · FIG_W")
````

````python
# 미션 2 — 프롬프트 (함정)   # ai: prompt
# [heavy] 고친 뒤에는 두 모델로 인코딩해 CPU 에서 1분 가까이 걸릴 수 있습니다
# 이 미션은 함정입니다 — 아래는 전임 담당자가 남긴 코드라 오류 없이 실행되지만 결론을 믿을 수 없습니다.
# 고칠 지시는 AI 튜터 대화창이 아니라 아래 PROMPT 따옴표 안에 적습니다. 따옴표 안이 비어 있으면 아래 코드의 진단만 나오고 완성본 풀이는 잠깁니다
# 지시를 적고 AI 튜터에 "위 PROMPT 대로 코드를 써 줘" → 「이 셀에 넣기」 → ▶ 하면 AI 튜터가 AI 코드 표시선 아래를 다시 씁니다
PROMPT = """

"""
# --- 여기부터 AI 코드 ---
# ↓ 전임 담당자의 코드 — 「리뷰 대부분이 복사한 글」이라고 판단한 근거인 중복 집계. 그대로 실행하면 채점이 ❌ 를 냅니다
E = embed("e5-small", REV["본문_정리"])                  # 인코더만 e5-small 로 바꿨다
S = E @ E.T                                             # 길이 1 벡터끼리의 내적 = 모든 쌍의 코사인
iu = np.triu_indices(len(REV), 1)                       # 위 삼각 — 같은 쌍을 두 번 집계하지 않게
cos = S[iu]
m = cos >= OLD_THRESHOLD
split = (LOW[iu[0]] != LOW[iu[1]])[m]                   # 별점 1 · 2점과 4 · 5점이 짝인 쌍
inpair = np.zeros(len(REV), bool)
inpair[iu[0][m]] = True
inpair[iu[1][m]] = True
print(f"모델 = e5-small · 임계값 = {OLD_THRESHOLD:.2f}")
print(f"임계값 이상인 쌍 = {int(m.sum()):,} 쌍 · 별점 갈린 쌍 = {rnd(100 * split.mean(), 1):.1f} %")
print(f"거의 같은 글이 있는 리뷰 = {int(inpair.sum()):,} 건 ({rnd(100 * inpair.mean(), 1):.1f} %)")
print("판단 : 리뷰 대부분이 복사한 글이라 별점 평균을 믿을 수 없다")

grid = np.round(np.arange(0.85, 0.9701, 0.01), 2)       # 임계값 0.85 ～ 0.97
fig, ax = plt.subplots(figsize=(FIG_W * 0.7, 3.6))
ax.plot(grid, [int((cos >= t).sum()) for t in grid], marker="o", color=COL["회색"])
ax.axvline(OLD_THRESHOLD, color=COL["남색"], linestyle="--")
ax.set_xlabel("임계값")
ax.set_ylabel("임계값 이상인 쌍 수")
plt.show()
````

## 오늘의 결론

<div class="lms-summary">

**정리**

- 키워드 「환불」로 집계하면 같은 사정을 「환급」으로 쓴 민원이 빠집니다. 질의의 이웃 20건 가운데 KURE-v1 1건 · e5-small 2건입니다.
- 0.90 을 e5-small 에 그대로 쓰면 GPU 의 리뷰 3,000건 가운데 94.7% 가 거의 같은 글이 됩니다. 쌍 수를 맞춰 다시 정하면 3.1% 입니다(CPU 의 1,000건은 89.2% → 1.4%).
- 가장 가까운 답의 코사인은 가운데 절반이 길이 0.1 도 안 되는 구간에 모입니다. 그래서 임계값을 0.02 만 옮겨도 대응 없는 민원의 비율이 크게 바뀝니다.
- 할 일: 임계값은 쓰는 모델에서 코사인 구간마다 쌍을 무작위로 뽑아 읽고 정밀도로 정합니다. 같은 이야기인지는 원문을 읽어 확인합니다. 별점은 단서입니다. 결과 수 옆에는 `MODEL_NAME` 과 임계값의 근거를 함께 적습니다.

코사인 몇 이상을 같은 이야기로 볼지는 모든 모델에 통하는 수 하나로 정할 수 없습니다.
