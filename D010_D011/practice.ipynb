```python
#패키지 import
import re
import time
from urllib.parse import urljoin, urlparse

import pandas as pd
import requests
from bs4 import BeautifulSoup

url = 'https://news.naver.com/section/103'
print('라이브러리 로드 완료!')
```

```python
# request 조건 정의
response = requests.get(url, timeout=5)

print('상태 코드:', response.status_code)    # 200 = 정상
print('인코딩:   ', response.encoding)
print('HTML 길이:', len(response.text), '자')
print()
print('HTML 앞부분 미리보기:')
print(response.text[:200])    
```

```python
# parser 및 selector 정의
soup = BeautifulSoup(response.text, "html.parser")
# 이건 못쓰는 selector, "bcr7h" 부분이 매번 접속시마다 랜덤생성 selector = '#_SECTION_HEADLINE_LIST_bcr7h > li:nth-child(1) > div > div > div.sa_text > a > strong'
# selector 정의: ul 태그를 사용한 태그들 중 id 속성 안의 값이 _SECTION_HEADLINE_LIST로 시작하는 애들을 가져와라
selector = 'ul[id^="_SECTION_HEADLINE_LIST"]'

headlines = soup.select_one(selector)
for i, li in enumerate(headlines.select("li"), 1):
    title = li.select_one("strong.sa_text_strong").get_text(strip=True)
    press = li.select_one("div.sa_text_press").get_text(strip=True)
    print(f"{i}. {title} - {press}")  
```

```python
for i, li in enumerate(headlines.find_all("li"), 1):
    title = li.find("strong", class_="sa_text_strong").get_text(strip=True)
    print(f"{i}. {title}")
```

```python
for i, li in enumerate(headlines.select("li"), 1):
    title = li.select_one("strong.sa_text_strong").get_text(strip=True)
    
    press_tag = li.select_one("div.sa_text_press")
    press = press_tag.get_text(strip=True) if press_tag else "매체명 없음"
    
    print(f"{i}. [{press}] {title}")
```
