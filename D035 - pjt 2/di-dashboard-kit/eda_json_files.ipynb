## 파일별 상세 분석

### Profile 파일

```python
print(profile.describe())
print()
print(profile['age'].value_counts().head())
print("\nage=118 개수:", (profile['age'] == 118).sum())
print()
# age=118과 income 결측이 실제로 겹치는지 확인
missing_age = profile['age'] == 118
missing_income = profile['income'].isnull()

print("age=118 개수:", missing_age.sum())
print("income 결측 개수:", missing_income.sum())
print("둘 다 겹치는 개수:", (missing_age & missing_income).sum())
print("gender 결측 개수:", profile['gender'].isnull().sum())
```

```python
missing_age = profile['age'] == 118
missing_income = profile['income'].isnull()
missing_gender = profile['gender'].isnull()

print("age=118 개수:", missing_age.sum())
print("income 결측 개수:", missing_income.sum())
print("gender 결측 개수:", missing_gender.sum())

# 세 조건이 정확히 같은 행 집합인지 확인
same_as_age_income = (missing_age == missing_income).all()
same_as_age_gender = (missing_age == missing_gender).all()

print("\nage=118 == income 결측 (완전 일치 여부):", same_as_age_income)
print("age=118 == gender 결측 (완전 일치 여부):", same_as_age_gender)

# 혹시 완전히 일치하지 않는다면, 어긋나는 행이 몇 개인지 확인
mismatch = (missing_age != missing_income) | (missing_age != missing_gender)
print("\n세 조건 중 하나라도 어긋나는 행 개수:", mismatch.sum())

if mismatch.sum() > 0:
    print("\n어긋나는 행 예시:")
    display(profile[mismatch][['age', 'income', 'gender']].head(10))
```

```python
#became_member_on 칼럼 datetime 형식으로 변환
profile['became_member_on'] = pd.to_datetime(profile['became_member_on'].astype(str), format='%Y%m%d')

# age=118, income, gender 인 고객들을 프로필 미입력 고객으로 따로 정의
profile['is_missing_demo'] = profile['age'] == 118
profile_clean = profile[~profile['is_missing_demo']].copy()

print("전체:", profile.shape)
print("인구통계 있는 고객만:", profile_clean.shape)
```

### Transcript 파일

```python
print(transcript.head())
print(transcript.shape)
print(transcript.dtypes)
print("\n=== event 종류별 개수 ===")
print(transcript['event'].value_counts())

print("\n=== 이벤트별 value 예시 ===")
for event_type in transcript['event'].unique():
    example = transcript[transcript['event'] == event_type]['value'].iloc[0]
    print(f"\n[{event_type}]")
    print(example)
```

```python
# 1. offer id / offer_id 키 이름부터 통일
def get_offer_id(v):
    if 'offer id' in v:
        return v['offer id']
    elif 'offer_id' in v:
        return v['offer_id']
    return None

def get_amount(v):
    return v.get('amount', None)

def get_reward(v):
    return v.get('reward', None)

transcript['offer_id'] = transcript['value'].apply(get_offer_id)
transcript['amount'] = transcript['value'].apply(get_amount)
transcript['reward'] = transcript['value'].apply(get_reward)

# 2. 확인
display(transcript.head(10))
print(transcript[['offer_id', 'amount', 'reward']].isnull().sum())

pd.crosstab(transcript['event'], transcript['offer_id'].notnull())
```

```python
#time 범위 확인 (시간 기준, 관측기간 30일로 최대 720시간)
print("time 범위:", transcript['time'].min(), "~", transcript['time'].max())
print("\ntime 히스토그램용 통계:")
print(transcript['time'].describe())
```

```python
# Unique 고객 & 오퍼 숫자
print("고유 고객 수:", transcript['person'].nunique())
print("고유 오퍼 수:", transcript['offer_id'].nunique())
```

```python
#각 오퍼 id에 따른 event 퍼널 
funnel = transcript[transcript['event'] != 'transaction'].groupby(['offer_id', 'event']).size().unstack(fill_value=0)
funnel = funnel[['offer received', 'offer viewed', 'offer completed']]
display(funnel)
```

```python
transaction_df = transcript[transcript['event'] == 'transaction']
print(transaction_df['amount'].describe())
```

```python
print(transcript.groupby('person')['event'].count().describe())
```

```python
#portfolio 파일 기준, 캠페인 퍼널 퍼포먼스 (Key: offer_id)
funnel_with_info = funnel.merge(
    portfolio[['id', 'offer_type', 'difficulty', 'reward', 'duration', 'channels']],
    left_on='offer_id',
    right_on='id'
).drop(columns='id')

# 완료율(전환율)도 같이 계산해두면 비교하기 편해요
funnel_with_info['view_rate'] = funnel_with_info['offer viewed'] / funnel_with_info['offer received']
funnel_with_info['completion_rate'] = funnel_with_info['offer completed'] / funnel_with_info['offer received']

display(funnel_with_info.sort_values('completion_rate', ascending=False))
```
