### 핵심 질문: 헬스장 민원이 늘었는가 [필수]

**예시 창구**는 소비자 민원을 받는 상담 창구 한 곳을 가정한 가상의 예입니다. 헬스장 민원은 이 창구의 민원 가운데 테마 「헬스장·운동 강습 해지 환불」(줄여 헬스장 테마)에 속한 민원입니다.

이 테마는 토픽을 검증하고 코드북을 만들 때 예시로 썼습니다. 테마는 월간 불만 요약에서 민원을 나누어 집계하는 불만 유형입니다. 요약에 「헬스장 민원이 늘었다」고 쓸지는 달마다 집계한 수를 차례로 비교해 판단합니다.

> 예시 창구의 헬스장 민원은 1～4월에 늘었는가?

늘었다는 근거는 없습니다. 2월 · 3월의 높은 비율은 헬스장 B 한 곳의 민원 때문이고 헬스장 B 를 빼면 비율이 1월과 비슷합니다(「3. 원문을 확인할 달」).

| 달 | 그 달 민원 전체 | 헬스장 테마 민원 |
|---|---:|---:|
| 1월 | 400 | 20 |
| 2월 | 400 | 40 |
| 3월 | 500 | 55 |
| 4월 | 113 | 11 |

헬스장 테마 민원 수만 달마다 비교하면 「3월에 가장 많고 4월에 거의 없어졌다」처럼 보입니다.

그런데 민원의 주제 구성이 그대로여도 아래 같은 일로 수가 바뀝니다.

- 그 달 민원 전체가 늘거나 줄어듭니다.
- 자료가 어느 달 중간에서 끊깁니다.
- 한 판매처에 민원이 몰립니다.
- 민원을 고르는 기준을 고칩니다. 찾을 단어 목록(사전)에 단어를 더하면, 같은 민원에서도 수가 늘어납니다.

4월처럼 건수가 크게 줄어든 달도 확인할 4가지(분모 · 달력 · 원문 · 버전)로 봅니다.

### 확인할 4가지 [필수]

**확인할 4가지**는 주제가 바뀌었다고 판단하기 전에 수가 바뀐 달의 분모 · 달력 · 원문 · 버전을 하나씩 보는 일입니다.

| 확인 | 무엇을 보는가 | 확인하지 않으면 |
|---|---|---|
| 분모 | 그 달 민원 전체로 나눈 비율로 비교했는가, 두 달 차이의 95% 구간이 0 을 포함하는가(「1. 건수인가 비율인가」) | 민원 전체가 늘어난 것을 헬스장 민원이 늘어난 것으로 잘못 해석한다 |
| 달력 | 그 달 접수가 말일까지 이어졌는가, 달마다 공휴일 수가 같은가 | 접수가 덜 된 달이나 공휴일이 많은 달을 민원이 줄어든 달로 잘못 해석한다 |
| 원문 | 원문을 확인할 달(기준은 「3. 원문을 확인할 달」)의 민원이 한 판매처에 몰렸는가 | 한 판매처의 일을 테마 전체의 일로 잘못 해석한다 |
| 버전 | 달마다 다른 기준(사전 · 코드북)으로 집계했는가 | 기준을 바꿔 생긴 차이를 민원의 변화로 잘못 해석한다 |

> 4가지를 모두 확인한 뒤에도 남는 변화만 주제의 변화로 봅니다.

민원 50만 건에 같은 확인을 한 결과는 「5. 민원 50만 건 추이와 팀 라운드」에 모았습니다.

## 1. 건수인가 비율인가

헬스장 민원이 3월에 늘었는지 보려면 건수를 볼까요, 비율을 볼까요? 그 달 민원 전체가 늘면 헬스장 테마의 비중이 그대로여도 건수는 늘어나므로 달끼리는 비율로 비교합니다.

| 민원 수 | 2월 | 3월 | 2월 → 3월 |
|---|---:|---:|---:|
| 헬스장 테마 민원 | 40 | 55 | +37.5% |
| 그 달 민원 전체 | 400 | 500 | +25% |

### 비율 추이 [필수]

헬스장 테마 민원 수만 보면 두 해석이 구분되지 않습니다. 「헬스장 민원이 늘었다」일 수도 있고 「민원이 다 같이 늘었다」일 수도 있습니다.

**비율 추이**는 그 달 민원 전체 가운데 한 테마가 차지하는 비율을 달마다 구해 이은 것입니다.

> 달마다 민원 전체가 다르므로, 달끼리는 그 달 민원 전체로 나눈 비율로 비교합니다.

달 $m$ 의 테마 $t$ 비율 $\hat p_{t,m}$ 은 그 테마의 민원 수를 그 달 민원 전체 수로 나눈 값입니다.

$$
\hat p_{t,m} = \frac{d_{t,m}}{N_m}
$$

- $d_{t,m}$: 달 $m$ 에 들어온 민원 가운데 테마 $t$ 에 속한 민원 수
- $N_m$: 달 $m$ 에 들어온 민원 전체 수(분모)

$t$ 가 헬스장 테마일 때 2월과 3월의 수를 대입합니다.

$$
\hat p_{t,2} = \frac{40}{400} = 10\%, \qquad \hat p_{t,3} = \frac{55}{500} = 11\%
$$

> **[예측] 헬스장 테마 민원이 2월 40건에서 3월 55건으로 늘었고, 비율은 10% 에서 11% 가 되었습니다. 아래 설명 가운데 이 비율만으로 틀렸다고 말할 수 있는 것은 무엇일까요?**
>
> 1. 헬스장 서비스가 3월에 나빠졌다
> 2. 3월에 헬스장 해지가 늘어 환불 민원도 늘었다
> 3. 3월 민원 전체가 늘어 헬스장 비율이 커 보인다
>
> 답을 정한 뒤 아래를 읽어 보세요(정답 채점은 없습니다).

답은 셋째 보기 「3월 민원 전체가 늘어 헬스장 비율이 커 보인다」입니다. 비율은 이미 그 달 민원 전체로 나눈 값이라, 전체와 헬스장 테마 민원이 같은 배율로 늘면 비율은 그대로입니다.

전체가 절반으로 줄 때도 같습니다.

| 전체가 절반이 될 때 | 이번 달 | 다음 달 |
|---|---:|---:|
| 그 달 민원 전체 | 400건 | 200건 |
| 어느 테마의 민원 | 20건 | 10건 |
| 그 테마의 비율 | 5% | 5% |

건수만 본 사람은 「그 테마의 민원이 줄었다」고 판단하지만 그 달 민원 가운데 그 테마의 비중은 바뀌지 않았습니다.

- 둘째 보기 「3월에 헬스장 해지가 늘어 환불 민원도 늘었다」는 그럴듯하지만 원문을 읽기 전의 해석입니다. 원문을 읽기 전에는 「…로 보인다」고 씁니다.
- 첫째 보기 「헬스장 서비스가 3월에 나빠졌다」도 비율로는 맞는지 알 수 없습니다. 민원이 조금 늘었다는 사실만으로는 서비스가 나빠졌는지 모르기 때문입니다.

1～4월의 월 건수와 비율을 그려 비교합니다.

위 그림에서 건수로는 3월이 가장 많고 4월이 가장 적지만 비율은 2월부터 10% 안팎입니다.

**비율이 높아지면 헬스장 민원이 늘어난 것인가요?**

꼭 그렇지는 않습니다. 비율은 그 달 민원 전체 가운데 차지하는 비중이라, 헬스장 테마 민원 수가 같아도 다른 민원이 줄면 높아집니다. 비율이 높아지면 헬스장 테마 민원 수와 그 달 민원 전체 수가 각각 어떻게 달라졌는지 함께 봅니다.

2월 → 3월은 두 수가 모두 늘었고 헬스장 테마 민원이 조금 더 늘어 비율이 높아졌습니다. 비율이 1%p 높아진 것이 우연으로도 생길 만한 크기인지는 두 달 차이의 95% 구간으로 판단합니다.

### 두 달 차이의 95% 구간 [필수]

헬스장 테마 비율은 2월 10% 에서 3월 11% 로 1.00%p 높아졌습니다. 이 정도 차이는 우연으로도 생길 만한 크기라 「헬스장 민원이 늘었다」는 근거가 되지 못합니다.

**표본이 아니라 민원 전체로 구한 비율도 달라지나요?**

예시 창구의 비율은 그 달 민원 전체로 구했습니다. 하지만 민원은 다른 사람이 다른 날 내므로 2월과 조건이 같은 달이 다시 와도 비율은 조금 다르게 나옵니다.

한 달의 민원도 조건이 같은 달에서 한 번 나온 표본으로 봅니다. **표본 변동**은 같은 조건에서 나온 표본이라도 값이 표본마다 달라지는 것입니다. 표준오차는 표본 변동의 크기를 나타내는 값입니다.

**두 달 차이의 95% 구간**은 표본 변동을 감안할 때 두 달 비율의 차이가 있을 만한 범위입니다. 식은 통계적 검정과 인과추론에서 A/B 테스트의 두 집단 비율 차이에 쓴 구간과 같고 두 집단 대신 두 달을 비교합니다.

$$
(\hat p_{\text{3월}} - \hat p_{\text{2월}}) \pm 1.96\sqrt{\frac{\hat p_{\text{2월}}(1-\hat p_{\text{2월}})}{N_{\text{2월}}} + \frac{\hat p_{\text{3월}}(1-\hat p_{\text{3월}})}{N_{\text{3월}}}}
$$

- $\hat p_{\text{2월}}, \hat p_{\text{3월}}$: 두 달의 헬스장 테마 비율
- $N_{\text{2월}}, N_{\text{3월}}$: 두 달의 민원 전체 수
- 제곱근 전체: 두 달 차이의 표준오차. 두 달의 민원은 겹치지 않아 표본 변동이 따로 생기므로 두 달의 항을 더합니다
- 1.96: 두 달 비율이 같다면, 차이의 크기가 표준오차의 1.96배를 넘는 일은 약 5% 입니다
- ± 뒤의 값: **오차 한계(margin of error)**. 골드 표본의 일치율에서 본 것처럼 구간 길이의 절반입니다

**2월 → 3월의 차이는 표본 변동으로 설명되나요?**

두 달의 비율 0.10 과 0.11 을 식에 대입합니다.

```text
차이: 0.11 − 0.10 = 0.0100 → +1.00%p
표준오차: √(0.10 × 0.90 ÷ 400 + 0.11 × 0.89 ÷ 500) = 0.0205
오차 한계: 1.96 × 0.0205 = 0.0402 → ±4.02%p
구간: 0.0100 ± 0.0402 → −3.02～5.02%p
```

1월 → 2월도 같은 식으로 구해 비교합니다. 구간이 0 을 포함하는지에 따라 뜻이 달라집니다.

| 두 달 | 차이 | 오차 한계 | 95% 구간 | 뜻 |
|---|---:|---:|---|---|
| 2월 → 3월 | +1.00%p | ±4.02%p | −3.02～5.02%p, 0 을 포함 | 우연한 표본 변동만으로도 생길 수 있는 차이 |
| 1월 → 2월 | +5.00%p | ±3.63%p | 1.37～8.63%p, 0 을 포함하지 않음 | 표본 변동만으로는 생기기 어려운 차이 |

> 두 달 차이의 95% 구간이 0 을 포함하지 않을 때만 「바뀌었다」고 봅니다.

「3월에 헬스장 민원이 늘었다」는 결론은 이 기준에 맞지 않습니다.

달마다 비율과 구간을 함께 그리면, 두 달의 구간이 겹치는지만 보고 비교하기 쉽습니다. 하지만 두 달은 두 구간이 겹치는지가 아니라, 차이의 구간이 0 을 포함하는지로 판단합니다.

한 달 비율의 오차 한계는 구간 식의 제곱근 안에서 그 달 항 하나만 남긴 값입니다.

$$
1.96\sqrt{\frac{\hat p_{t,m}(1-\hat p_{t,m})}{N_m}}
$$

```text
2월 헬스장 테마: 1.96 × √(0.10 × 0.90 ÷ 400) = 0.0294 → ±2.94%p
3월 헬스장 테마: 1.96 × √(0.11 × 0.89 ÷ 500) = 0.0274 → ±2.74%p
두 달 차이: √(2.94² + 2.74²) ≈ 4.02 → ±4.02%p
```

차이의 오차 한계 ±4.02%p 는 두 오차 한계를 제곱해 더한 뒤 제곱근을 구한 값이라, 두 오차 한계의 합 5.68%p 보다 작습니다. 그래서 두 구간이 겹쳐도 차이의 구간은 0 을 포함하지 않을 수 있습니다. 1월 → 2월이 그런 경우입니다.

```text
두 달의 오차 한계(같은 식): 1월 ±2.14%p · 2월 ±2.94%p → 두 오차 한계의 합 5.08%p
1월 → 2월 차이 5.00%p 는 5.08%p 보다 작아 두 구간이 겹침
차이의 구간 1.37～8.63%p 는 0 을 포함하지 않음
```

다만 구간에는 표본 변동만 나타납니다. 자료가 끊겼는지, 한 판매처에 몰렸는지, 사전이 바뀌었는지는 구간으로 알 수 없습니다. 구간이 0 을 포함하지 않는 1월 → 2월도 이 세 가지를 알아본 뒤에 「바뀌었다」고 판단합니다.

**여러 테마를 한꺼번에 비교하면 무엇이 달라지나요?**

두 달 비율이 실제로 같아도 95% 구간은 100번에 5번쯤 0 을 포함하지 않습니다. 비교를 여러 번 하면 이런 우연한 「차이 있음」도 함께 늘어납니다.

예시 창구의 테마는 헬스장 · 여행 · 예식의 3개입니다. 이 테마들을 1～4월의 이웃한 달마다 비교할 때, 우연한 「차이 있음」이 몇 번 나올지 계산합니다.

```text
비교 횟수: 테마 3개 × 이웃한 달 3쌍(1월 → 2월 · 2월 → 3월 · 3월 → 4월) = 9번
차이가 하나도 없어도: 9 × 0.05 = 0.45번쯤 「차이 있음」
헬스장 테마 하나만 보면: 비교 3번, 3 × 0.05 = 0.15번쯤
```

**다중비교 문제**(multiple comparisons)는 비교가 많을수록 차이가 없어도 기준을 넘는 결과가 늘어나는 문제입니다. 집단 특징 단어를 찾을 때 본 문제가 달끼리 비교할 때도 생깁니다. 따라서 볼 테마를 먼저 정하고 그 구간만 해석합니다.

그래도 우연한 「차이 있음」이 생길 수 있어 원문까지 읽어 봅니다. 원문을 확인할 달은 앞 달보다 비율이 1.5배 이상 오르고 두 달 차이의 95% 구간이 0 을 포함하지 않는 달입니다.

집단 특징 단어(distinctive words)를 찾을 때도 같은 일이 있었습니다. 별점 라벨을 무작위로 다시 배정해 차이가 없게 만든 비교에서도 기준을 넘는 단어가 나왔습니다. 어휘 2,748개 가운데 평균 66.6개의 $|z|$ 가 1.96 을 넘었습니다.

비교할 테마나 단어를 미리 정해 두면 모두 비교한 뒤 차이가 커 보이는 것만 골라 차이가 있다고 말하는 일도 막습니다.

### 테마 비율의 분모 [필수]

예시 창구의 민원은 세 테마 가운데 하나에 속하거나, 어느 테마에도 속하지 않아 「기타」가 됩니다. 「기타」를 빼면 그 달 민원이 늘거나 줄어든 것이 분모에 나타나지 않으므로 테마 비율의 분모에는 「기타」도 포함합니다.

**테마 비율**은 그 달 민원 전체 가운데 한 테마에 속한 민원의 비율로, 앞의 식 $\hat p_{t,m}$ 입니다.

**「기타」를 빼고 나누면 어떻게 되나요?**

| 민원 수와 비율 | 2월 | 3월 |
|---|---:|---:|
| 헬스장 · 여행 · 예식 테마 | 40 · 60 · 20 | 55 · 45 · 20 |
| 기타 | 280 | 380 |
| 헬스장 비율: 테마 3개의 합으로 나눔 | 40 ÷ 120 = 33.3% | 55 ÷ 120 = 45.8% |
| 헬스장 비율: 「기타」 포함 전체로 나눔 | 40 ÷ 400 = 10% | 55 ÷ 500 = 11% |

테마 3개의 합은 두 달 모두 120건이라, 「기타」가 280건에서 380건으로 늘어난 것이 분모에서 빠집니다. 「기타」를 빼면 헬스장 비율이 12.5%p 높아진 것처럼 보입니다.

「기타」를 뺀 비율은 「테마가 정해진 민원 가운데 몇 % 인가」를 나타냅니다. 월간 불만 요약에는 「그 달 민원 가운데 몇 % 인가」가 필요해 분모가 다릅니다.

> 테마 비율의 분모는 「기타」까지 포함한 그 달 민원 전체입니다.

한 달 오차 한계 식에서 $N_m$ 은 제곱근 안의 분모라 한 달 민원이 적을수록 오차 한계가 커집니다. 예시 창구처럼 한 달 민원이 수백 건이면 테마 비율은 ±3%p 안팎의 오차 한계 안에서만 믿을 수 있습니다.

<div class="lms-summary">

**정리**

달끼리는 건수가 아니라 「기타」까지 포함한 그 달 민원 전체로 나눈 비율로 비교합니다. 두 달 차이의 95% 구간이 0 을 포함하면 「늘었다」고 보지 않습니다. 헬스장 테마의 2월 → 3월이 그런 차이입니다.

**할 일**

- 볼 테마를 먼저 정하고 그 구간만 해석합니다
- 앞 달보다 비율이 1.5배 이상 오르고 구간이 0 을 포함하지 않는 달은 원문을 읽습니다
- 원문을 읽기 전에는 늘어난 이유를 「…로 보인다」고 씁니다

구간에는 표본 변동만 나타나므로 달력 · 원문 · 버전은 따로 살펴봅니다.

</div>

> **[문제] 예시 창구의 헬스장 테마 민원은 2월 40건, 3월 55건이고, 두 달의 민원 전체는 400건과 500건입니다. 두 달을 비교하는 문장으로 맞는 것은 무엇일까요?**
>
> 1. 헬스장 테마 민원 건수가 37.5% 늘었으니 비율도 37.5% 늘었다
> 2. 헬스장 테마 민원은 37.5%, 그 달 민원 전체는 25% 늘어 비율은 10% 에서 11% 로 조금 높아졌다
> 3. 그 달 민원 전체도 늘었으므로, 비율을 구하지 않아도 헬스장 민원은 늘지 않았다고 말할 수 있다

## 2. 달력 효과

월 건수가 달라지면 그 달 민원이 늘거나 줄었다고 할 수 있을까요?

영업일이 적은 달은 민원이 그대로여도 월 건수가 작아지므로 월 건수만으로는 알 수 없습니다.

**달력 효과(calendar effect)** 는 달마다 영업일 · 공휴일 수가 달라 월 건수가 달라지는 현상입니다. 여기서 월 건수는 그 달에 접수된 민원 수입니다.

### 영업일당 건수 [필수]

**영업일** 은 월요일～금요일 가운데 공휴일을 뺀 날입니다. 공휴일에는 상담 창구가 쉬어 민원이 거의 접수되지 않습니다. 그래서 영업일이 많은 달은 민원이 그대로여도 월 건수가 커집니다.

예시 창구는 1월과 2월의 월 건수가 같지만 2월은 설 연휴로 영업일이 이틀 적습니다.

| 달 | 월 건수 | 영업일 |
|---|---:|---:|
| 1월 | 400 | 20 |
| 2월(설 연휴) | 400 | 18 |
| 3월 | 500 | 22 |

영업일 수가 다른 달의 월 건수를 그대로 비교하면, 영업일 수의 차이와 민원의 차이가 구분되지 않습니다.

**영업일당 건수** 는 그 달 건수를 그 달 영업일 수로 나눈 값입니다. 영업일 수로 나누므로, 영업일 수가 다른 달도 같은 기준으로 비교할 수 있습니다.

$$
\text{영업일당 건수}_m = \frac{c_m}{b_m}
$$

- $m$: 달(1월 · 2월 · 3월 …)
- $c_m$: 달 $m$ 의 월 건수. 주말 · 공휴일에 접수된 민원도 포함합니다
- $b_m$: 달 $m$ 의 영업일 수

1～3월에 대입합니다.

```text
1월: 400 ÷ 20 = 20.0건
2월: 400 ÷ 18 = 22.2건
3월: 500 ÷ 22 = 22.7건
```

**왜 날 수가 아니라 영업일 수로 나누나요?**

- 민원이 접수되는 날은 거의 영업일뿐이라, 분모도 접수가 있는 날의 수로 맞춥니다.
- 날 수로 나누면 쉬는 날이 많은 달은 분모만 커져, 영업일 하루 접수가 같아도 값이 작게 나옵니다.

**영업일 수가 다른 두 달은 월 건수로 비교해도 될까요?**

두 쌍을 비교하면 월 건수로 본 결론과 영업일당 건수로 본 결론이 다릅니다.

| 비교 | 월 건수 | 영업일당 건수 |
|---|---|---|
| 1월 → 2월 | 같음 | 2월이 많음 |
| 2월 → 3월 | 3월이 25% 많음 | 3월이 약 2% 많음 |

월 건수 차이 25% 는 대부분 3월 영업일이 4일 많아서 생겼습니다. 「3월에 민원이 25% 늘었다」는 영업일 수의 차이를 민원의 차이로 본 결론입니다.

> 영업일 수가 다른 달끼리는 월 건수가 아니라 영업일당 건수로 비교합니다.

> **[문제] 1월과 2월의 영업일당 건수**
>
> 1월과 2월은 월 건수가 400건으로 같고, 영업일은 1월 20일 · 2월 18일입니다. 두 달의 영업일당 건수를 구하고, 「2월 민원은 1월과 같았다」는 문장을 고쳐 한 문장으로 적어 보세요.

1～3월에 400～500건이던 예시 창구의 월 건수가 4월에는 크게 줄었습니다. 무엇을 먼저 볼지 예상해 봅니다.

> **[예측] 월 건수만 보면 4월 민원은 113건으로 3월 500건보다 77% 적습니다. 이 감소가 민원이 줄어서인지 확인하려면 무엇을 먼저 볼까요?**
>
> 1. 4월의 영업일 수
> 2. 4월의 하루 접수 건수
> 3. 4월의 헬스장 테마 비율
>
> 답을 정한 뒤 아래를 읽어 보세요(정답 채점은 없습니다).

### 잘린 달 [필수]

답은 둘째 보기 「4월의 하루 접수 건수」입니다. 영업일 수로 77% 감소를 설명하려면 4월 영업일이 5일쯤이어야 합니다. 한 달 영업일은 보통 20일 안팎이라 영업일 수로는 설명되지 않습니다. 자료가 끊겨도 비율은 거의 그대로라, 감소가 어디서 생겼는지는 비율에 드러나지 않습니다.

4월 접수를 하루 단위로 나누면 접수가 끊긴 날이 드러납니다. 4-7 까지는 영업일마다 20건 이상 접수되었지만 4-8 부터는 거의 접수되지 않았습니다.

이 자료의 접수는 4-7 을 끝으로 끊긴 것으로 봅니다. 4월에 민원이 실제로 줄었는지는 끊긴 뒤의 접수가 자료에 없어 알 수 없습니다.

**잘린 달(truncated period)** 은 자료 수집이 그 달 중간에서 끊겨, 그 달 민원의 일부만 자료에 있는 달입니다. 예시 창구의 4월이 잘린 달입니다.

> 잘린 달의 월 건수는 1일부터 말일까지 받은 달의 월 건수와 비교하지 않습니다.

**끊긴 날은 어떻게 정하나요?**

**끊긴 날** 은 일별 건수가 영업일당 건수의 절반 아래로 떨어진 뒤 다시 오르지 않는 첫 영업일입니다. 예시 창구의 영업일당 건수는 22건 안팎이라 기준은 11건이고 4-8 이 끊긴 날입니다. 접수가 이어진 마지막 날은 그 전 영업일인 4-7 입니다.

절반은 평소의 22건 안팎보다 낮고 끊긴 뒤의 0～2건보다는 높습니다. 기준을 절반으로 정하면 평소보다 조금 적게 들어온 날을 끊긴 날로 잘못 보지 않습니다. 아래 그림에서 점선 아래로 처음 떨어진 하늘색 막대가 끊긴 날 4-8 입니다.

잘린 달은 다른 자료에서도 생기므로, 달마다 비교하기 전에 아래 순서로 접수가 이어진 마지막 날부터 찾습니다.

| 순서 | 할 일 | 예시 창구 |
|---|---|---|
| ① | 접수일자의 가장 이른 날과 가장 늦은 날을 본다 | 가장 늦은 날 4-9 |
| ② | 일별 건수를 그리고, 앞 달의 영업일당 건수를 구한다 | 3월 22.7건 |
| ③ | 끊긴 날의 기준이 될 하루 건수를 정한다 | 영업일당 건수의 절반, 약 11건 |
| ④ | 기준 아래로 떨어진 뒤 다시 오르지 않는 첫 영업일과 그날부터의 건수 합을 구한다 | 4-8, 3건 |
| ⑤ | 추이 표의 마지막 달을 정하고, 뺀 달이나 기간으로 부른 달은 표 아래에 쓴다 | 4월을 빼거나 「4-1～7 접수분」 |

가장 늦은 접수일은 4-9 이지만 4-8 부터 접수된 민원은 모두 3건뿐입니다. 접수가 이어진 마지막 날은 접수일자의 끝이 아니라 일별 건수로 정합니다.

4월을 접수가 이어진 마지막 날까지로 자를 때와 말일까지 포함할 때를 비교합니다.

| 4월 1일부터 | 누적 건수 | 영업일 | 영업일당 건수 | 헬스장 테마 비율 |
|---|---:|---:|---:|---:|
| 4-7 까지 | 110 | 5 | 22.0 | 11 ÷ 110 = 10.0% |
| 4-30 까지 | 113 | 22 | 5.1 | 11 ÷ 113 = 9.7% |

- 영업일당 건수는 4-7 까지로 자르면 3월의 0.97배로 거의 같습니다. 말일까지 포함해 나누면 끊긴 뒤의 영업일 17일이 분모에 더해져 크게 떨어집니다.
- 헬스장 테마 비율은 거의 그대로입니다. 비율의 분모는 영업일 수가 아니라 민원 수라서, 끊긴 뒤 접수된 3건이 분모에 더해질 뿐입니다.

그래서 잘린 달의 영업일당 건수는 건수와 영업일을 모두 접수가 이어진 마지막 날까지만 계산합니다.

다만 비율이 거의 같아도 잘린 달의 비율은 그대로 쓰지 않습니다.

- 분모가 3월의 5분의 1 남짓으로 작습니다. 오차 한계는 분모의 제곱근에 반비례해, 같은 비율이어도 약 √5 ≈ 2.2배 큽니다.
- 한 달을 대표하지 않습니다. 첫 주의 민원만 있어, 그 주의 주제 구성이 한 달 전체와 같은지 이 자료로는 알 수 없습니다.

월별 추이 표에서 4월은 빼거나 「4-1～7 접수분」처럼 기간으로 부릅니다.

**비율도 영업일 수로 나눠야 하나요?**

영업일이 적은 달은 분자와 분모가 함께 작아져 비율은 영업일 수의 영향을 거의 받지 않습니다. 달력 효과는 월 건수를 비교할 때 고려하고 비율은 영업일 수로 나누지 않습니다.

<div class="lms-summary">

**정리**

월 건수는 영업일 수와 자료가 끝나는 날에 따라 달라져 민원이 그대로여도 크게 바뀔 수 있습니다. 영업일 수가 다른 달은 영업일당 건수로 비교하고 잘린 달은 일별 건수로 끊긴 날을 찾습니다.

**할 일**

- 잘린 달은 접수가 이어진 마지막 날까지 잘라 영업일당 건수를 다시 구합니다
- 월별 추이 표에서는 잘린 달을 빼거나, 「4-1～7 접수분」처럼 달 이름 대신 기간을 씁니다

예시 창구의 4월 월 건수가 줄어든 것은 자료가 4-7 을 끝으로 끊긴 결과입니다. 끊긴 뒤에 민원이 줄었는지는 자료로 알 수 없습니다.

## 3. 원문을 확인할 달

헬스장 테마의 비율이 1월 5% 에서 2월 10% 로 두 배가 되었습니다. 「2월에 헬스장 민원이 늘었다」고 할 수 있을까요?

2월 원문을 읽으면 헬스장 B 한 곳의 민원이 절반을 넘습니다. 헬스장 B 를 빼면 2월과 3월 모두 1월보다 늘어난 것이 없습니다.

> 원문을 확인할 달은 앞 달보다 비율이 1.5배 이상 오르고 두 달 차이의 95% 구간이 0 을 포함하지 않는 달입니다.

헬스장 테마의 2월은 1월의 2배이고 두 달 차이의 95% 구간도 0 을 포함하지 않아 두 조건을 모두 만족합니다.

한 달 민원이 수만 건이면 1%p 보다 작은 차이도 구간이 0 을 포함하지 않습니다. 그래서 구간만으로는 원문을 확인할 달을 정하지 않습니다. 1.5배는 이 수업에서 정한 기준입니다.

### 드릴다운 [필수]

**드릴다운(drill-down)** 은 월별 합계에서 민원 한 건 한 건의 원문으로 내려가 무엇이 늘었는지 찾는 일입니다. 원문을 확인할 달마다 합니다. 예를 들어 2월 헬스장 민원의 제목에 같은 상호가 거듭 나오면, 늘어난 것은 헬스장 테마 전체가 아니라 한 판매처의 민원입니다.

원문은 무작위로 골라 읽습니다. 가장 많이 나온 제목만 읽으면 그 달 전체가 한 판매처의 일로 보입니다. 토픽 이름을 검토할 때 대표 문서와 함께 무작위 문서를 읽은 것도 같은 이유였습니다.

**드릴다운은 어떤 순서로 하나요?**

- ① 원문을 확인할 달과 앞 달의 그 테마 민원 수를 구합니다. 헬스장 테마라면 1월과 2월의 민원 수입니다.
- ② 원문을 확인할 달의 그 테마 민원을 무작위로 10건 읽고 같은 상호가 여러 번 나오는지 봅니다.
- ③ 거듭 나오는 상호가 있으면 그 판매처의 비중을 구합니다. 분모는 읽은 10건이 아니라 그 달 그 테마의 민원 전체입니다.
- ④ 비중이 0.5 를 넘으면 그 판매처의 민원을 빼고 비율을 다시 구합니다.
- ⑤ 늘어난 이유는 원문에 있는 말로만 씁니다.

> 원문을 확인할 달에는 늘어난 이유를 원문에서 찾은 뒤에 씁니다.

### 한 판매처 집중 [필수]

2월 헬스장 민원 40건 가운데 무작위로 10건을 골라 제목을 읽었습니다. 예시 창구 자료에는 판매처 열이 없어, 판매처는 제목에 있는 상호로 찾습니다.

| 번호 | 민원 제목(상호는 헬스장 B 로 바꿈) |
|---:|---|
| 1 | 헬스장 B 회원권 환불 거부 |
| 2 | 헬스장 B 3개월 이용권 해지 위약금 과다 |
| 3 | 필라테스 학원 수강료 환불 지연 |
| 4 | 헬스장 B 갑작스러운 휴관 후 환불 거부 |
| 5 | 수영장 강습 중도 해지 환불 |
| 6 | 헬스장 B PT 잔여 횟수 환불 요구 |
| 7 | 요가원 회원권 중도 해지 환불 거절 |
| 8 | 헬스장 B 연간 회원권 해지 거절 |
| 9 | 헬스장 B 남은 기간 환불 금액 다툼 |
| 10 | 크로스핏 체육관 계약 해지 위약금 |

10건 가운데 6건이 「헬스장 B」 한 곳의 민원입니다. 나머지 4건은 필라테스 학원 · 수영장 · 요가원 · 크로스핏 체육관으로 모두 다른 곳입니다.

**한 판매처 집중** 은 한 판매처의 민원이 그 달 그 테마 민원의 절반을 넘게 차지한 경우입니다. 가장 많이 나온 판매처 하나의 비중 $s$ 로 판단합니다.

$$
s = \frac{\text{그 판매처 상호가 제목에 있는 민원 수}}{\text{그 달 그 테마의 민원 수}}
$$

- 분자: 가장 많이 나온 판매처 하나의 민원 수
- 분모: 그 달 그 테마의 민원 전체 수. 읽은 10건이 아닙니다

> 한 판매처의 비중이 0.5 를 넘으면($s$ > 0.5) 한 판매처 집중으로 봅니다.

읽은 10건으로는 어느 상호의 비중을 구할지만 정합니다. 다른 10건을 고르면 헬스장 B 의 건수도 달라질 수 있습니다. 2월 헬스장 민원 40건 전체에서 헬스장 B 상호가 있는 제목을 찾으니 24건이었습니다.

$$
s = \frac{24}{40} = 0.6
$$

0.5 를 넘으므로 2월 헬스장 민원은 헬스장 B 한 곳에 집중되어 있습니다. 이 예는 읽은 10건의 비율도 0.6 이지만 두 값이 늘 같지는 않습니다.

**왜 절반을 기준으로 하나요?**

- $s$ 가 0.5 를 넘으면 그 판매처 하나의 민원이 나머지 판매처를 모두 합친 것보다 많습니다.
- 이때 「헬스장 민원이 늘었다」고 결론 내리면 한 판매처의 일을 여러 판매처의 일로 보게 됩니다.

**헬스장 B 를 뺀 증가가 있나요?**

헬스장 B 의 민원을 빼고 비율을 다시 구합니다. 비교 기준은 헬스장 B 민원이 늘기 전인 1월입니다. 1월과 3월에서도 제목에 헬스장 B 상호가 있는 민원을 찾아 같은 방식으로 뺍니다. 뺀 비율도 그 달 민원 가운데 몇 % 인지를 나타내므로 분모는 그 달 전체 민원 그대로 씁니다.

| 달 | 헬스장 B | 헬스장 B 포함 | 헬스장 B 를 뺌 | 뺀 비율의 1월 대비 차이(95% 구간) |
|---|---:|---:|---:|---|
| 1월 | 1 | 20 ÷ 400 = 5.00% | 19 ÷ 400 = 4.75% | 기준 |
| 2월 | 24 | 40 ÷ 400 = 10.00% | 16 ÷ 400 = 4.00% | −0.75%p(−3.58～2.08%p) |
| 3월 | 30 | 55 ÷ 500 = 11.00% | 25 ÷ 500 = 5.00% | +0.25%p(−2.58～3.08%p) |

3월에도 헬스장 B 의 비중이 30 ÷ 55 = 0.545 라 한 판매처 집중입니다.

헬스장 B 를 빼면 두 달 모두 1월과 비교한 차이의 95% 구간이 0 을 포함합니다. 헬스장 테마의 증가는 헬스장 B 한 곳의 민원이고 헬스장 B 를 뺀 증가는 없습니다.

<details class="lms-extra"><summary>자세히 — 헬스장 B 를 뺀 비율의 계산</summary>

1월과의 차이의 95% 구간은 앞에서 두 달 차이의 구간을 구한 식으로 구합니다.

```text
2월: 4.00% − 4.75% = −0.75%p
     오차 한계 1.96 × √(0.0475 × 0.9525 ÷ 400 + 0.04 × 0.96 ÷ 400) = 2.83%p → −3.58～2.08%p
3월: 5.00% − 4.75% = +0.25%p
     오차 한계 1.96 × √(0.0475 × 0.9525 ÷ 400 + 0.05 × 0.95 ÷ 500) = 2.83%p → −2.58～3.08%p
```

판매처를 뺀 비율은 포함한 비율의 (1 − $s$)배입니다. 분모를 그 달 전체 민원 그대로 쓰기 때문에 생기는 관계입니다.

- $a$: 그 달 그 테마의 민원 수(2월 헬스장 민원 40건)
- $b$: 그 가운데 그 판매처의 민원 수(헬스장 B 24건)
- $T$: 그 달 전체 민원 수(2월 400건)

포함한 비율은 $a / T$ 이고 $s = b / a$ 입니다. 뺀 비율의 분자 · 분모에 $a$ 를 곱하고 두 분수의 곱으로 바꾼 뒤 두 정의를 대입합니다.

$$
\begin{aligned}
\text{뺀 비율:}\quad & \frac{a - b}{T} \\
\text{분자 · 분모에 } a \text{ 를 곱함:}\quad & = \frac{a\,(a - b)}{T\,a} \\
\text{두 분수의 곱으로 바꿈:}\quad & = \frac{a}{T} \times \frac{a - b}{a} \\
\text{뒤 분수를 둘로 나눔:}\quad & = \frac{a}{T} \times \left(1 - \frac{b}{a}\right) \\
\text{앞의 두 정의를 대입:}\quad & = \text{포함한 비율} \times (1 - s)
\end{aligned}
$$

2월 값을 대입하면 10% × (1 − 0.6) = 4.0% 입니다. 헬스장 B 를 직접 빼서 나눈 (40 − 24) ÷ 400 = 4.0% 와 같습니다.

이 관계로 $s$ 에 따라 추이 표에 쓰는 비율을 정합니다. 0.5 는 한 판매처 집중의 기준과 같고 0.3 은 이 수업에서 정한 기준입니다.

| $s$ | 뜻 | 추이 표에 쓰는 비율 |
|---|---|---|
| 0.5 초과 | 한 판매처 집중 | 그 판매처를 뺀 비율, 집중은 따로 설명 |
| 0.3 초과 0.5 이하 | 집중은 아니지만 한 판매처의 비중이 큼 | 포함한 비율과 뺀 비율 둘 다 |
| 0.3 이하 | 한 판매처의 비중이 작음 | 포함한 비율만 |

$s$ 가 0.3 을 넘으면 뺀 비율이 포함한 비율보다 30% 넘게 작아, 포함한 비율 하나로는 그 판매처의 영향을 알 수 없습니다.

</details>

한 판매처 집중과 그 판매처를 뺀 증가는 조치가 달라 결과에서 구분해 말합니다.

- 한 판매처 집중이면 그 판매처에 조치합니다.
- 그 판매처를 뺀 증가가 있으면 여러 판매처의 해지 · 환불 관행을 살핍니다.

2월 헬스장 테마의 결론입니다.

> 2월 헬스장 테마 민원의 60% 는 헬스장 B 한 곳의 민원이다.
> 헬스장 B 를 빼면 1월과 차이가 없어, 헬스장 B 를 뺀 증가는 없다.

표와 결론의 「헬스장 B」는 원문의 상호를 바꾼 이름입니다. 결론에서 원문의 상호를 부르면 민원을 낸 사람의 주장만으로 그 판매처를 지목하게 됩니다.

**헬스장 B 의 민원은 어떻게 따로 기록하나요?**

**판매처 표시 열** 은 민원마다 한 판매처 집중으로 드러난 판매처의 민원인지를 「헬스장 B」 또는 「없음」으로 기록하는 열입니다. 앞의 10건 가운데 3건에 판매처 표시 열을 더했습니다.

| 번호 | 민원 제목 | 테마 | 판매처 표시 열 |
|---:|---|---|---|
| 1 | 헬스장 B 회원권 환불 거부 | 헬스장 테마 | 헬스장 B |
| 3 | 필라테스 학원 수강료 환불 지연 | 헬스장 테마 | 없음 |
| 4 | 헬스장 B 갑작스러운 휴관 후 환불 거부 | 헬스장 테마 | 헬스장 B |

- 헬스장 B 민원을 지우거나 테마를 바꾸지 않고 판매처 표시 열을 하나 더 만듭니다. 그래야 헬스장 B 를 포함한 비율과 뺀 비율을 같은 자료에서 구할 수 있습니다.
- 판매처 표시 열은 달과 상관없이 제목에 헬스장 B 상호가 있으면 「헬스장 B」입니다. 한 판매처 집중이 아닌 1월에도 이 열로 헬스장 B 를 뺀 비율을 구할 수 있습니다.

<div class="lms-summary">

**정리**

원문을 확인할 달에는 그 달 민원 제목을 무작위로 읽어 여러 번 나오는 상호를 찾습니다. 그 판매처의 비중은 그 달 그 테마의 민원 전체에서 구합니다. 비중이 0.5 를 넘으면 한 판매처 집중이고 그 판매처를 뺀 증가가 있는지 다시 알아봅니다.

**할 일**

- 판매처는 원래 열을 고치지 않고 판매처 표시 열에 따로 기록합니다
- 결과에서는 한 판매처 집중과 그 판매처를 뺀 증가를 구분해 말합니다

헬스장 테마의 2월 증가분은 헬스장 B 한 곳에서 나와 헬스장 테마 민원이 늘었다고 보지 않습니다.

</div>

>

## 4. 버전 기록

같은 민원인데 결과 파일마다 비율이 다르면 무엇부터 의심해야 할까요?

민원을 고른 사전이나 코드북이 바뀌었는지부터 봅니다. 사전만 고쳐도 같은 민원에서 비율이 달라지므로, 값마다 어느 사전으로 구했는지 남겨 둡니다.

예시 토픽 「헬스장」의 이름 초안을 검토할 때 읽은 무작위 10건을 다시 씁니다. 그때는 민원이 이름에 맞는지 읽고 판단했습니다. 여기서는 사전의 단어가 있는지로 「환불 문제」 민원을 고릅니다.

**「환불 문제」 사전**은 환불 문제를 말하는 민원을 고르는 단어 목록입니다. 민원 제목을 형태소로 나눈 명사에 사전 단어가 하나라도 있으면 「환불 문제」 민원으로 봅니다.

- 사전 v1 = {환불}
- 사전 v2 = {환불, 위약금, 해지}. v1 에 위약금 · 해지를 더했습니다.

### 같은 민원에서 값이 달라지는가 [필수]

같은 10건에서 「환불 문제」 민원을 두 사전으로 골라 봅니다.

| # | 민원 제목 | 사전 v1 {환불} | 사전 v2 {환불, 위약금, 해지} |
|---:|---|:---:|:---:|
| 1 | 헬스장 회원권 중도 해지 환불 요청 | 예 | 예 |
| 2 | 헬스장 회원권 환불을 거절당했어요 | 예 | 예 |
| 3 | 피트니스 센터 이용권 환불을 안 해 줍니다 | 예 | 예 |
| 4 | 헬스장 중도 해지 위약금이 과합니다 | 아니오 | 예(위약금 · 해지) |
| 5 | PT 10회권 남은 횟수 환불 요청 | 예 | 예 |
| 6 | 요가원 회원권을 중도에 해지하고 싶어요 | 아니오 | 예(해지) |
| 7 | 헬스장 트레이너가 너무 불친절해요 | 아니오 | 아니오 |
| 8 | 필라테스 수강권 환불을 거절당했어요 | 예 | 예 |
| 9 | 헬스장 양도 수수료가 너무 비싸요 | 아니오 | 아니오 |
| 10 | 수영장 강습 취소 환불 문의 | 예 | 예 |
| | 「환불 문제」 비율 | 6/10 = 60% | 8/10 = 80% |

민원은 같은데 두 비율이 20%p 다릅니다.

- 4번과 6번만 v2 에서 「예」로 바뀌었습니다. 6번의 「해지하고」도 형태소로 나누면 명사 「해지」가 나옵니다.
- v2 는 v1 의 단어를 모두 포함해서 v1 로 고른 민원은 v2 로도 고릅니다. 사전을 넓히면 비율은 내려가지 않습니다.

즉, 같은 10건에서 나온 20%p 는 **민원의 변화가 아니라 사전을 바꿔 생긴 차이**입니다.

**버전(version)** 은 민원을 고르거나 값을 매기는 데 쓴 것을 바꿀 때마다 정하는 이름입니다. 대상은 사전 · 코드북 · 토픽 배정 · 프롬프트 · 모델이고 사전 v1 · v2 가 그 예입니다. **버전 기록**은 값마다 그 값을 구한 버전을 결과 값 옆에 남기는 일입니다.

**버전이 다른 두 값을 비교하면 어떻게 될까요?**

지난번 결과 파일에는 사전 v1 로 구한 60% 가, 이번 파일에는 v2 로 구한 80% 가 있다고 해 봅니다. 두 값만 비교하면 「환불 문제」 민원이 20%p 늘어난 것처럼 보이지만 늘어난 민원은 하나도 없습니다.

두 달을 비교할 때도 같습니다. 한 달은 v1, 다른 달은 v2 로 구하면 두 달을 비교한 값에 사전을 바꿔 생긴 차이가 더해집니다.

```text
버전이 다른 두 달의 차이 = 민원의 변화 + 사전을 바꿔 생긴 차이
```

민원이 줄어든 달도 사전을 바꿔 생긴 차이가 더 크면 늘어난 달로 나옵니다. 프롬프트나 모델을 바꿔도 같은 일이 생깁니다. 골드 표본으로 LLM 라벨을 확인할 때 표본과 전체를 같은 모델 · 같은 프롬프트 버전으로 라벨링한 것도 이 때문입니다.

> 버전이 다른 두 값의 차이는 민원의 변화로 보지 않습니다.

소프트웨어의 버전 이름은 흔히 숫자 3개로 씁니다.

앞 수가 바뀌면 이전 버전의 결과와 그대로 비교할 수 없습니다. 사전이 v1 에서 v2 로 바뀐 것도 같습니다.

</details>

### 버전 열에 남길 4가지 [필수]

값 옆에 사전 이름이 있으면 파일을 받은 사람이 사전이 바뀌었는지 바로 알 수 있습니다. **버전 열**은 값마다 어느 버전으로 계산했는지 기록한 열입니다.

| 민원 | 「환불 문제」 비율 | 버전 |
|---|---:|---|
| 같은 10건 | 60% | 사전 v1 |
| 같은 10건 | 80% | 사전 v2 |

**버전 열에는 무엇을 남기나요?**

위 버전 열에는 대상과 번호만 있어서 v2 를 언제 만들었고 무엇이 바뀌었는지 모릅니다. 그래서 버전 열에는 4가지를 남깁니다.

- 버전을 매긴 대상: 사전 · 코드북 · 토픽 배정 · 프롬프트 · 모델 가운데 무엇인지
- 이름과 번호: 「사전 v2」처럼. LLM 라벨이면 모델 이름 · 모델 버전 · 프롬프트 버전입니다
- 만든 날짜
- 앞 버전에서 바뀐 것: 한 문장으로

80% 옆의 버전 열에 4가지를 모두 채우면 v1 과 무엇이 다른지까지 알 수 있습니다.

| 대상 | 이름과 번호 | 만든 날짜 | 앞 버전에서 바뀐 것 |
|---|---|---|---|
| 사전 | v2 | v2 를 만든 날(연-월-일) | v1 에 위약금 · 해지를 더함 |

**사전이나 코드북을 고치면 무엇을 다시 하나요?**

사전으로 고른 값이나, 사람 · LLM 이 코드북 기준으로 다시 분류한 값은 버전이 바뀌면 달라질 수 있습니다.

```text
새 버전으로 모든 달을 처음부터 다시 집계 (고친 달부터가 아님)
→ 같은 민원의 옛 버전 값과 새 버전 값을 비교 (같은 10건이면 v1 60% 대 v2 80%)
→ 추이 그림에는 새 버전의 선 하나만, 버전을 바꾼 날짜와 차이는 각주로
```

토픽 배정으로 집계한 테마 비율은 코드북의 기준 문장을 고쳐도 그대로입니다. 지금 쓰는 민원 테마 코드북은 v0 입니다. 테마 이름을 검토한 원문 10건을 보고 예식 테마의 기준 문장을 다시 썼으므로, 고친 코드북은 다음 버전 v1 로 기록합니다. 다음 달 민원을 테마로 분류하거나 LLM 에 기준을 입력할 때는 고친 기준이 쓰이기 때문입니다.

**토픽 배정은 왜 한 버전으로 고정할까요?**

BERTopic 은 적합할 때마다 민원이 많은 토픽부터 0, 1, 2 … 로 번호를 새로 매깁니다. 시드를 바꾸거나 달마다 따로 적합하면 같은 번호가 다른 내용의 토픽일 수 있습니다.

시드 안정성을 볼 때 쓴 민원 12건(문서 1～12)에서도 같은 일이 있었습니다. 시드 0 의 「항공권 환불」과 「여행사 패키지 환불」이 시드 1 에서는 「여행·항공 환불」 하나로 합쳐졌습니다.

| 토픽 번호 | 시드 0 | 시드 1 |
|---:|---|---|
| 0 | 휴대폰 요금(문서 1～5) | 여행·항공 환불(문서 6～12) |
| 1 | 항공권 환불(문서 6～9) | 휴대폰 요금(문서 1～5) |
| 2 | 여행사 패키지 환불(문서 10～12) | 없음 |

같은 번호 0 이 시드 0 에서는 휴대폰 요금이고 시드 1 에서는 여행·항공 환불입니다. 그래서 다른 시드의 토픽은 같은 번호로 부르지 않고 토픽 배정의 버전에는 시드까지 기록합니다. 월별 비율도 토픽 배정 한 버전을 모든 달에 그대로 쓰고 달마다 나눠 구합니다.

<div class="lms-summary">

**정리**

같은 10건도 「환불 문제」 사전 v1 로는 60%, v2 로는 80% 입니다. 버전이 다른 두 값의 차이는 민원의 변화로 보지 않습니다.

**할 일**

- 추이 표의 모든 값 옆에 버전 열을 만들고 4가지(대상 · 이름과 번호 · 만든 날짜 · 앞 버전에서 바뀐 것)를 남깁니다
- 사전으로 고르거나 코드북 기준으로 분류한 값은 모든 달을 새 버전으로 처음부터 다시 집계합니다

토픽 배정도 한 버전으로 고정하고 시드까지 기록합니다.

</div>

헬스장 테마는 4가지를 확인하자 「늘었다」는 근거가 없어졌습니다. 민원 50만 건의 「여행」은 반대로 같은 절차를 거쳐도 감소가 남습니다.

2월 민원 3건으로 명사 열과 「여행」 민원을 보겠습니다.

셋째 민원은 펜션 취소 민원이지만 명사 목록에 「여행」이 없어 아니오입니다. 「여행」의 문서 비율은 「여행」이 있는 민원 수를 그 달 전체 민원 수로 나눈 값입니다.

1～9월의 「여행」 문서 비율은 2월이 가장 높고 4월부터는 1% 아래입니다.

```text
「여행」 문서 비율(%)
1～4월: 5.64 → 6.54 → 2.34 → 0.93
5～9월: 0.71 → 0.59 → 0.64 → 0.68 → 0.53
```

2월과 4월의 건수를 그 달 전체 민원과 비교합니다.

| 건수 | 2월 | 4월 | 4월 ÷ 2월 |
|---|---:|---:|---:|
| 「여행」이 있는 민원 | 4,022 | 461 | 약 9분의 1 |
| 그 달 전체 민원 | 61,536 | 49,404 | 약 5분의 4 |

```text
2월: 4,022 ÷ 61,536 = 6.54%
4월:   461 ÷ 49,404 = 0.93%
```

헬스장 테마는 2월 → 3월에 건수가 37.5% 늘어도 비율은 10% 에서 11% 였습니다. 「여행」 민원은 전체 민원보다 크게 줄어, 비율도 함께 내려갔습니다.

1～10월의 두 수로 달마다 문서 비율과 전월 대비를 구합니다.

````python
# 「여행」이 있는 민원 수를 그 달 전체 민원 수로 나눠 달마다 문서 비율을 구한다(10월은 접수가 10-08 을 끝으로 끊긴 달)
import pandas as pd

month_n = [52317, 61536, 57942, 49404, 50194, 54436, 55260, 57796, 51589, 9370]   # 그 달 전체 민원
travel_n = [2949, 4022, 1355, 461, 357, 323, 352, 391, 272, 55]                  # 명사 열에 「여행」이 있는 민원

travel = pd.DataFrame({"전체 민원": month_n, "「여행」 민원": travel_n}, index=[f"{m}월" for m in range(1, 11)])
travel["문서 비율(%)"] = (100 * travel["「여행」 민원"] / travel["전체 민원"]).round(2)
travel["전월 대비(%p)"] = travel["문서 비율(%)"].diff().map(lambda v: "없음" if pd.isna(v) else f"{v:+.2f}")
for name, *vals in travel.itertuples():
    print(f"{name}: " + " · ".join(f"{k} {v}" for k, v in zip(travel.columns, vals)))
````

출력에서 비율이 가장 크게 내려간 달은 3월입니다. 2월과 3월의 두 수를 비교하면 무엇이 줄었는지 보입니다.

```text
「여행」 민원: 4,022건 → 1,355건 (약 3분의 1)
전체 민원: 61,536건 → 57,942건 (약 6% 감소)
문서 비율: 6.54% → 2.34% (−4.20%p)
```

다른 민원이 늘어서가 아니라 「여행」 민원이 줄어 비율이 내려갔습니다. 5～9월도 0.53～0.71% 라, 비율은 2월 수준으로 다시 높아지지 않았습니다.

10월은 접수가 10-08 을 끝으로 끊긴 달이라(근거는 아래 「10월 잘린 달」), 비율 추이는 1～9월로 봅니다.

**2월과 4월의 차이는 표본 변동으로 설명되나요?**

민원 50만 건 전체로 구한 비율도 조건이 같은 다른 달이 오면 조금 다르게 나옵니다. 그래서 두 달 차이의 95% 구간을 구합니다. 반올림하기 전의 두 비율 0.06536 과 0.00933 을 대입합니다.

```text
차이: 0.06536 − 0.00933 = 0.05603 → 5.60%p
표준오차: √(0.06536 × 0.93464 ÷ 61,536 + 0.00933 × 0.99067 ÷ 49,404) = 0.001086
오차 한계: 1.96 × 0.001086 = 0.00213 → ±0.21%p
구간: 0.05603 ± 0.00213 → 5.39～5.82%p
(반올림한 두 비율로 빼면 6.54 − 0.93 = 5.61 이라 0.01%p 다르다)
```

구간이 0 을 포함하지 않으므로 2월과 4월의 차이는 표본 변동으로 설명되지 않습니다.

헬스장 테마 두 달 비교보다 한 달 분모가 커서 오차 한계가 더 작습니다.

```text
헬스장 테마 비율: 한 달 분모 400～500건 → 오차 한계 ±4.02%p
여행 문서 비율: 한 달 분모 5만～6만 건 → 오차 한계 ±0.21%p
```

같은 식으로 2월 − 4월과, 구간이 0 을 포함하는 예인 6월 − 7월을 구합니다.

````python
# 두 달의 「여행」 문서 비율 차이와 95% 구간을 계산한다. 맨 아래 줄의 달을 바꿔 다른 두 달도 확인할 수 있다
import numpy as np
import pandas as pd

month_n = [52317, 61536, 57942, 49404, 50194, 54436, 55260, 57796, 51589, 9370]   # 그 달 전체 민원
travel_n = [2949, 4022, 1355, 461, 357, 323, 352, 391, 272, 55]                  # 「여행」이 있는 민원


def month_gap(m1, m2):
    p1, n1 = travel_n[m1 - 1] / month_n[m1 - 1], month_n[m1 - 1]
    p2, n2 = travel_n[m2 - 1] / month_n[m2 - 1], month_n[m2 - 1]
    gap = p1 - p2
    moe = 1.96 * np.sqrt(p1 * (1 - p1) / n1 + p2 * (1 - p2) / n2)        # 오차 한계(± 뒤의 값)
    return {"비교": f"{m1}월 − {m2}월", "차이(%p)": 100 * gap, "오차 한계(%p)": 100 * moe,
            "하한(%p)": 100 * (gap - moe), "상한(%p)": 100 * (gap + moe), "0 을 포함": bool(gap - moe <= 0 <= gap + moe)}


table = pd.DataFrame([month_gap(2, 4), month_gap(6, 7)]).set_index("비교").round(2)
DIGITS = {"차이(%p)": 2, "오차 한계(%p)": 2, "하한(%p)": 2, "상한(%p)": 2}                                   # 반올림 자릿수 그대로 출력(끝자리 0 포함)
table = table.apply(lambda col: col.map(("{:.%df}" % DIGITS[col.name]).format) if col.name in DIGITS else col)
for name, *vals in table.itertuples():
    print(f"{name}: " + " · ".join(f"{k} {v}" for k, v in zip(table.columns, vals)))
````

출력의 「0 을 포함」 열이 False 면 표본 변동만으로는 생기기 어려운 차이입니다. True 면 「바뀌었다」고 보지 않습니다.

6월 − 7월은 차이 −0.04%p, 구간 −0.14～0.05%p 로 0 을 포함합니다. 두 달 사이에 「여행」 비율이 바뀌었다는 근거는 없습니다.

비교를 여러 번 하면 차이가 없어도 우연한 「차이 있음」이 함께 늘어납니다. 민원 50만 건에서 단어 4개로 1～9월의 이웃한 달을 모두 비교하면, 우연한 「차이 있음」이 1.6번쯤 나옵니다.

```text
비교 횟수: 단어 4개(여행 · 코로나 · 마스크 · 숙박) × 1～9월의 이웃한 달 8쌍 = 32번
차이가 하나도 없어도: 32 × 0.05 = 1.6번쯤 「차이 있음」
```

헬스장 · 여행 · 예식 테마의 9번 비교에서는 0.45번쯤이었습니다. 볼 단어를 2개로 먼저 정하면 비교는 16번이라, 우연한 「차이 있음」도 0.8번쯤으로 줄어듭니다.

단어 대신 테마 열로 구하면 분모는 민원 표본 2만 건의 그 달 표본입니다. 2월 여행 테마는 15.02% 로 단어 「여행」 비율의 두 배를 넘습니다. 펜션 취소 민원처럼 「여행」이라는 단어 없이 숙박 · 항공 취소를 쓴 민원도 테마에 속하기 때문입니다.

8월에는 「여행」이 없는 숙박 민원이 늘어 여행 테마 비율이 4월보다 다시 높아졌습니다. 숙박 민원이 늘어난 이유는 8월 원문을 읽기 전에는 알 수 없습니다. 이름이 같아도 단어의 문서 비율과 테마 비율은 다른 값이라, 한 그림의 같은 세로축에 그리지 않습니다.

2월이 원문을 확인할 달이 아니므로, 4월까지의 감소를 2월에 한 판매처로 몰렸다가 빠진 일로 보기도 어렵습니다. 2월 「여행」 민원의 제목에 있는 상호로도 한 판매처 집중이 있는지 알아보았습니다.

```text
판매처 A 를 직접 빼서 나눔: (514 − 195) ÷ 61,536 = 0.52%
```

판매처 A 의 실제 비중은 0.599 보다 크거나 같습니다. 제목에 상호가 없는 판매처 A 민원은 집계에 포함되지 않기 때문입니다. 0.599 는 실제 비중의 하한입니다.

품목과 「배송」·「지연」 조건 없이 민원 50만 건에서 제목에 판매처 A 상호가 있는 민원을 찾으면 1월 43건, 4월 1,233건, 8월 20건으로 추이가 같습니다. 앞의 원문 제목 3개는 4월 판매처 A 민원의 서로 다른 제목 249개에서 시드 0 으로 골랐습니다. 한 제목이 368번 나온 것도 있어서 서로 다른 제목에서 골라야 같은 제목을 여러 번 읽지 않습니다.

다른 분석에서도 같은 판매처가 나왔습니다. BERTopic 으로 토픽을 만들 때의 토픽 34 는 상위 단어가 「판매처 A · 배송 · 옷 · 지연」입니다. 크기 120건인 이 토픽의 4월 문서 46건은 모두 제목에 판매처 A 상호가 있습니다. 상위 단어에 상호가 있는 토픽은 한 판매처 집중의 신호입니다.

가중 로그 오즈로 성별 특징 단어를 구하면 판매처 A 의 상호가 여성 민원의 특징 단어 상위 10개 가운데 다섯째로 나옵니다. 이처럼 상호가 다른 분석에도 그대로 나오므로, 결과에는 실제 상호 대신 「판매처 A」를 씁니다.

</details>

### 6월 비율은 사전 버전에 따라 달라지는가 [필수]

같은 6월 민원에서 「배송 문제」 민원의 비율을 사전 두 개로 구했습니다. 「배송 문제」 민원은 명사 열에 사전 단어가 하나라도 있는 민원입니다. 의류 민원에서 두 단어가 함께 있어야 했던 「배송·지연」과 달리, 모든 민원에서 단어 하나만 있어도 되는 넓은 조건입니다.

| 사전 | 사전 단어 | 6월 「배송 문제」 민원 | 비율 |
|---|---|---:|---:|
| v1 | 배송 · 지연 | 4,778건 | 8.78% |
| v2 | v1 에 택배 · 도착 · 배달을 더함 | 5,329건 | 9.79% |

「환불 문제」 사전의 60% 와 80% 처럼, 같은 6월 민원도 사전만 바꾸자 비율이 1.01%p 달라졌습니다. v2 로만 찾은 민원 551건만큼의 차이라 민원의 변화가 아니라 사전을 바꿔 생긴 차이입니다.

**5월과 6월을 서로 다른 사전으로 비교하면 어떻게 될까요?**

5월은 사전 v1 로 9.64% 입니다.

| 비교 | 5월 | 6월 | 6월 − 5월 |
|---|---:|---:|---:|
| 두 달 모두 사전 v1 | 9.64% | 8.78% | −0.86%p |
| 6월만 사전 v2 | 9.64% | 9.79% | +0.15%p |

6월만 v2 로 본 변화는 같은 v1 로 본 변화에 사전을 바꿔 생긴 차이를 더한 값입니다.

```text
−0.86%p(같은 v1 로 본 변화) + 1.01%p(사전을 바꿔 생긴 차이) = +0.15%p(6월만 v2 로 본 변화)
```

같은 v1 로 본 감소 0.86%p 는 두 달 차이의 오차 한계 0.35%p 보다 커서 표본 변동으로 보기 어렵습니다. 그런데 6월만 v2 로 바꾸면 사전을 바꿔 생긴 차이가 더해져 늘어난 달처럼 보입니다.

프롬프트를 바꿔도 같은 일이 생깁니다. 골드 표본으로 LLM 라벨을 확인할 때 뜻이 같은 프롬프트 5개로 리뷰 100건을 다시 라벨링했습니다. 부정 비율은 51～56% 로 달라졌습니다.

그래서 값마다 버전 열을 만듭니다. 6월 값 옆에 「사전 v2」가 있으면 v1 로 구한 값과는 그대로 비교하지 않습니다.

| 달 | 「배송 문제」 비율 | 버전 | 앞 버전에서 바뀐 것 |
|---|---:|---|---|
| 5월 | 9.64% | 사전 v1 | 첫 버전 |
| 6월 | 9.79% | 사전 v2 | v1 에 택배 · 도착 · 배달을 더함 |

명사 열 · 토픽 배정 파일 · 코드북 v0 처럼 과정 결과로 받아서 쓰는 파일은 버전 이름만 그대로 기록합니다. 만든 날짜와 바뀐 것은 그 파일을 만든 교육 과정이 기록합니다.

| 버전을 매긴 대상 | 버전 이름 |
|---|---|
| 명사 열(형태소 분석기) | kiwipiepy 0.23.1 |
| 토픽 배정 파일 | `topics-v1-20260924` (시드 0) |
| 코드북(테마 이름과 기준) | `codebook-v0-20260924`, 예식 테마의 기준을 다시 쓴 코드북은 `codebook-v1-20261026` |

테마 비율은 토픽 배정 한 버전(시드 0)을 1～9월에 그대로 쓰고 달마다 나눠 구했습니다. 예식 테마의 기준을 다시 쓴 코드북 v1 은 새 버전이지만 테마 열은 토픽 배정으로 정해 v0 로 구한 값과 같습니다.

<details class="lms-extra"><summary>더 깊이 — 토픽 배정: 시점별 계산 · 시드 · 토픽 번호 · 코드북 v0</summary>

BERTopic 은 문장 임베딩을 군집해 토픽을 만들고 c-TF-IDF 로 토픽마다 상위 단어를 매기는 방법입니다. 시간에 따른 토픽을 구할 때도 BERTopic 은 전체 문서로 토픽 모델을 한 번 적합한 뒤, 시점마다 그 토픽의 상위 단어만 다시 계산합니다.

BERTopic 공식 문서는 시간 정보가 없는 것처럼 먼저 적합하라고 설명합니다. 원문은 「fit BERTopic as if there were no temporal aspect」입니다.

<img src="https://maartengr.github.io/BERTopic/getting_started/topicsovertime/topicsovertime.svg" alt="토픽마다 문서를 시점별로 나누고 시점마다 c-TF-IDF 를 다시 계산하는 흐름도" width="562" height="405" />
<p align="center"><em>▲ BERTopic 의 시점별 토픽 계산. 토픽 1～n 의 문서를 시점(Timestep) 1～m 으로 다시 나눈다. 시점마다 나눈 문서에 미리 적합한 c-TF-IDF 로 상위 단어를 다시 매긴다. 아래쪽 두 상자는 시점별 c-TF-IDF 를 전체 값이나 앞 시점 값과 평균해 조정하는 단계이며 생략할 수 있다 (출처: BERTopic 문서 「Dynamic Topic Modeling」, Maarten Grootendorst, MIT)</em></p>

시점마다 상위 단어를 다시 매기려면 c-TF-IDF 를 달마다 한 번 더 계산합니다. 토픽 배정은 그대로이고 시점마다 상위 단어만 달라집니다. 여기서는 상위 단어도 다시 계산하지 않고 배정을 고정한 채 비율만 구합니다.

시드에 따른 차이는 시드 안정성에서 다뤘습니다. 시드 0 · 1 · 2 로 세 번 적합한 배정을 둘씩 비교했고 ARI 는 0.821～0.841 이었습니다. ARI 는 두 배정이 문서를 나눈 결과가 완전히 같으면 1, 우연 수준이면 0 입니다.

세 값 모두 1 에 가까워 같은 토픽에 속한 문서들은 대체로 같았습니다. 다만 일부 문서는 다른 토픽이나 이상치(−1)로 배정되었습니다. 같은 버전 이름이라도 시드가 다르면 일부 문서의 배정이 달라지므로 버전 이름 뒤에 「(시드 0)」까지 기록합니다.

코드북 v0 에 있는 토픽 번호도 `topics-v1-20260924` 시드 0 의 번호라 번호는 버전 이름과 짝지어 기록합니다.

코드북 v0 에도 이름을 고쳐 온 기록이 있습니다. 큰 토픽 8개의 LLM 이름 초안은 무작위 10건으로 검토했습니다. 토픽 5 의 초안 「세탁 관련 민원」은 첫 검토에서 10건 가운데 4건만 맞았습니다. 이름을 두 번 고쳐 「세탁·수선 사고(훼손·분실)」로 바꾸자 10건이 모두 맞았습니다.

`codebook-v0-20260924` 는 검토를 마친 테마 이름 8개를 처음 모은 코드북입니다. 그 전의 이름 초안에는 버전 번호를 매기지 않았습니다. 예식 테마의 기준을 다시 쓴 코드북부터는 v1 로 기록합니다.

명사 열의 분석 설정에는 품사도 포함됩니다. kiwipiepy 0.23.1 의 일반명사 NNG 와 고유명사 NNP 만 추출했습니다. 이 분석기의 기본 단어 사전은 사전 v1 · v2 와 다른 것입니다.

**4가지를 확인해 무엇을 알았나요?**

| 확인 | 대상 | 민원 50만 건에서 알게 된 것 | 다음 할 일 |
|---|---|---|---|
| 분모 | 「여행」 민원 | 2월 − 4월 차이 5.60%p, 95% 구간 5.39～5.82%p 로 0 을 포함하지 않음 | 그 달 전체로 나눈 비율과 두 달 차이의 95% 구간으로 비교한다 |
| 달력 | 월 건수 | 10월은 접수가 10-08 을 끝으로 끊긴 달. 10-08 까지의 영업일당 건수는 9월의 약 93% | 잘린 10월은 추이 표에서 빼거나 접수 기간으로 부른다 |
| 원문 | 의류 「배송·지연」 민원 | 원문을 확인할 달인 4월의 60% 가 판매처 A. 판매처 A 를 빼도 4월 1.13% 로 1월 0.53% 의 두 배 남짓 | 판매처 표시 열로 판매처 A 의 민원을 따로 기록하고, 뺀 비율도 계산한다 |
| 버전 | 「배송 문제」 민원 | 같은 6월 비율이 사전 v1 로 8.78%, v2 로 9.79% | 모든 달을 한 버전으로 집계한다 |

**「줄었다」는 문장은 두 비율만으로 충분한가요?**

2월과 4월의 두 비율만으로는 두 가지를 알 수 없습니다.

- 이 차이가 표본 변동보다 큰지: 두 달 차이의 95% 구간으로 압니다.
- 두 달을 같은 기준으로 구했는지: 분모와 버전으로 압니다.

「여행」처럼 차이가 표본 변동보다 크고 두 비율의 분모와 버전이 같을 때 「변했다」고 할 수 있습니다.

> 「여행」이 있는 민원의 문서 비율은 2월 6.54% 에서 4월 0.93% 로 줄었다.
> 차이의 95% 구간은 5.39～5.82%p 이고 분모는 그 달 전체 민원, 명사 열은 kiwipiepy 0.23.1 버전이다.

잘린 10월을 뺀 이유나 판매처 A 에 몰린 민원처럼 달력 · 원문에서 나온 결과는 문제가 있던 달에만 결론에 더합니다.

<details class="lms-extra"><summary>자세히 — 테마 비율의 구간에 반영되지 않는 오차</summary>

코드북 v0 의 테마 이름은 LLM 이 낸 이름 초안을 무작위 10건으로 검토해 정했습니다. 10건이 이름에 맞는지도 사람이 아니라 LLM 이 판단했고 사람이 라벨링한 골드 표본은 없습니다.

골드 표본으로 LLM 라벨을 확인할 때 쓴 예측 기반 추론(PPI)은 참값이 있는 표본으로 LLM 라벨의 오차만큼 비율을 보정합니다. 그런데 테마 열에는 그런 표본이 없습니다. 테마 비율의 구간에는 표본 변동만 반영되고 라벨이 틀려 생긴 차이는 반영되지 않습니다.

테마 비율의 구간은 「테마 이름은 무작위 10건으로만 확인함(LLM 판단)」이라는 조건에서 구한 값입니다.

</details>

민원 50만 건에서 주제가 바뀐 시기는 「여행」 민원의 비율이 크게 줄어든 2월에서 4월 사이입니다. 분모 · 달력 · 버전을 확인한 뒤에도 이 감소는 남습니다. 「마스크」 · 「코로나」는 원문을 확인할 달이 있어 그 달 원문을 읽은 뒤에 결론에 더합니다.

- 비율이 바뀐 달마다 분모 · 달력 · 원문 · 버전을 차례로 확인합니다
- 「변했다」고 하려면 무엇의 비율인지 정하고 분모 · 버전을 맞춰 두 달 차이의 95% 구간을 구합니다

판매처 A 처럼 한 판매처를 뺀 증가가 있을 수 있으니, 원문을 확인할 달은 뺀 비율까지 구한 뒤에 결론을 씁니다.

### 팀 라운드: 「코로나 민원 재확산」이라는 제목 [필수]

아래 질문의 답은 수업에서 팀이 함께 정합니다. 지금은 자기 답과 근거만 적어 두면 됩니다.

각자 답과 확신도(1～5)를 적어 잠그면 팀의 답 분포가 공개됩니다. 근거 세 문장을 나눠 토론한 뒤 팀이 답 하나로 모아 제출합니다.

> **[문제] 팀 라운드: 「코로나 민원 재확산」이라는 제목**
>
> 분석팀이 달마다 하는 경영진 브리핑의 8월 제목을 정합니다. 「코로나」가 있는 민원의 비율이 7월보다 높아졌고, 확인할 4가지 가운데 분모 · 달력 · 버전은 마쳤지만 원문은 아직 읽지 않았습니다. 기획팀 동료는 「코로나 민원이 세 배 가까이 늘었으니 제목을 "코로나 민원 재확산"으로 하자」고 합니다.
> 
> - **문서 비율**(민원 50만 건 전체): 7월 716 ÷ 55,260 = 1.30% → 8월 2,172 ÷ 57,796 = 3.76%
> - **8월 품목**: 가장 많은 품목은 예식서비스 810건(8월 「코로나」 민원의 37.3%)
> - **8월 접수일**: 2,172건 가운데 1,968건(90.6%)이 8-15～8-31 에 접수, 몰린 이유는 파일에 없음
> - **테마 열**(표본 2만 건): 「예식 취소·위약금」 7월 30 ÷ 2,297 = 1.31% → 8월 92 ÷ 2,314 = 3.98%
> 
> 팀 안에서 의견이 3가지로 나뉘었습니다.
>
