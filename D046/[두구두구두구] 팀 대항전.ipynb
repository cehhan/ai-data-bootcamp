## 0. 팀 대항전 진행 방법

문항은 여섯 개이고, 초급·중급·고급이 두 개씩입니다. 문항마다 상황과 자료를 읽고 보기 다섯 가운데 하나를 고릅니다.

1. 팀원이 각자 진영(A～E), 고른 까닭, 확신도(1～5)를 적어 냅니다. 내 답을 내면 팀의 답 분포가 바로 보입니다.
2. 두 명 이상이 같은 보기를 고르면 「변호인 질문」이 뜹니다. 팀이 질문에 답을 적고, 그 보기를 지킬지 바꿀지 정합니다.
3. 출석한 팀원이 모두 낸 뒤, 한 명이 팀 답 칸에 고른 보기와 까닭을 두세 문장으로 적어 냅니다. 팀 답은 한 번만 낼 수 있습니다.
4. 팀 답을 내면 그 팀 화면에 정답과 해설이 열립니다. 다른 팀이 모두 낼 때까지 정답을 말하지 않습니다.

| 난이도 | 맞히면 | 설명이 핵심을 짚으면 |
|---|---|---|
| 초급 | 10점 | +5점 |
| 중급 | 20점 | +5점 |
| 고급 | 30점 | +5점 |

오답이어도 설명이 핵심을 짚으면 3점을 받습니다. 핵심은 해설의 채점 키워드로 확인합니다. 1～5일차 이론과 레퍼런스, 복습 프로젝트의 출력은 풀면서 봐도 됩니다.
