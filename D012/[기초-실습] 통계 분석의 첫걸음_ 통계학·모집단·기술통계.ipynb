# 문제 1. 기초통계량 계산

**📘 문제**

- 온라인 쇼핑몰 고객 1,000명을 대상으로 1점부터 10점 사이의 만족도 조사를 실시했습니다.
  이 데이터를 바탕으로 고객 만족도의 **중심 경향성**과 **분포 특성**을 파악해 봅시다.

```python
# 데이터 생성
np.random.seed(2025)
satisfaction = np.random.normal(7.5, 1.5, 1000)
satisfaction = np.clip(satisfaction, 1, 10)
df1 = pd.DataFrame({'score': satisfaction})
```

```python
# 아래를 수행해 보세요:
# [문제 1] 여기에 코드를 작성해주세요.

#평균, 중앙값, 최빈값, 표준편차, 값 확인용
s= df1["score"]

print(f"평균: {s.mean():.4f}")
print(f"중앙값: {s.median():.4f}")
print(f"최빈값: {s.mode().iloc[0]:.4f}")
print(f"표준편차: {s.std():.4f}")

print(df1.describe())

#히스토그램 (평균, 중앙값 표시)

mean_val = s.mean().round(3)
median_val = s.median().round(3)

plt.figure(figsize=(8,5))
plt.hist(s, bins=30, color='skyblue', edgecolor='black', alpha=0.7)

plt.axvline(mean_val, color = 'red', linestyle='-', linewidth=1, label=f'평균: {mean_val:.3f}')
plt.axvline(median_val, color = 'yellow', linestyle='--', linewidth=1, label=f'중앙값: {median_val:.3f}')
plt.legend()
print()
print("고객만족사항 데이터는 6점에서 9점 사이에 집중 분포하고 있으며 평균과 중앙값의 차이는 0.007정도로 미미함,") 
print("낮은 점수는 거의 발생하지 않을것으로 봐서 불만족 고객들은 리뷰를 남기지 않고 이탈하는 경향으로 추정,")
print("소수이지만 낮은 점수를 준 6점 미만의 고객들의 VOC 체크, 최우선 조치하여 빠르게 점수 개선 실행")
```

# 문제 2. 평균 vs 중앙값 비교

**📘 문제**

- A지역과 B지역의 연봉 데이터를 비교해 봅시다.
- B 지역에는 일부 고소득자가 포함되어 있습니다.

```python
# 데이터 생성
np.random.seed(2025)

# A 지역: 평균 300, 표준편차 50
region_a = np.random.normal(300, 50, 100)

# B 지역: 일반 소득자 + 극단적인 고소득자 포함
region_b = np.concatenate([
    np.random.normal(280, 40, 95),    # 일반 소득자 95명
    np.random.normal(1500, 200, 5)    # 고소득자 5명
])

df2 = pd.DataFrame({
    'region': ['A'] * 100 + ['B'] * 100,
    'salary': np.concatenate([region_a, region_b])
})
```

**📌 아래를 수행해 보세요:**

- 각 지역의 평균 연봉과 중앙값을 구해 봅시다.
- 지역별 박스플롯(Boxplot)을 그려서 분포의 차이를 시각적으로 확인해 봅시다.

```python
# 아래를 수행해 보세요:
# [문제 2] 여기에 코드를 작성해주세요.
print(df2.head())
print()
# 지역별 연붕 총합, 평균, 중앙값
print("지역별 연봉 총합/평균/중앙값")
salary_region = df2.groupby("region")["salary"].agg(["sum", "mean", "median"]).round(2)
print(salary_region)

# boxplot 그리기

plt.figure(figsize=(8, 6))
sns.boxplot(data=df2, x='region', y='salary', hue='region', palette='Set2', legend=False)

plt.title("지역별 소득 분포")
plt.xlabel("Region")
plt.ylabel("Salary")
plt.show()
```

# 문제 3. 이상값과 분산 비교

**📘 문제**

- 1반과 2반의 시험 점수 분포를 비교해 봅시다.
- 2반에는 극단적인 이상값(예: 부정행위로 0점)이 포함되어 있습니다.

```python
# 데이터 생성
np.random.seed(2025)  # 재현 가능성을 위해 시드 고정

# 1반: 평균 75, 표준편차 8
class1 = np.random.normal(75, 8, 30)

# 2반: 일반 학생 + 이상값(0점, 130점)
class2 = np.concatenate([
    np.random.normal(75, 15, 28),  # 일반 학생
    [0, 130]                       # 극단적인 이상값
])

df3 = pd.DataFrame({
    'class': ['1반'] * 30 + ['2반'] * 30,
    'score': np.concatenate([class1, class2])
})
```

**📌 아래를 수행해 보세요:**

1. 각 반의 평균, 표준편차, 최소/최대값을 계산해 봅시다.
2. 박스플롯을 사용하여 두 반의 분포 차이를 시각적으로 비교해 봅시다.
3. 이상값이 통계 지표에 어떤 영향을 주는지 생각해 봅시다.

```python
# 아래를 수행해 보세요:
# [문제 3] 여기에 코드를 작성해주세요.
print("반별 성적 종합")
df_group = df3.groupby("class")["score"].agg(["mean","median", "std", "min", "max"]).round(1)
print(df_group)

plt.figure(figsize=(8,6))
sns.boxplot(data=df3, x='class', y='score', hue='class', palette='Set2', legend=False)
plt.title("반별 성적 분포")
plt.xlabel("반")
plt.ylabel("성적")
plt.show()
```
